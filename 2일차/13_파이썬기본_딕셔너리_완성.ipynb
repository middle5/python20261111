# 딕셔너리

## 딕셔너리(dict)

▪️ 딕셔너리는 key:value 쌍을 이루어 하나의 요소로 저장됩니다.

```
딕셔너리명 = { key1:value1, key2:value2, key3:value3, … }
```

- `key1:value1` 한 쌍이 하나의 요소
- key : 자료를 구분
- value : 자료의 내용

딕셔너리에는 인덱스가 없으며, 자료에 접근할 때는 key로 접근합니다.

## 딕셔너리(dict)

```python
menu = {  '김밥':2000 , '떡볶이':2500 , '어묵':2000 , '튀김':3000  }
```


## 딕셔너리(dict)

◾ 딕셔너리 생성하기

````python
menu = {
    "김밥":4000,
    "떡볶이":3000,
    "어묵":2000,
    "튀김":3500
}
print(menu)
````

## 딕셔너리(dict)

◾ 딕셔너리의 특징 ①

딕셔너리의 키(key)가 중복되면 하나만 반영됩니다.

````python
menu = {
    "김밥":4000,
    "떡볶이":3000,
    "어묵":2000,
    "튀김":3500,
    "김밥":3500
}
print(menu)
````

## 딕셔너리(dict)

◾ 딕셔너리의 특징 ②

딕셔너리는 키(key)를 이용하여 값(value)에 접근합니다.

````python
menu = {
    "김밥":4000,
    "떡볶이":3000,
    "어묵":2000,
    "튀김":3500
}
print("김밥:", menu["김밥"],'원', sep='')
````

## 딕셔너리(dict)

◾ 키와 값의 특징

| | 키(key) | 값(value) |
|---|---|---|
| 가능한 자료형 | 숫자, 문자열, 변수, 튜플 | 모든 자료형, 변수 |
| 변경 가능 여부 | 변경 불가 | 변경 가능 |
| 중복 가능 여부 | 중복 불가 | 중복 가능 |

## 딕셔너리(dict)

◾ 딕셔너리의 값 수정/추가하기

key가 존재하면 수정, 존재하지 않으면 추가됩니다.

````python
menu = {
    "김밥":4000,
    "떡볶이":3000,
    "어묵":2000,
    "튀김":3500
}
menu['김밥'] = 4500    # 수정
menu['라면'] = 3500    # 추가
print(menu)
````

## 딕셔너리(dict)

◾ 딕셔너리의 요소 삭제하기

key를 이용하여 삭제합니다.

````python
menu = {
    "김밥":4000,
    "떡볶이":3000,
    "어묵":2000,
    "튀김":3500
}
del menu["튀김"]
print(menu)
````

## 딕셔너리(dict)

◾ 딕셔너리의 멤버연산자 : in, not in

딕셔너리의 key 존재 여부를 확인합니다.

````python
menu = {
    "김밥":4000,
    "떡볶이":3000,
    "어묵":2000,
    "튀김":3500
}
print("김밥" in menu)
print("김밥" not in menu)
````

## 딕셔너리(dict)

◾ 딕셔너리의 key, value, (key, value) 가져오기

````python
menu = {
    "김밥":4000,
    "떡볶이":3000,
    "어묵":2000,
    "튀김":3500
}
print(menu.keys())
print(menu.values())
print(menu.items())
````

## [미션] 여러 품목 한 번에 추가하기

포스트잇(2,500원)과 클립(1,500원)을 한 번에 추가하고, 전체 품목 수를 출력하세요.

````python
price = {'A4 용지': 5000, '볼펜': 800, '스테이플러': 7000, '형광펜': 1200}
price.update({'포스트잇': 2500, '클립': 1500})
print(price)
print('품목 수:', len(price))
````

## [미션] 이미 있는 품목은 추가하지 않기

새 품목을 등록할 때, 이미 가격표에 있는 품목이면 값을 덮어쓰지 않고 안내 문구를 출력하세요.

````python
price = {'A4 용지': 5000, '볼펜': 900, '스테이플러': 7000, '형광펜': 1200, '포스트잇': 2500, '클립': 1500, '가위': 3000}
new_item = input("추가할 품목 : ")
new_price = int(input("금액"))

if new_item not in price:
  price[new_item] = new_price
  print(f'{new_item}({new_price}원) 추가했습니다.')
else:
  print(f'{new_item}은(는) 이미 등록된 품목입니다.')
  print(f'{new_item}:{new_price}')

print(f'가격표:{price}')
````

실행 결과 예시

```
볼펜은(는) 이미 등록된 품목입니다.
볼펜:1000
가격표:{'A4 용지': 5000, '볼펜': 900, '스테이플러': 7000, '형광펜': 1200, '포스트잇': 2500, '클립': 1500, '가위': 3000}
```

```
테이프(2000원) 추가했습니다.
가격표:{'A4 용지': 5000, '볼펜': 900, '스테이플러': 7000, '형광펜': 1200, '포스트잇': 2500, '클립': 1500, '가위': 3000, '테이프': 2000}
```

## [미션] 빈 딕셔너리에 하나씩 채우기

빈 주문서를 만들고, A4 용지 3박스와 볼펜 10자루를 차례로 추가한 뒤 출력하세요. 이어서 가격표를 써서 주문 금액을 계산하세요.

````python
price = {'A4 용지': 5000, '볼펜': 900, '스테이플러': 7000, '형광펜': 1200, '포스트잇': 2500, '클립': 1500, '가위': 3000}

order = {}
order['A4 용지'] = 3
order['볼펜'] = 10
print('주문서:', order)

total = price['A4 용지'] * order['A4 용지'] + price['볼펜'] * order['볼펜']
print('주문 금액:', total)
````
