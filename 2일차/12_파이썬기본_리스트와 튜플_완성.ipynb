# 리스트

## 리스트(list)

- 대괄호`[ ]` 안에 여러 요소들을 콤마(`,`)로 구분하여 묶어서 저장
- 요소들은 순서를 가지고 있으며 이 순서를 인덱스라고 한다.
- 인덱스를 이용하여 요소를 선택하거나 변경, 추가, 삭제한다.
- 인덱스는 0부터 시작한다. 음수 인덱스를 사용할 수 있다.

**리스트 개념도**

| | 요소1 | 요소2 | 요소3 | ∙∙∙ | 요소n |
|---|---|---|---|---|---|
| 인덱스 | 0 | 1 | 2 | ∙∙∙ | n-1 |
| 음수 인덱스 | -n | -n+1 | -n+2 | ∙∙∙ | -1 |

## 리스트(list)

◾ 리스트 인덱싱 : 리스트에서 하나의 값 가져오기

````python
todo_list = ['주간업무보고서 작성','거래처 미팅','견적서 발송','회의록 정리','월말 정산']
print('할 일:', todo_list)
print('첫번째로 할 일 :', todo_list[0])
print('세번째로 할 일 :', todo_list[2])
print('마지막으로 할 일 :', todo_list[-1])
````

## 리스트(list)

◾ 리스트 슬라이싱 : 리스트에서 값 잘라오기

````python
todo_list = ['거래처 미팅','회의록 작성','견적서 발송','월말 정산','주간업무보고서 작성']
print('두번째~네번째 할 일 :', todo_list[1:4])
print('처음 할 일 2가지 :', todo_list[:2])
print('3번째 이후로 할 일 :', todo_list[2:])
print('할 일 모두:', todo_list[:])
````

## 리스트(list)

◾ 리스트 끝에 요소 추가하기

```
리스트.append(추가할 요소)
```

````python
todo_list = []
todo_list.append('거래처 미팅')
todo_list.append('회의록 정리')
todo_list.append('주간업무보고서 작성')
todo_list
````

## 리스트(list)

◾ 리스트 중간에 요소 추가하기

```
리스트.insert(추가할 위치, 추가할 요소)
```

````python
todo_list = ['거래처 미팅', '회의록 정리', '주간업무보고서 작성']
todo_list.insert (1, '견적서 발송')
todo_list.insert (3, '월말 정산')
todo_list
````

## 리스트(list)

◾ 요소 수정하기

```
리스트[수정할 요소의 인덱스] = 값
```

````python
todo_list = ['거래처 미팅', '견적서 발송', '회의록 정리', '월말 정산', '주간업무보고서 작성']
todo_list[3] = '연말정산'
todo_list[4] = '월말보고서 작성'
todo_list
````

## 리스트(list)

◾ 특정 인덱스의 요소 삭제하기

```
del 리스트[인덱스]
```

````python
todo_list = ['거래처 미팅', '견적서 발송', '회의록 정리', '월말 정산', '주간업무보고서 작성']
del todo_list[1]
todo_list
````

## 리스트(list)

◾ 마지막 요소 리턴하고 삭제하기

```
pop 리스트
```

````python
todo_list = ['거래처 미팅', '견적서 발송', '회의록 정리', '월말 정산', '주간업무보고서 작성']
last_todo = todo_list.pop()
print(last_todo)
print(todo_list)
````

## 리스트(list)

◾ 리스트 정렬하기

```
리스트.sort()
```

````python
todo_list = ['거래처 미팅', '견적서 발송', '회의록 정리', '월말 정산', '주간업무보고서 작성']
result = todo_list.sort()
print(todo_list)
print(result)
````

## 리스트(list)

◾ 리스트 정렬하기

```
sorted(리스트)
```

````python
todo_list = ['거래처 미팅', '견적서 발송', '회의록 정리', '월말 정산', '주간업무보고서 작성']
result = sorted(todo_list)
print(todo_list)
print(result)
````

## 함수와 메소드

파이썬의 모든 데이터는 객체입니다.

| 함수 | 메소드 |
|---|---|
| `sorted(리스트)` | `리스트.sort()` |

메소드 : 객체가 가지는 함수

```python
fruit_list = ['banana','apple', 'kiwi']
fruit_list.sort()
```

`fruit_list`는 객체(object), `sort()`는 메소드입니다.

# 튜플

## 튜플(tuple)

▪️ 튜플은 아래의 특징을 제외하곤 리스트와 거의 동일합니다.

- 리스트는 `[ ]`로 둘러싸지만 튜플은 `( )`로 둘러싼다.
- 리스트는 값의 생성, 삭제, 수정이 가능하지만 튜플은 값을 바꿀 수 없다.
- 튜플은 요소가 1개일 때에도 콤마(`,`)를 붙여주어야 한다.
- 튜플은 괄호를 생략할 수 있다.
- 리스트와 동일한 방법으로 인덱싱/슬라이싱 한다.

※ 튜플은 함수에 값을 전달하거나, 값을 리턴 받을 때 자주 볼 수 있어요!

## 튜플(tuple)

◾ 튜플 생성하기

````python
t1 = ()      # 빈 튜플
t2 = (1,)    # 요소가 1개인 튜플
t3 = 1,      # 요소가 1개인 튜플
t4 = (1,2,3)
t5 = 1,2,3
t6 = (1)     # 튜플이 아님
````
