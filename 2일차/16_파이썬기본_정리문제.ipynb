### [쉬운 미션] 주문 확인서 출력하기

가격표와 장바구니가 있습니다. 아래 순서대로 주문 확인서를 출력하세요.

1. 장바구니의 품목마다 "품목 수량개 x 단가원 = 금액원"을 출력하세요.
2. 상품 금액 합계를 구하세요.
3. 상품 금액 합계가 50,000원 이상이면 배송비는 0원, 그렇지 않으면 3,000원입니다.
4. 상품 금액, 배송비, 결제 금액(상품 금액 + 배송비)을 출력하세요.

힌트: 장바구니 딕셔너리를 `items()`로 순회하고, 단가는 가격표에서 품목 이름(키)으로 꺼냅니다.

````python
price = {'A4 용지': 5000, '볼펜': 800, '스테이플러': 7000, '형광펜': 1200}
cart = {'A4 용지': 4, '볼펜': 10, '형광펜': 5}
total_price = 0
delivery = 3000

# 여기에 코드를 작성하세요
for item, qty in cart.items():
    print('=== 주문 확인서 ===')
    print(f"{item} {qty}개 X {price[item]:,}원 = {price[item] * cart[item]:,}원")
    total_price += price[item] * cart[item]

if total_price >= 50000:
    delivery = 0

print('-' * 30)
print(f"상품 금액 : {total_price:,}원")
print(f"배송비 : {delivery:,}원")
print(f"결제 금액 : {total_price + delivery:,}원")
````

실행 결과 예시

```
=== 주문 확인서 ===
A4 용지 4개 x 5000원 = 20000원
볼펜 10개 x 800원 = 8000원
형광펜 5개 x 1200원 = 6000원
--------------------
상품 금액: 34000원
배송비: 3000원
결제 금액: 37000원
```

장바구니에 스테이플러를 3개 넣으면 배송비가 어떻게 바뀌는지도 확인해 보세요.
