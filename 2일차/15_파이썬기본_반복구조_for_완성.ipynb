## for 반복구조

iterable(반복 가능한) 객체를 순회하며 각 요소에 대해 내부의 코드를 실행.

```
for 제어변수 in iterable객체:
        코드블록
```

- iterable객체 : list, tuple, dict, set, str, range, file object
- 요소들을 순서대로 하나씩 제어변수에 대입하면서 각 요소에 대해 내부의 코드를 실행

## for 반복구조

■ range() 함수

연속된 정수 시퀀스를 만들어내는 함수

사용법

```
range(stop)
range(start, stop)
range(start, stop, step)
```

- start: 시작값(포함). 생략하면 기본값은 0입니다.
- stop: 종료값(미포함). 반드시 지정해야 합니다.
- step: 증가(또는 감소) 폭. 기본값은 1이며, 양수 혹은 음수로 지정할 수 있습니다.

## [연습문제] 상반기 매출 합계와 월평균

sum()을 쓰지 않고, 반복문으로 상반기 매출 합계와 월평균을 구하세요. (단위: 만 원)

````python
sales = [1250, 980, 1410, 1120, 1360, 1540]
total = 0

for s in sales:
    total += s

print('합계:', total)
print('월평균:', total / len(sales))
````

## [연습문제] 주문 내역과 금액 계산하기

가격표와 주문서를 써서 품목별 금액과 전체 주문 금액을 출력하세요.

````python
price = {'A4 용지': 5000, '볼펜': 800, '스테이플러': 7000}
order = {'A4 용지': 3, '볼펜': 10, '스테이플러': 1}
total = 0

for item, qty in order.items():
    cost = price[item] * qty
    print(f'{item} {qty}개 : {cost}원')
    total += cost
print('주문 금액:', total)
````

## [연습문제] 반려 문서가 나오면 멈추기

결재 결과 목록을 앞에서부터 확인하다가, '반려'가 처음 나오면 몇 번째 문서인지 출력하고 확인을 멈추세요.

````python
docs = ['승인', '승인', '반려', '승인', '반려']

for i in range(len(docs)):
    if docs[i] == '반려':
        print(f'{i + 1}번째 문서가 반려되었습니다. 확인을 멈춥니다.')
        break
    print(f'{i + 1}번째 문서: 승인')
````

## for 반복구조

◾ 반복을 건너뛰는 continue

continue를 만나면 이번 차례는 건너뛰고 다음 차례로 넘어갑니다.

````python
total = 0
for i in range(1,101):
  if i%2==0:
    continue
  total = total + i
print('1부터 101까지 홀수만 더한 결과 --> ', total)
````

## [연습문제] 반려 문서만 골라 출력하기

결재 결과 목록에서 승인된 문서는 건너뛰고, 반려된 문서만 모두 출력하세요.

````python
docs = ['승인', '승인', '반려', '승인', '반려']

for i in range(len(docs)):
    if docs[i] == '승인':
        continue
    print(f'{i + 1}번째 문서가 반려되었습니다.')
````

## [연습문제] 휴무일 빼고 평균 근무 시간 구하기

일주일 동안의 근무 시간 기록입니다. 0은 휴무일입니다.
휴무일을 빼고 근무일 수, 총 근무 시간, 하루 평균 근무 시간을 출력하세요.

````python
hours = [8, 9, 0, 8, 10, 0, 0]
total = 0
work_days = 0

for h in hours:
    if h == 0:
        continue
    total += h
    work_days += 1

print('근무일 수:', work_days)
print('총 근무 시간:', total)
print('하루 평균:', total / work_days)
````

## [연습문제] 미응답 빼고 평균 만족도 구하기

고객 만족도 설문 결과입니다.
빈 문자열('')은 응답하지 않은 고객입니다. 응답한 고객 수와 평균 점수를 출력하세요.

````python
answers = [5, 4, '', 3, 5, '', 4]
total = 0
count = 0

for a in answers:
    if a == '':
        continue
    total += a
    count += 1

print('응답 수:', count)
print('평균 점수:', total / count)
````

## [연습문제] 성적 계산하기

학생들의 시험 점수 리스트에 결석한 학생의 점수는 -1로 기재되어 있습니다.
아래 3가지를 출력하세요.

- 출석한 학생 수
- 결석한 학생 수
- 출석한 학생 수의 평균 점수

````python
# 학생들의 시험 점수 리스트
scores = [85, 92, 78, 90, -1, 87, 95, -1, 88, 93]
print("=== 학생 성적 분석 ===")
total = 0
present_count = 0
absent_count = 0
for i in scores:
    if i == -1:
        absent_count += 1
        continue
    total += i
    present_count += 1
mean = total / present_count

print(f'''
출석한 학생 수 : {present_count}
결석한 학생 수 : {absent_count}
평균 점수 : {mean}
''')
````
