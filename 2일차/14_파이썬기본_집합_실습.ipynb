# 집합

## 집합(set)

▪️ 집합(set)은, 자료들이 순서와 중복 없이 저장되는 자료형입니다.

```
집합명 = { 값1, 값2, 값3 }
```

````python
# 회의 참석 부서 확인하기
dept = {'영업팀','인사팀','마케팅팀','영업팀','회계팀'}
print(dept)
````

중복을 제거하고 한 개만 저장된 것을 볼 수 있어요!

집합에도 인덱스가 존재하지 않으며 순서가 없는 자료형이므로, 순서가 바뀌어 출력될 수 있습니다.

## 집합(set)

▪️ 집합 연산

| | 합집합 | 교집합 | 차집합 |
|---|---|---|---|
| 메소드 | `a.union(b)` | `a.intersection(b)` | `a.difference(b)` |
| 연산자 | `a\|b` | `a&b` | `a-b` |

## 집합(set)

▪️ 집합 연산

| sweet에만 | sweet와 salty 모두 | salty에만 |
|---|---|---|
| 빼빼로, 칸초 | 허니버터 | 포카칩, 새우깡 |

````python
# 달콤한 과자, 짭짤한 과자
sweet= {'빼빼로','칸초','허니버터'}
salty = {'포카칩','새우깡','허니버터'}
print(sweet)
print(salty)
````

## 집합(set)

▪️ 집합 연산 - 합집합

````python
# 모든 과자 집합
print("합집합:")
print("합집합:")
````

## 집합(set)

▪️ 집합 연산 - 교집합

````python
# 단짠을 찾아라
print("교집합:")
print("교집합:")
````

## 집합(set)

▪️ 집합 연산 - 차집합

````python
# 달콤하기만 한 과자
print("차집합(sweet에만 속하는 과자):")
print("차집합(sweet에만 속하는 과자):")
````

## 집합(set)

▪️ 집합 연산 - 차집합

````python
# 짭짤하기만 한 과자
print("차집합(salty에만 속하는 과자):")
print("차집합(salty에만 속하는 과자):")
````
