### [연습문제 5-3] 집합 연산

1. 두 달 모두 구매한 고객 (교집합)
2. 한 번이라도 구매한 고객 (합집합)

```python
jan = {'한빛상사', '다온물산', '새봄유통', '푸른테크'}
feb = {'다온물산', '푸른테크', '하늘물류'}

print('두 달 모두 구매:', jan & feb)
print('한 번이라도 구매:', jan | feb)
print('1월에만 구매:', jan - feb)
print('2월 신규 구매:', feb - jan)
```

### [연습문제 6-3] 합계 구하기

이번 주 주문 금액 리스트입니다. `sum()`을 쓰지 않고 for 반복구조로 주문 금액 합계와 주문 건수를 출력하세요.

```python
order_amounts = [120000, 45000, 380000, 76000, 510000]

total = 0
for amount in order_amounts:
    total += amount

print('주문 금액 합계:', total)
print('주문 건수:', len(order_amounts))
```

### [연습문제 6-4] 조건에 맞는 개수 세기

주문 금액이 100,000원 이상인 주문이 몇 건인지 출력하세요.

```python
order_amounts = [120000, 45000, 380000, 76000, 510000]

count = 0
for amount in order_amounts:
    if amount >= 100000:
        count += 1

print('100,000원 이상 주문:', count, '건')
```

### [연습문제 6-5] 조건에 맞는 값만 리스트에 담기

월 매출 목표는 1,200만 원입니다. 목표를 달성한 달의 이름만 빈 리스트에 담아 출력하세요. 월 이름과 매출은 같은 순서로 저장되어 있습니다.

```python
months = ('1월', '2월', '3월', '4월', '5월', '6월')
sales = [1250, 980, 1410, 1120, 1360, 1540]   # 단위: 만 원

good_months = []
for i in range(len(sales)):
    if sales[i] >= 1200:
        good_months.append(months[i])

print('목표 달성한 달:', good_months)
```

### [연습문제 6-6] 문자열 순회하고 continue로 건너뛰기

전화번호에서 '-'를 건너뛰고 숫자만 이어 붙여서 출력하세요. 빈 문자열 `digits = ''`에서 시작합니다.

```python
phone = '010-1234-5678'

digits = ''
for ch in phone:
    if ch == '-':
        continue
    digits += ch

print(digits)
```

### [연습문제 6-9] break로 멈추기

회의실 상태를 앞에서부터 확인하다가, 처음으로 '비어있음'인 회의실을 찾으면 "○○을 예약합니다."를 출력하고 확인을 멈추세요. 사용 중인 회의실은 "○○: 사용 중"을 출력합니다.

```python
rooms = {'A룸': '사용 중', 'B룸': '사용 중', 'C룸': '비어있음', 'D룸': '비어있음'}

for room, status in rooms.items():
    if status == '비어있음':
        print(f'{room}을 예약합니다.')
        break
    print(f'{room}: 사용 중')
```

가격표와 장바구니가 있습니다. 아래 순서대로 주문 확인서를 출력하세요.

1. 장바구니의 품목마다 "품목 수량개 x 단가원 = 금액원"을 출력하세요.
2. 상품 금액 합계를 구하세요.
3. 상품 금액 합계가 50,000원 이상이면 배송비는 0원, 그렇지 않으면 3,000원입니다.
4. 상품 금액, 배송비, 결제 금액(상품 금액 + 배송비)을 출력하세요.

힌트: 장바구니 딕셔너리를 `items()`로 순회하고, 단가는 가격표에서 품목 이름(키)으로 꺼냅니다.

```python
price = {'A4 용지': 5000, '볼펜': 800, '스테이플러': 7000, '형광펜': 1200}
cart = {'A4 용지': 4, '볼펜': 10, '형광펜': 5}

print('=== 주문 확인서 ===')
total = 0
for item, qty in cart.items():
    cost = price[item] * qty
    print(f'{item} {qty}개 x {price[item]}원 = {cost}원')
    total += cost

if total >= 50000:
    shipping = 0
else:
    shipping = 3000

print('-' * 20)
print(f'상품 금액: {total}원')
print(f'배송비: {shipping}원')
print(f'결제 금액: {total + shipping}원')
```

### [어려운 미션] 거래처별 월간 주문 분석하기

이번 달 주문 기록입니다. 주문 한 건은 (거래처, 품목, 수량) 튜플이고, 주문 기록은 이 튜플의 리스트입니다. 아래 다섯 가지를 차례로 구하세요.

1. 주문 건마다 "거래처 | 품목 수량개 | 금액원"을 출력하세요. 수량이 0인 주문은 취소된 주문이므로 `continue`로 건너뜁니다.
2. 거래처별 주문 금액 합계를 딕셔너리 `customer_total`에 모으고 출력하세요.
3. 이번 달 전체 매출을 출력하세요.
4. 주문 금액 합계가 100,000원 이상인 거래처를 "우수 거래처"로 출력하세요.
5. 전체 거래처 집합 `all_customers` 가운데 이번 달 주문이 없는 거래처를 출력하세요. 취소된 주문만 있는 거래처도 주문이 없는 거래처입니다.

힌트

- `for customer, item, qty in orders:`처럼 튜플을 변수 세 개로 나눠 받을 수 있습니다. 15 반복구조 for의 `for item, qty in order.items():`와 같은 방식입니다.
- 2번은 거래처가 `customer_total`에 이미 있으면 금액을 더하고, 없으면 새 키로 추가합니다. `in`으로 확인합니다.
- 5번은 실제로 주문한 거래처를 집합에 모은 뒤 차집합을 씁니다.

```python
price = {'A4 용지': 5000, '볼펜': 800, '스테이플러': 7000, '형광펜': 1200}

orders = [
    ('한빛상사', 'A4 용지', 10),
    ('다온물산', '볼펜', 30),
    ('한빛상사', '스테이플러', 4),
    ('새봄유통', 'A4 용지', 5),
    ('다온물산', 'A4 용지', 20),
    ('푸른테크', '형광펜', 0),
    ('새봄유통', '형광펜', 10),
]

all_customers = {'한빛상사', '다온물산', '새봄유통', '푸른테크', '하늘물류'}

# 1. 주문 건별 금액 출력, 2. 거래처별 합계 모으기
print('=== 주문 내역 ===')
customer_total = {}
ordered_customers = set()

for customer, item, qty in orders:
    if qty == 0:
        continue
    cost = price[item] * qty
    print(f'{customer} | {item} {qty}개 | {cost}원')

    if customer in customer_total:
        customer_total[customer] += cost
    else:
        customer_total[customer] = cost
    ordered_customers.add(customer)

print()
print('=== 거래처별 주문 금액 ===')
for customer, amount in customer_total.items():
    print(f'{customer}: {amount}원')

# 3. 전체 매출
total_sales = 0
for amount in customer_total.values():
    total_sales += amount
print()
print(f'이번 달 전체 매출: {total_sales}원')

# 4. 우수 거래처
print()
print('=== 우수 거래처 (100,000원 이상) ===')
for customer, amount in customer_total.items():
    if amount >= 100000:
        print(customer)

# 5. 주문이 없는 거래처
print()
print('이번 달 주문이 없는 거래처:', all_customers - ordered_customers)
```

5번 결과는 집합이므로 출력 순서가 다를 수 있습니다.
