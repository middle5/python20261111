# 반복구조

## 반복구조

| for 반복구조 | while 반복구조 |
|---|---|
| iterable(반복 가능한) 객체를 순회하며 각 요소에 대해 내부의 코드를 실행. | 조건이 True인 동안 반복 |
| 반복 가능한 객체 : list, tuple, str, dict, set, range | 반복 횟수는 조건에 따라 달라진다. |
| 반복이 시작되기 전에 반복 횟수를 알 수 있다.<br>(반복횟수=시퀀스의 크기) | 조건은 반복이 시작되기 전에도 검사하고, 반복이 한번 끝날 때마다 검사한다. |
| 무한루프에 빠지지 않는다. | 조건이 False가 되면 반복을 빠져나온다. |
| | 무한루프를 만들 수 있다. |

## for 반복구조

iterable(반복 가능한) 객체를 순회하며 각 요소에 대해 내부의 코드를 실행.

```
for 제어변수 in iterable객체:
        코드블록
```

- iterable객체 : list, tuple, dict, set, str, range, file object
- 요소들을 순서대로 하나씩 제어변수에 대입하면서 각 요소에 대해 내부의 코드를 실행

## for 반복구조

■ range() 함수

연속된 정수 시퀀스를 만들어내는 함수

사용법

```
range(stop)
range(start, stop)
range(start, stop, step)
```

- start: 시작값(포함). 생략하면 기본값은 0입니다.
- stop: 종료값(미포함). 반드시 지정해야 합니다.
- step: 증가(또는 감소) 폭. 기본값은 1이며, 양수 혹은 음수로 지정할 수 있습니다.

## [연습문제] 상반기 매출 합계와 월평균

## [연습문제] 주문 내역과 금액 계산하기

## [연습문제] 반려 문서가 나오면 멈추기

결재 결과 목록을 앞에서부터 확인하다가, '반려'가 처음 나오면 몇 번째 문서인지 출력하고 확인을 멈추세요.

````python
docs = ['승인', '승인', '반려', '승인', '반려']

for i in range(len(docs)):
    if docs[i] == '반려':
        print(f"{i+1}번째 문서가 반려되었습니다. 확인을 멈춥니다.")
        break
    print(f"{i+1}번째 문서 : {docs[i]}")
````

## for 반복구조

◾ 반복을 건너뛰는 continue

continue를 만나면 이번 차례는 건너뛰고 다음 차례로 넘어갑니다.

````python
total = 0

for i in range(1,102):
    if i % 2 == 0:
        continue
    total += i
print(f"1부터 101까지 홀수만 더한 결과 --> {total}")
````

## [연습문제] 휴무일 빼고 평균 근무 시간 구하기

일주일 동안의 근무 시간 기록입니다. 0은 휴무일입니다.
휴무일을 빼고 근무일 수, 총 근무 시간, 하루 평균 근무 시간을 출력하세요.

````python
hours = [8, 9, 0, 8, 10, 0, 0]
total = 0
work_days = 0

for i in hours:
    if i > 0:
        total += i
        work_days += 1
avg = total / work_days

print(f"근무일 수 : {work_days}\n총 근무 시간 : {total}\n하루 평균 : {avg}")
````
