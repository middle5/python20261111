# 집합

## 집합(set)

▪️ 집합(set)은, 자료들이 순서와 중복 없이 저장되는 자료형입니다.

```
집합명 = { 값1, 값2, 값3 }
```

````python
# 회의 참석 부서 확인하기
dept = {'영업팀','인사팀','마케팅팀','영업팀','회계팀'}
print(dept)
````

중복을 제거하고 한 개만 저장된 것을 볼 수 있어요!

집합에도 인덱스가 존재하지 않으며 순서가 없는 자료형이므로, 순서가 바뀌어 출력될 수 있습니다.

## 집합(set)

▪️ 집합 연산

| | 합집합 | 교집합 | 차집합 |
|---|---|---|---|
| 메소드 | `a.union(b)` | `a.intersection(b)` | `a.difference(b)` |
| 연산자 | `a\|b` | `a&b` | `a-b` |

## 집합(set)

▪️ 집합 연산

| sweet에만 | sweet와 salty 모두 | salty에만 |
|---|---|---|
| 빼빼로, 칸초 | 허니버터 | 포카칩, 새우깡 |

````python
# 달콤한 과자, 짭짤한 과자
sweet= {'빼빼로','칸초','허니버터'}
salty = {'포카칩','새우깡','허니버터'}
print(sweet)
print(salty)
````

## 집합(set)

▪️ 집합 연산 - 합집합

````python
# 모든 과자 집합
print("합집합:", sweet|salty)
print("합집합:", sweet.union(salty))
````

## 집합(set)

▪️ 집합 연산 - 교집합

````python
# 단짠을 찾아라
print("교집합:", sweet&salty)
print("교집합:", sweet.intersection(salty))
````

## 집합(set)

▪️ 집합 연산 - 차집합

````python
# 달콤하기만 한 과자
print("차집합(sweet에만 속하는 과자):", sweet-salty)
print("차집합(sweet에만 속하는 과자):", sweet.difference(salty))
````

## 집합(set)

▪️ 집합 연산 - 차집합

````python
# 짭짤하기만 한 과자
print("차집합(salty에만 속하는 과자):", salty-sweet)
print("차집합(salty에만 속하는 과자):", salty.difference(sweet))
````

## [연습문제] 신청자 명단 중복 제거하기

회의 참석 부서에 총무팀을 추가하고, 영업팀도 한 번 더 추가해 보세요. 부서 수가 어떻게 바뀌는지 출력하세요.

````python
names = ['김하늘', '이도윤', '박서준', '김하늘', '최지우', '이도윤']
unique_names = set(names)

print('신청 건수:', len(names))
print('실제 인원:', len(unique_names))
print('명단:', unique_names)
````

## [연습문제] 참석 부서 추가하기

회의 참석 부서에 총무팀을 추가하고, 영업팀도 한 번 더 추가해 보세요. 부서 수가 어떻게 바뀌는지 출력하세요.

````python
dept = {'영업팀', '인사팀', '마케팅팀', '회계팀'}

dept.add('총무팀')
dept.add('영업팀')
print(dept)
print('부서 수:', len(dept))
````

## [연습문제] 모두 신청한 사람

엑셀 교육과 파이썬 교육 신청자 명단이 있습니다. 두 교육을 모두 신청한 사람을 출력하세요.

````python
excel = {'김하늘', '이도윤', '박서준', '최지우'}
python = {'이도윤', '최지우', '정민호'}

print('모두 신청:', excel & python)
````

## [연습문제] 하나라도 신청한 사람

두 교육 중 하나라도 신청한 사람과 그 인원을 출력하세요.

````python
excel = {'김하늘', '이도윤', '박서준', '최지우'}
python = {'이도윤', '최지우', '정민호'}

print('모두 신청:', excel | python)
print('인원:', len(excel | python))
````

## [연습문제] 서류 미제출자 찾기

전체 직원 명단과 서류 제출자 명단이 있습니다. 아직 제출하지 않은 직원과 그 인원을 출력하세요.

````python
staff = {'김하늘', '이도윤', '박서준', '최지우', '정민호', '한서연'}
submitted = {'이도윤', '최지우', '한서연'}

not_submitted = staff - submitted
print('미제출자:', not_submitted)
print('미제출 인원:', len(not_submitted))
````
