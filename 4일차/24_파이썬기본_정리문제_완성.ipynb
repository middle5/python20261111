### [연습문제 1-1] def 함수를 lambda로 바꾸기

아래 셀의 `calc_vat` 함수와 같은 일을 하는 lambda 함수를 만들어 같은 이름 `calc_vat`에 저장하세요. `calc_vat(50000)`을 출력하세요.

### [연습문제 1-2] 매개변수가 두 개인 lambda

단가와 수량을 받아서 공급가액을 돌려주는 lambda 함수를 `calc_supply`에 저장하세요. 단가 800원, 수량 30개로 호출해서 출력하세요.

### [연습문제 1-3] lambda 안에서 if-else 쓰기

주문 금액을 받아서 배송비를 돌려주는 lambda 함수 `shipping_fee`를 만드세요. 50,000원 이상이면 0원, 아니면 3,000원입니다. 42,000원과 58,000원으로 호출해서 출력하세요.

힌트: 16 리스트컴프리헨션에서 배운 `참일때_값 if 조건식 else 거짓일때_값` 모양을 lambda의 돌려줄 값 자리에 씁니다.

### [연습문제 1-5] sorted()의 key에 lambda 넣기 (글자 수)

품목 이름을 글자 수가 적은 순서로 정렬해서 출력하세요.

힌트: `sorted(리스트, key=lambda 변수: 정렬 기준)`. 정렬 기준은 `len(이름)`입니다.

```python
items = ['스테이플러', '볼펜', '포스트잇', '클립']

print(sorted(items, key=lambda item: len(item)))
```

힌트: `key=lambda d: abs(d)`, 큰 순서는 `reverse=True`

```python
diffs = [-120, 45, -30, 260]   # 단위: 만 원

print(sorted(diffs, key=lambda d: abs(d), reverse=True))
```

### [연습문제 1-7] 튜플 리스트를 두 번째 값으로 정렬하기

(담당자, 매출) 튜플의 리스트입니다. `sort()`와 lambda로 매출이 높은 순서로 정렬해서 출력하세요.

힌트: 튜플 `s`에서 매출은 두 번째 값이므로 `s[1]`입니다.

```python
sales = [('김민수', 1200), ('이서연', 980), ('박지훈', 1530)]

sales.sort(key=lambda s: s[1], reverse=True)
print(sales)
```

`sort()`는 원본 리스트를 바꾸고 None을 돌려줍니다. 그래서 `sales = sales.sort(...)`처럼 쓰면 안 됩니다.

### [연습문제 1-8] 딕셔너리를 값 기준으로 정렬하기

가격표 딕셔너리를 가격이 싼 순서로 정렬해서 "품목 : 가격원" 모양으로 출력하세요.

힌트: `price.items()`는 (품목, 가격) 튜플을 꺼냅니다. 연습문제 1-7처럼 두 번째 값으로 정렬합니다.

```python
price = {'A4 용지': 5000, '볼펜': 800, '토너': 32000, '스테이플러': 7000}

for item, p in sorted(price.items(), key=lambda x: x[1]):
    print(f'{item} : {p}원')
```

### [연습문제 2-11] map()에 def 함수 넣기

연간 구매 금액을 받아서 등급을 돌려주는 함수 `get_grade`를 map()에 넣어서, 금액 리스트를 등급 리스트로 바꾸세요.

힌트: map()에 함수를 넣을 때는 이름만 씁니다. `map(get_grade, amounts)`처럼 괄호를 붙이지 않습니다.

```python
def get_grade(amount):
    if amount >= 1000000:
        return 'VIP'
    elif amount >= 500000:
        return 'GOLD'
    else:
        return 'BASIC'

amounts = [1200000, 650000, 90000, 2300000]

grades = list(map(get_grade, amounts))
print(grades)
```

### [연습문제 2-12] filter()에 lambda 넣기

주문 금액 리스트에서 500,000원 이상인 주문만 골라 리스트로 출력하세요. 골라낸 주문의 건수도 출력하세요.

### [연습문제 2-14] map()과 filter() 이어 쓰기

엑셀에서 복사한 일별 방문자 수가 문자열입니다. map()으로 숫자로 바꾼 뒤, filter()로 방문자가 100명 미만인 날의 방문자 수만 골라 출력하세요.

```python
raw_visits = ['132', '87', '154', '96', '210', '64']

visits = list(map(int, raw_visits))
low_days = list(filter(lambda v: v < 100, visits))
print(visits)
print(low_days)
```

### [연습문제 2-15] map(), filter()를 리스트 컴프리헨션으로 바꾸기

아래 셀의 두 줄과 같은 결과가 나오도록 리스트 컴프리헨션으로 바꿔 쓰세요.

```python
prices = [18000, 25000, 9000, 32000]

print([int(p * 1.1) for p in prices])       # 부가세 포함 가격
print([p for p in prices if p >= 20000])    # 2만 원 이상
```

map()은 `[표현식 for ...]`, filter()는 `[변수 for ... if 조건식]`으로 바꿔 쓸 수 있습니다.

프로젝트 시작일은 2026년 9월 1일, 종료일은 2026년 12월 31일입니다. 두 날짜를 빼서 프로젝트 기간이 며칠인지 출력하세요. 시작일과 종료일을 모두 포함하므로 1을 더합니다.

### [쉬운 미션] 주간 회의 일정표 만들기

11월 9일(월)부터 매주 월요일에 4번 회의를 합니다. 아래 순서대로 일정표를 만드세요.

1. `from datetime import date, timedelta`로 필요한 것만 가져오세요.
2. 리스트 컴프리헨션으로 회의 날짜 4개를 담은 리스트 `dates`를 만드세요. i번째 회의는 첫 회의 날짜 + `timedelta(days=7 * i)`입니다.
3. map()과 lambda로 날짜를 "11월 09일" 모양의 문자열 리스트 `labels`로 바꾸세요.
4. enumerate()로 "1회차 : 11월 09일" 모양으로 출력하세요.
5. 기준일(11월 6일)부터 마지막 회의까지 며칠 남았는지 출력하세요.

```python
# 1. 필요한 것만 가져오기
from datetime import date, timedelta

first = date(2026, 11, 9)
base = date(2026, 11, 6)   # 기준일

# 2. 회의 날짜 4개
dates = [first + timedelta(days=7 * i) for i in range(4)]

# 3. 날짜를 문자열로 바꾸기
labels = list(map(lambda d: d.strftime('%m월 %d일'), dates))

# 4. 일정표 출력
print('=== 주간 회의 일정 ===')
for i, label in enumerate(labels, 1):
    print(f'{i}회차 : {label}')

# 5. 마지막 회의까지 남은 일수
print(f'마지막 회의까지 {(dates[-1] - base).days}일 남았습니다.')
```

회의 횟수를 바꾸려면 `range(4)`의 숫자만 바꾸면 됩니다.

### [어려운 미션] 영업팀 실적 분석 모듈 만들기

영업팀 실적을 분석하는 모듈을 만들고, 그 모듈로 보고서를 출력하세요. 월 목표는 1,200만 원입니다.

**1단계: 모듈 파일 `sales_report.py` 만들기** (`%%writefile sales_report.py`)

- 목표 금액 상수 `TARGET = 1200` (단위: 만 원)
- `to_number(text)` : "1,380"처럼 쉼표가 들어간 문자열을 정수로 바꿔 돌려줍니다.
- `achieve_rate(sale, target=TARGET)` : 달성률(매출 ÷ 목표 × 100)을 `round()`로 소수점 첫째 자리까지 돌려줍니다.
- `grade(rate)` : 달성률 110 이상 'A', 100 이상 'B', 그 밖 'C'를 돌려줍니다.

**2단계: 모듈을 불러와 보고서 만들기**

1. `import sales_report as sr`, `import statistics`로 불러옵니다.
2. map()과 `sr.to_number`로 매출 리스트 `sales`를 만듭니다.
3. 리스트 컴프리헨션과 `sr.achieve_rate`로 달성률 리스트 `rates`를 만듭니다.
4. enumerate()와 zip()으로 "1. 김민수 | 1,380만 원 | 115.0% | A" 모양의 보고서를 출력합니다.
5. sorted()와 lambda로 매출 상위 3명을 "1위 박지훈 (1,530만 원)" 모양으로 출력합니다.
6. 요약을 출력합니다.
   - 팀 평균 매출과 중앙값 (statistics)
   - 모든 직원이 달성률 80% 이상인지 (all)
   - 달성률 110% 이상인 직원이 있는지 (any)
   - 목표 미달(달성률 100% 미만) 직원 이름 리스트 (filter와 lambda)

힌트

- 5번은 `sorted(zip(names, sales), key=lambda x: x[1], reverse=True)`로 정렬한 뒤 앞의 3개만 슬라이싱합니다.
- 6번 미달 직원은 `filter(lambda x: x[1] < 100, zip(names, rates))`로 (이름, 달성률) 쌍을 고른 뒤, 이름만 꺼냅니다.

```python
%%writefile sales_report.py
# 영업 실적 분석 모듈
TARGET = 1200   # 월 목표 (단위: 만 원)

# "1,380" → 1380
def to_number(text):
    return int(text.replace(',', ''))

# 달성률(%)
def achieve_rate(sale, target=TARGET):
    return round(sale / target * 100, 1)

# 달성률에 따른 등급
def grade(rate):
    if rate >= 110:
        return 'A'
    elif rate >= 100:
        return 'B'
    else:
        return 'C'
```

```python
import sales_report as sr
import statistics

names = ['김민수', '이서연', '박지훈', '최유진', '정다은']
raw_sales = ['1,380', '1,120', '1,530', '960', '1,250']   # 단위: 만 원

# 2. 매출 리스트
sales = list(map(sr.to_number, raw_sales))

# 3. 달성률 리스트
rates = [sr.achieve_rate(s) for s in sales]

# 4. 보고서
print('=== 영업팀 실적 보고서 ===')
for i, (name, sale, rate) in enumerate(zip(names, sales, rates), 1):
    print(f'{i}. {name} | {sale:,}만 원 | {rate}% | {sr.grade(rate)}')

# 5. 매출 상위 3명
print()
print('=== 매출 상위 3명 ===')
ranking = sorted(zip(names, sales), key=lambda x: x[1], reverse=True)
for rank, (name, sale) in enumerate(ranking[:3], 1):
    print(f'{rank}위 {name} ({sale:,}만 원)')

# 6. 요약
print()
print('=== 요약 ===')
print(f'평균 매출: {statistics.mean(sales):,}만 원')
print(f'중앙값: {statistics.median(sales):,}만 원')
print('모두 80% 이상:', all([r >= 80 for r in rates]))
print('110% 이상 있음:', any([r >= 110 for r in rates]))
under = list(filter(lambda x: x[1] < 100, zip(names, rates)))
print('목표 미달:', [name for name, rate in under])
```

목표가 바뀌면 sales_report.py의 `TARGET` 한 줄만 고치면 됩니다. `sr.achieve_rate(s, 1300)`처럼 목표를 직접 넘겨 계산해 볼 수도 있습니다.
