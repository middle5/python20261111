# 모듈

> 이 노트북 파일에는 모듈 파일을 만드는 셀(`%%writefile 파일이름`)이 있습니다. 셀을 실행하면 노트북 파일과 같은 폴더에 `company_utils.py`, `travel.py`, `office` 폴더, `보고서` 폴더가 만들어집니다.

## 모듈은 자주 쓰는 기능을 담아두는 '도구함'입니다.

함수·변수·클래스를 별도 .py 파일에 저장하고, 필요한 곳에서 불러와 사용하는 방식입니다.

```
report_utils.py        ← 모듈명 : report_utils (파일명이 모듈명이 됩니다)
├─ calc_growth()
├─ format_won()
└─ make_summary()
```

- ① **반복 제거** : 같은 계산 코드를 매번 복사하지 않습니다.
- ② **수정 용이** : 한 파일만 고치면 모든 업무에 반영됩니다.
- ③ **협업 향상** : 역할이 명확해 인수인계가 쉬워집니다.

비유로 이해하기 : 엑셀의 자주 쓰는 수식을 템플릿으로 저장하는 것과 비슷합니다.

## 모듈의 종류

| | 표준 모듈 | 외부 모듈 | 내가 만든 모듈 |

## datetime의 자주 쓰는 기능

날짜·시간과 관련된 자료형 `"""Specific date/time and related types."""`

**date.today()** : 오늘 날짜를 date 객체로 가져옵니다.

`today`의 값은 셀을 실행한 날짜입니다.

````python
import datetime

today = datetime.date.today()
print(today, type(today))
````

**timedelta(days=n)** : 날짜에 더하면 n일 뒤, 빼면 n일 전이 됩니다. 두 날짜를 빼도 timedelta가 나옵니다.

````python
next_week = today + datetime.timedelta(days=7)
print(next_week)
````

**strftime(형식)** : 날짜를 원하는 모양의 문자열로 바꿉니다. %Y 연도, %m 월, %d 일

````python
today.strftime("%Y년 %m월 %d일")
````

## datetime 예제 1: 결재 마감일 계산

- **상황** : 오늘 올린 결재의 마감일을 7일 뒤로 정해 안내 문구로 출력합니다.
- **처리 순서**
  1. 오늘 날짜 구하기
  2. 결재 마감일 구하기 (오늘 날짜에 7일 더하기)
  3. 결재 마감일을 형식에 맞게 출력하기

````python
import datetime

# 1. 오늘 날짜 구하기
today = datetime.date.today()
# 2. 결재 마감일 구하기
due_date = today + datetime.timedelta(days=7)
# 3. 형식에 맞게 출력하기
print("결재 마감일:",
      due_date.strftime("%Y년 %m월 %d일"))
````

**핵심 포인트** : 월말·연말을 넘겨도 날짜가 정확히 계산되고, 읽기 쉬운 형식으로 바꿀 수 있습니다.

## datetime 예제 2: 근속 일수와 근속연수 계산

- **상황** : 입사일부터 오늘까지 며칠, 몇 년 근무했는지 구합니다.
- **처리 순서**
  1. 오늘 날짜 구하기
  2. 입사일 정하기
  3. 근속 기간 구하기 (오늘 날짜에서 입사일 빼기)
  4. 근속 일수·연수 출력하기

````python
import datetime

# 1. 오늘 날짜 구하기
today = datetime.date.today()
# 2. 입사일 정하기
join_date = datetime.date(2023, 3, 1)
# 3. 근속 기간 구하기
service = today - join_date
# 4. 근속 일수·연수 출력하기
print(f"근속 일수: {service.days:,}일")
print(f"근속 연수: {service.days // 365}년")
````

**핵심 포인트** : 날짜끼리 빼면 기간(timedelta)이 나오고, .days로 일수를 꺼냅니다. 윤년도 자동 반영됩니다.

## math의 자주 쓰는 기능

C 표준에 정의된 수학 함수 제공 `"""This module provides access to the mathematical functions defined by the C standard."""`

**math.ceil(x)** : 소수점을 올려 정수로 만듭니다. 박스·인원처럼 모자라면 안 되는 수를 셀 때 씁니다.

````python
import math

math.ceil(52.1)
````

**math.floor(x)** : 소수점을 버려 정수로 만듭니다. 꽉 채운 묶음이 몇 개인지 셀 때 씁니다.

````python
math.floor(52.9)
````

**math.prod(목록)** : 목록의 값을 모두 곱합니다. 가로×세로×단처럼 여러 수를 한 번에 곱할 때 씁니다.

````python
math.prod([4, 5, 3])  # 가로 4 × 세로 5 × 3단
````

## math 예제: 포장 박스 수량 계산

- **상황** : 주문 1,250개를 24개들이 박스에 담을 때 필요한 박스 수를 구합니다.
- **처리 순서**
  1. 주문 수량 정하기
  2. 박스당 수량 정하기
  3. 필요한 박스 개수 구하기 (주문 수량 ÷ 박스당 수량을 올림)
  4. 결과 출력하기

````python
import math

# 1. 주문 수량 정하기
order = 1250
# 2. 박스당 수량 정하기
per_box = 24
# 3. 필요한 박스 개수 구하기
boxes = math.ceil(order / per_box)
# 4. 결과 출력하기
print(f"필요한 박스: {boxes}개")
````

**핵심 포인트** : 1250 ÷ 24는 52 나머지 2입니다. 남은 2개도 박스가 필요하므로 올림(ceil)을 씁니다.

## random의 자주 쓰는 기능

무작위 값을 만드는 생성기 `"""Random variable generators."""`

아래 셀들은 실행할 때마다 결과가 달라집니다.

**random.choice(목록)** : 목록에서 하나를 무작위로 뽑습니다. 실행할 때마다 결과가 달라집니다.

````python
import random

random.choice(["A", "B", "C"])
````

**random.sample(목록, k)** : 같은 값을 두 번 뽑지 않고 k개를 뽑습니다. 로또 번호처럼 중복 없는 추첨에 씁니다.

````python
nums = [1, 2, 3, 4, 5]
random.shuffle(nums)
print(nums)
````

- **상황** : 응모한 직원 5명 중 경품 당첨자 3명을 뽑아 가나다순으로 발표합니다.
- **처리 순서**

## statistics의 자주 쓰는 기능

기본 통계 계산 모듈 `"""Basic statistics module."""`

**statistics.mean(data)** : 평균을 구합니다. 한두 개의 튀는 값에 크게 흔들립니다.

**statistics.median(data)** : 크기순으로 늘어놓은 가운데 값입니다. 튀는 값이 있어도 잘 흔들리지 않습니다.

## statistics 예제: 월 매출 평균과 중앙값

- **상황** : 상반기 월 매출에 튀는 값이 있는지 평균과 중앙값으로 분석합니다.
- **처리 순서**
  1. 1~6월 매출 정하기
  2. 평균 구하기 (모두 더해 개수로 나누기, mean)
  3. 중앙값 구하기 (크기순으로 놓은 가운데 값, median)
  4. 결과 출력하기

````python
import statistics

# 1. 1~6월 매출 정하기 (단위: 만원)
sales = [1200, 1350, 980, 1420, 5100, 1280]
# 2. 평균 구하기
avg = statistics.mean(sales)
# 3. 중앙값 구하기
mid = statistics.median(sales)
# 4. 결과 출력하기
print(f"평균: {avg:,.0f}만원")
print(f"중앙값: {mid:,.0f}만원")
````

## os 예제: 월별 보고서 파일 12개 만들기

- **상황** : 매년 1월에 손으로 만들던 월별 보고서 파일 12개를 한 번에 만듭니다.
- **처리 순서**
  1. 01~12 목록 만들기 (range + 리스트 컴프리헨션, :02d로 두 자리)
  2. 보고서 폴더 만들기
  3. 월별 파일 12개 만들기 (join으로 경로 만들기 → open()으로 쓰기)
  4. 만든 파일 확인하기

이 셀을 실행하면 노트북 파일과 같은 폴더에 `보고서` 폴더와 텍스트 파일 12개가 만들어집니다.

````python
import os
# 1. 01~12 목록 만들기
months = [f"{m:02d}" for m in range(1, 13)]
# 2. 보고서 폴더 만들기
os.makedirs("보고서", exist_ok=True)
# 3. 월별 파일 12개 만들기
for month in months:
    path = os.path.join("보고서", f"{month}월_매출.txt")
    with open(path, "w", encoding="utf-8") as f:
        f.write(f"{month}월 매출 보고서")
# 4. 만든 파일 확인하기
files = sorted(os.listdir("보고서"))
print(files[0], "~", files[-1], f"(총 {len(files)}개)")
````

**핵심 포인트** : 두 자리(01)로 맞추면 이름순이 곧 월 순서가 됩니다.

## json의 자주 쓰는 기능

JSON 형식 데이터 주고받기 `"""JSON (JavaScript Object Notation) ... used as a lightweight data interchange format."""`

**json.dumps(데이터)** : 사전·목록을 JSON 글자로 바꿉니다. 한글은 ensure_ascii=False를 붙입니다.

````python
import json

data = {"상호": "한빛상사"}
json_str = json.dumps(data, ensure_ascii=False)
print(json_str, type(json_str))
````

**json.loads(글자)** : JSON 글자를 사전·목록으로 되돌립니다. 받은 데이터를 꺼내 쓸 때 씁니다.

````python
json_dict = json.loads(json_str)
print(json_dict, type(json_dict))
````

**json.dumps(데이터, indent=2)** : 줄을 바꾸고 들여 써서 사람이 읽기 좋은 모양으로 만듭니다.

````python
print(json.dumps(data, indent=2, ensure_ascii=False))
````

## json 예제: 받은 주문 데이터 합계 내기

- **상황** : 쇼핑몰 시스템에서 받은 주문 데이터(JSON)를 읽어 주문 건수와 합계를 냅니다.
- **처리 순서**
  1. 받은 주문 데이터 확인하기 (목록 [ ] 안에 사전 { }이 3개)
  2. 파이썬 목록으로 바꾸기 (글자 → 목록, loads)
  3. 주문 합계 구하기 (사전마다 "금액"을 꺼내 더하기)
  4. 결과 출력하기

````python
import json
# 1. 쇼핑몰에서 받은 주문 데이터 (JSON 글자)
received = '''[
  {"상품": "볼펜", "금액": 5000},
  {"상품": "노트", "금액": 10000},
  {"상품": "파일", "금액": 3000}
]'''
# 2. 파이썬 목록으로 바꾸기
orders = json.loads(received)
# 3. 주문 합계 구하기
total = 0
for order in orders:
    total += order["금액"]
# 4. 결과 출력하기
print(f"주문 {len(orders)}건, 합계 {total:,}원")
````

**핵심 포인트** : loads 후에는 평범한 목록·사전이라 for문으로 바로 계산합니다.

**① company_utils.py (모듈 파일)**

노트북에서는 셀 첫 줄에 `%%writefile 파일이름`을 쓰면, 셀 내용이 그 이름의 파일로 저장됩니다.

````python
%%writefile company_utils.py
# 부가가치세율: 세율이 바뀌면 이 줄만 고칩니다
VAT_RATE = 0.1 # 10%

# 공급가액에 부가가치세를 더한 최종 결제 금액 계산
def add_vat(price):
    return int(price * (1 + VAT_RATE))

# 금액을 원화 문자열로 변환
def won(amount):
    return f"{amount:,}원"

# 부가세 금액만 계산하는 함수
def vat_amount(price):
    return int(price * VAT_RATE)
````

````python
%%writefile company_utils.py
# 부가가치세율: 세율이 바뀌면 이 줄만 고칩니다
VAT_RATE = 0.1 # 10%

# 공급가액에 부가가치세를 더한 최종 결제 금액 계산
def add_vat(price):
    return int(price * (1 + VAT_RATE))

# 금액을 원화 문자열로 변환
def won(amount):
    return f"{amount:,}원"

# 부가세 금액만 계산하는 함수
def vat_amount(price):
    return int(price * VAT_RATE)

# 할인가 계산
def sale_price(price, rate):
    return int(price * (1 - rate))
````

## 패키지는 모듈 여러 개를 폴더로 묶은 것이다

여러 모듈(.py)을 한 폴더에 모아 두고, 폴더 이름으로 불러 쓰는 방식입니다.

```
업무폴더/
├─ main.py  # 패키지를 쓰는 파일
└─ office/  # 패키지 (폴더)
    ├─ __init__.py
    ├─ stock.py  # 재고 확인 모듈
    └─ order.py  # 주문 수량 계산 모듈
```

**`__init__.py`** : 이 폴더가 패키지라는 표시입니다. 파이썬 3.3부터는 없어도 동작하지만, 빈 파일이라도 넣어 두는 것을 권장합니다.

## 패키지 안의 모듈을 불러오는 방법

| 불러오는 형태 | 사용할 때 | 언제 쓰나 |
|---|---|---|
| `import office.stock` | `office.stock.is_low(12, 20)` | 패키지와 모듈이 분명하게 |
| `from office import stock` | `stock.is_low(12, 20)` | 모듈 이름만으로 짧게 |
| `from office.stock import is_low` | `is_low(12, 20)` | 기능 하나만 필요할 때 |

**점(.)으로 폴더 안을 찾아 들어갑니다** : office.stock는 office 폴더 안의 stock.py라는 뜻입니다. 모듈의 모듈명.기능()과 같은 규칙입니다.

**2. 재고가 부족한지 확인하는 모듈입니다** : office/stock.py

````python
%%writefile office/stock.py
def is_low(count, minimum):
    return count < minimum

# 최소 수량보다 적으면 True
# is_low(12, 20) → True
# is_low(35, 20) → False
````

**3. 주문할 수량을 계산하는 모듈입니다** : office/order.py

````python
%%writefile office/order.py
def order_qty(count, target):
    return target - count

# 목표 수량까지 모자란 개수
# order_qty(12, 50) → 38
# order_qty(8, 50)  → 42
````

## 예제: office 패키지로 부족한 사무용품을 주문한다

- **상황** : 총무팀이 사무용품 재고를 확인해, 최소 20개보다 적은 물품을 50개까지 채울 주문 목록을 만듭니다.
- **처리 순서**
  1. 사무용품별 현재 재고 현황 파악
  2. 재고가 부족한(20개 미만인) 물품 고르기 (stock.is_low(재고, 20)이 True인 것만)
  3. 주문할 수량(목표수량 50개)을 딕셔너리에 담기 (orders[물품] = order.order_qty(재고, 50))
  4. 주문 목록 출력하기

````python
from office import stock, order
# 1. 사무용품별 현재 재고 현황 파악
supplies = {"볼펜": 12, "A4 용지": 35, "포스트잇": 8}
orders = {}  # 주문할 사무용품을 담을 딕셔너리
for item, current in supplies.items():
    # 2. 재고가 부족한 물품 고르기 (최소 20개)
    if stock.is_low(current, 20):
        # 3. 주문 수량을 딕셔너리에 담기
        orders[item] = order.order_qty(current, 50)
# 4. 주문 목록 출력하기
print(orders)
````

**핵심 포인트** : 주문할 물품을 딕셔너리에 모아 두면 출력·저장·합계에 다시 쓸 수 있습니다.

## 외부 패키지 예제 1: holidays로 공휴일을 피해 결재 마감일을 정한다

- **처리 순서**
  1. 한국 공휴일 달력 준비하기 (pip install holidays 후 holidays.KR())
  2. 결재 마감일 구하기
  3. 공휴일·주말이면 하루씩 미루기 (in kr_holidays / weekday() >= 5 (토·일))
  4. 결과 출력하기

먼저 터미널에서 `pip install holidays`로 설치합니다.

````python
import holidays
from datetime import date, timedelta
# 1. 한국 공휴일 달력 준비하기
kr_holidays = holidays.KR()
# 2. 결재 마감일 구하기 (상신일 + 7일)
submit = date(2026, 9, 28)  # 상신일
deadline = submit + timedelta(days=7)
# 3. 공휴일이나 주말이면 하루씩 미루기
while deadline in kr_holidays or deadline.weekday() >= 5:
    deadline += timedelta(days=1)
# 4. 결과 출력하기
print("결재 마감일:", deadline)
````

## 외부 패키지 예제 2: requests로 오늘 환율을 받아온다

- **상황** : 해외 거래 금액을 원화로 바꾸려고, 환율 서버에서 오늘 환율을 JSON으로 받아 옵니다.
- **처리 순서**
  1. 환율 데이터 받아오기
  2. 받은 데이터 구조 확인하기 (indent=2로 펼쳐 보기, json.dumps)
  3. 원화 환율 꺼내기 (사전 안의 사전: data["rates"]["KRW"])

먼저 터미널에서 `pip install requests`로 설치합니다. 인터넷에 연결되어 있어야 실행됩니다.

````python
import json
import requests
# 1. 환율 데이터 받아오기
url = "https://open.er-api.com/v6/latest/USD"
data = requests.get(url).json()
# 2. 받은 데이터 구조 확인하기
print(json.dumps(data, indent=2)[:300])
# 3. 원화 환율 꺼내기
krw = data["rates"]["KRW"]
# 4. 결과 출력하기
print(f"1달러 = {krw:,.0f}원")
````
