### [연습문제 2-15] map(), filter()를 리스트 컴프리헨션으로 바꾸기

아래 셀의 두 줄과 같은 결과가 나오도록 리스트 컴프리헨션으로 바꿔 쓰세요.

````python
prices = [18000, 25000, 9000, 32000]

print(list(map(lambda p: int(p * 1.1), prices)))      # 부가세 포함 가격
print(list(filter(lambda p: p >= 20000, prices)))     # 2만 원 이상

# 여기에 리스트 컴프리헨션으로 작성하세요
````

실행 결과 예시

```
[19800, 27500, 9900, 35200]
[25000, 32000]
```

map()은 `[표현식 for ...]`, filter()는 `[변수 for ... if 조건식]`으로 바꿔 쓸 수 있습니다.
