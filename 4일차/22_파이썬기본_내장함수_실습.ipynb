# 내장 함수

## 내장 함수(Built-in Function)

파이썬에서 기본적으로 제공하는 함수로, 별도의 import 없이 바로 사용할 수 있는 함수

■ 주요 내장 함수

| 분류 | 주요 함수 | 기능 |
|---|---|---|
| 입력·출력 | print(), input() | 데이터 출력 및 입력 |
| 자료형 변환 | int(), float(), str(), bool() | 자료형 변환 |
| 컬렉션 생성 | list(), tuple(), set(), dict() | 컬렉션 객체 생성 |

전체 내장 함수 중 입문 과정에서 활용도가 높은 함수를 중심으로 정리한 것입니다.

## all()

모든 요소가 참(True)일 때만 True를 돌려줍니다.     
빈 리스트는 `all([])` → `True`입니다.

■ 코드

````python
# all([True, True, True])
# all([True, True, False])
# all([])
# all([1,2,3])
# all([0,2,3])
# all(["a","a","a"])
# all(["","a","b"])
````

````python
# 우리반 학생들이 시험을 봤는데 점수는 아래와 같습니다.
scores = [80, 95, 55]

# 모두 합격(60점 이상)인가요? 
all([score >= 60 for score in scores])
all(map(lambda x:x>=60, scores))
````

## enumerate()

**반복할 때** 번호(인덱스)와 값을 함께 꺼냅니다.

■ 동작 원리             
`enumerate(['민수','지은','서준'])`

| students | '민수' | '지은' | '서준' |
|---|---|---|---|
| enumerate() 결과 | (0, '민수') | (1, '지은') | (2, '서준') |

- 번호와 값이 튜플로 묶여 하나씩 나옵니다.
- `for i, students in ...`처럼 두 변수로 나눠 받습니다.
- 번호는 0부터 시작합니다.

■ 코드

````python
students = ["민수", "지은", "서준"]

for num, student in enumerate(students, start=1):
    print(num, student)
````

TIP enumerate(fruits, start=1)로 쓰면 번호가 1부터 시작합니다. range(len(fruits))는 필요 없습니다.

## zip()

여러 리스트에서 같은 위치의 값끼리 묶어 줍니다.

■ 동작 원리

| names | '민수' | '지은' | '서준' |
|---|---|---|---|
| scores | 90 | 85 | 77 |
| zip() 결과 | ('민수', 90) | ('지은', 85) | ('서준', 77) |

- 0번은 0번끼리, 1번은 1번끼리 튜플로 묶입니다.
- for name, score in ...으로 나눠 받습니다.

■ 코드

````python
students = ["민수", "지은", "서준"]
scores = [90, 85, 77]
````

TIP 길이가 다르면 짧은 쪽에 맞춰 멈춥니다. zip([1, 2, 3], ["a", "b"]) → (1, 'a'), (2, 'b')

## enumerate(), zip() 정리

반복할 때 값을 짝지어 꺼냅니다.

■ 한눈에 비교

| 함수 | 하는 일 | 요소 개수 | 예시 → 결과 |
|---|---|---|---|
| enumerate() | 번호와 값을 묶음 | 그대로 | `enumerate(["a", "b"])` → (0, 'a'), (1, 'b') |
| zip() | 같은 위치끼리 묶음 | 짧은 쪽 기준 | `zip([1, 2], ["a", "b"])` → (1, 'a'), (2, 'b') |

■ 함께 쓰기: 짝짓고 번호 붙이기

````python
students = ["민수", "지은", "서준"]
scores = [90, 85, 77]
````

- zip()으로 짝지은 뒤 enumerate()로 번호를 붙입니다.
- (n, s)처럼 괄호로 짝을 나눠 담습니다.
- 결과는 한 번만 꺼낼 수 있습니다.

## map()

모든 요소에 같은 함수를 적용해 새 값으로 바꿉니다.

■ 동작 원리

| nums | 1 | 2 | 3 | 4 |
|---|---|---|---|---|
| n * 10 결과 | 10 | 20 | 30 | 40 |

- 모든 요소에 같은 함수를 적용하므로 개수는 그대로입니다.
- 형태: map(함수, 리스트)
- 결과는 map 객체이므로 list()로 감싸야 값이 보입니다.

■ 코드

````python
# 리스트의 각 요소에 10을 곱합니다.
nums = [1, 2, 3, 4]
result = map(______________, nums)
print(list(result))
````

````python
# 리스트의 각 요소를 정수로 변환합니다.
data = list(map(___, ["1", "2", "3"]))
print(data)
````

TIP lambda n: n * 10은 def f(n): return n * 10을 한 줄로 쓴 이름 없는 함수입니다.

## filter()

조건이 참(True)인 요소만 골라냅니다.

■ 동작 원리

| nums | 1 | 2 | 3 | 4 | 5 | 6 |
|---|---|---|---|---|---|---|
| n % 2 == 0 결과 | | 2 | | 4 | | 6 |

- 조건이 True인 요소만 남아 개수가 줄어듭니다.
- 형태: filter(조건 함수, 리스트)
- 결과는 filter 객체이므로 list()로 감싸야 값이 보입니다.

■ 코드

````python
# 짝수만 출력합니다.
nums = [1, 2, 3, 4, 5, 6]
result = filter(______________, nums)
print(list(result))
````

````python
# 빈 문자열은 None으로 채웁니다.
words = ["hi", "", "ok", ""]
print(list(map(lambda word: None if word == "" else word, words)))
````

TIP filter(None, 리스트)는 거짓으로 판단되는 값(0, "", None)을 모두 걸러 냅니다.

## map(), filter() 정리

함수를 적용해 데이터를 가공합니다.

■ 한눈에 비교

| 함수 | 하는 일 | 요소 개수 | 예시 → 결과 |
|---|---|---|---|
| map() | 모든 요소를 변환 | 그대로 | `map(str, [1, 2])` → '1', '2' |
| filter() | 조건에 맞는 요소만 선택 | 줄어듦 | `filter(None, [0, 1, 2])` → 1, 2 |

■ 리스트 컴프리헨션으로 바꿔 쓰기

````python
nums = [1, 2, 3, 4]

list(map(lambda n: n * 10, nums))

list(filter(lambda n: n % 2 == 0, nums))
````

- 결과를 보려면 list()로 감싸야 합니다.
- 한 번 꺼내면 비어 있습니다. 여러 번 쓰려면 list()로 저장해 두세요.
- 간단한 변환·선별은 리스트 컴프리헨션으로도 쓸 수 있습니다.
