# 람다 함수

## lambda 함수

이름 없이 한 줄로 만드는 간단한 함수입니다.

■ 생김새

```python
# def
def add(a, b):
    return a + b

# 한 줄로 → lambda
add = lambda a, b: a + b
```

| 구성 | `lambda` | `a, b` | `:` | `a + b` |
|---|---|---|---|---|
| 뜻 | 키워드 | 매개변수 | | 돌려줄 값 |

주로 함수의 인자로 사용합니다.

■ 코드

````python
# def로 만든 함수
def square(x):
    return x * x

# lambda로 만든 함수
square2 = lambda x : x * x
square2(3)
````

TIP lambda에는 식 하나만 씁니다. return 없이도 식의 결과가 돌려집니다.

## lambda 활용

함수를 값처럼 다른 함수에 넘길 때 씁니다.

■ 자주 쓰는 곳

| 쓰는 곳 | 예 | 결과 |
|---|---|---|
| sorted(key=) | `sorted([3, -5, 1], key=lambda n: abs(n))` | [1, 3, -5] |

````python
print(sorted([3, -5, 1], key=lambda n: abs(n)))
````
