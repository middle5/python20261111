> **실습 파일 사용법**
>
> - **[직접 입력]** 표시가 있는 셀: 바로 위 마크다운 셀의 코드를 보고 직접 입력한 뒤 실행합니다.
> - **[직접 작성]** 표시가 있는 셀: 문제를 읽고 코드를 스스로 작성합니다. 정답은 완성 파일에 있습니다.
> - **[실행만]** 표시가 있는 셀: 코드를 입력하지 않고 그대로 실행합니다.

> 이 노트북 파일의 예제는 `input()`으로 값을 입력받습니다. 셀을 실행하면 입력 칸이 나타나므로, 값을 입력하고 Enter를 누릅니다. 정상 값과 잘못된 값을 번갈아 입력하면서 결과를 비교해 봅니다.

## 예외처리 기초

### 예외(Exception)란?

- 프로그램 실행 중 예상치 못한 오류가 발생하는 상황
- 예: 0으로 나누기, 잘못된 입력 값, 파일 없음 등..
- 이런 예외를 처리하지 않으면 프로그램이 중단됩니다.

### 예외처리의 목적

- 프로그램이 갑작스런 오류로 중단되는 것을 방지합니다.
- 사용자에게 명확한 에러 메시지를 보여줄 수 있습니다.

## 예외 처리 문법

### 기본 문법

```python
try:
    # 실행할 코드
except 예외종류:
    # 예외 발생 시 실행할 코드
```

- `try` 블록 : 예외가 발생할 수 있는 코드를 넣습니다.
- `except` 블록 : 예외가 발생했을 때 실행할 코드를 넣습니다.

## 예외 처리 문법

### 기본 문법

**[직접 입력]** 아래 코드를 다음 셀에 입력하고 실행합니다.

```python

```

## 예외 처리 문법

### 두 개 이상의 예외 처리

예외처리 전 코드

```python
# 두 숫자를 입력 받아 나누기
num1 = float(input("첫번째 숫자 >>> "))
num2 = float(input("두번째 숫자 >>> "))
result = 10/1
print(f'{num1} / {num2} = {num1/num2}')
```

위 코드에서는 두 가지 예외가 발생할 수 있습니다.

- 숫자가 아닌 글자를 입력하면 `float()`에서 `ValueError`
- 두번째 숫자에 0을 입력하면 `num1/num2`에서 `ZeroDivisionError`

`except`를 여러 개 이어 쓰면 예외 종류마다 다른 처리를 할 수 있습니다.

**[직접 입력]** 아래 코드를 다음 셀에 입력하고 실행합니다.

```python
# 두 숫자를 입력 받아 나누기
try:
  num1 = float(input("첫번째 숫자 >>> "))
  num2 = float(input("두번째 숫자 >>> "))
  result = 10/1
  print(f'{num1} / {num2} = {num1/num2}')
except ValueError:
  print("숫자만 입력할 수 있습니다.")
except ZeroDivisionError:
  print("0으로 나눌 수 없습니다.")
```

## 예외 처리 문법

### 의도적으로 예외 발생시키기

```python
raise 예외명("예외메시지")
```

`raise`는 조건에 맞지 않는 값이 들어왔을 때 예외를 직접 발생시킵니다.

예를 들어 위 연습에서 `0`이나 `-1`을 입력하면 `todo[-1]`, `todo[-2]`처럼 뒤에서부터 값을 꺼내므로 오류가 나지 않습니다. 이런 경우를 막으려면 `raise`로 예외를 직접 발생시킵니다.

`idx`가 0 이하이면 `raise`로 `IndexError`를 발생시킵니다. `0`, `-1`, `2`, `5`를 차례로 입력해 결과를 비교해 봅니다.

**[직접 입력]** 아래 코드를 다음 셀에 입력하고 실행합니다.

```python
try:
  todo = ["운동", "독서", "공부"]
```

````python

import os
import glob
import math

from pypdf import PdfReader, PdfWriter, Transformation
from pypdf.errors import PdfReadError
from pypdf.generic import RectangleObject
from pypdf._page import PageObject
from pypdf import PaperSize


# ==================================================
# 1. 사용자 설정
# ==================================================

# PDF 파일이 들어있는 폴더 경로
INPUT_FOLDER = r"C:\kjo\workspace\파일정리 테스트\파이썬수업자료\파이썬수업자료"

# 병합된 PDF 저장 경로 및 파일명
OUTPUT_FILE = r"C:\kjo\workspace\파일정리 테스트\테스트 결과"

# 한 페이지에 배치할 PDF 페이지 수
PAGES_PER_SHEET = 2

# 페이지 방향: "landscape" 또는 "portrait"
PAGE_ORIENTATION = "landscape"

# 용지 크기: A4
PAPER_SIZE = "A4"

# 페이지 바깥 여백
MARGIN = 20

# PDF 페이지 사이 간격
GAP = 10


# ==================================================
# 2. 용지 방향 및 크기 설정
# ==================================================

if PAPER_SIZE.upper() == "A4":
    PAGE_WIDTH, PAGE_HEIGHT = (595.28, 841.89)

elif PAPER_SIZE.upper() == "A3":
    PAGE_WIDTH, PAGE_HEIGHT = (841.89, 1190.55)

elif PAPER_SIZE.upper() == "LETTER":
    PAGE_WIDTH, PAGE_HEIGHT = (612, 792)

else:
    raise ValueError("지원하지 않는 용지 크기입니다.")

if PAGE_ORIENTATION.lower() == "landscape":
    PAGE_WIDTH, PAGE_HEIGHT = PAGE_HEIGHT, PAGE_WIDTH

elif PAGE_ORIENTATION.lower() != "portrait":
    raise ValueError("landscape 또는 portrait를 입력하세요.")


# ==================================================
# 3. 배치 설정
# ==================================================

if PAGES_PER_SHEET < 1:
    raise ValueError("PAGES_PER_SHEET는 1 이상이어야 합니다.")

# 행과 열 자동 계산
COLS = math.ceil(math.sqrt(PAGES_PER_SHEET))
ROWS = math.ceil(PAGES_PER_SHEET / COLS)

# 한 페이지에서 각 PDF가 차지할 영역
CELL_WIDTH = (
    PAGE_WIDTH - 2 * MARGIN - (COLS - 1) * GAP
) / COLS

CELL_HEIGHT = (
    PAGE_HEIGHT - 2 * MARGIN - (ROWS - 1) * GAP
) / ROWS

if CELL_WIDTH <= 0 or CELL_HEIGHT <= 0:
    raise ValueError("여백이나 간격이 너무 큽니다.")


# ==================================================
# 4. PDF 파일 목록 가져오기
# ==================================================

pdf_files = sorted(
    glob.glob(os.path.join(INPUT_FOLDER, "*.pdf"))
)

# 결과 파일이 입력 폴더에 있다면 제외
output_abs = os.path.abspath(OUTPUT_FILE)

pdf_files = [
    f for f in pdf_files
    if os.path.abspath(f) != output_abs
]

if not pdf_files:
    raise FileNotFoundError(
        f"PDF 파일을 찾을 수 없습니다: {INPUT_FOLDER}"
    )

print(f"발견된 PDF 파일: {len(pdf_files)}개")


# ==================================================
# 5. PDF 파일 읽기
# ==================================================

readers = []
all_pages = []

for pdf_file in pdf_files:

    print(f"읽는 중: {os.path.basename(pdf_file)}")

    reader = PdfReader(pdf_file)

    if reader.is_encrypted:
        try:
            result = reader.decrypt("")
            if not result:
                raise ValueError("암호가 설정된 PDF입니다.")
        except Exception as e:
            raise ValueError(
                f"PDF 암호 해제 실패: {pdf_file}"
            ) from e

    readers.append(reader)

    for page in reader.pages:
        all_pages.append(page)


total_pages = len(all_pages)

if total_pages == 0:
    raise ValueError("PDF에 페이지가 없습니다.")

print(f"전체 PDF 페이지 수: {total_pages}페이지")


# ==================================================
# 6. PDF 페이지 배치 및 병합
# ==================================================

writer = PdfWriter()

for start in range(0, total_pages, PAGES_PER_SHEET):

    batch = all_pages[start:start + PAGES_PER_SHEET]

    # 중요: 빈 페이지를 먼저 Writer에 등록
    # Writer에 등록된 PageObject를 직접 수정
    sheet_page = writer.add_blank_page(
        width=PAGE_WIDTH,
        height=PAGE_HEIGHT
    )

    for idx, pdf_page in enumerate(batch):

        # 배치 위치 계산
        row = idx // COLS
        col = idx % COLS

        x = MARGIN + col * (CELL_WIDTH + GAP)

        y = (
            PAGE_HEIGHT
            - MARGIN
            - (row + 1) * CELL_HEIGHT
            - row * GAP
        )

        # 원본 PDF 페이지 크기
        original_width = float(pdf_page.mediabox.width)
        original_height = float(pdf_page.mediabox.height)

        # 원본 비율을 유지하면서 셀 안에 맞게 축소
        scale = min(
            CELL_WIDTH / original_width,
            CELL_HEIGHT / original_height
        )

        # 실제 배치될 크기
        draw_width = original_width * scale
        draw_height = original_height * scale

        # 중앙 정렬
        offset_x = x + (CELL_WIDTH - draw_width) / 2
        offset_y = y + (CELL_HEIGHT - draw_height) / 2

        # 원본 PDF 페이지를 변환하여 현재 페이지에 합성
        transformation = (
            Transformation()
            .scale(scale)
            .translate(offset_x, offset_y)
        )

        sheet_page.merge_transformed_page(
            pdf_page,
            transformation,
            over=True,
            expand=False
        )

    print(
        f"페이지 생성 완료: "
        f"{len(writer.pages)} / "
        f"{math.ceil(total_pages / PAGES_PER_SHEET)}"
    )


# ==================================================
# 7. PDF 저장
# ==================================================

output_dir = os.path.dirname(OUTPUT_FILE)

try:
    # 저장 폴더 생성
    os.makedirs(output_dir, exist_ok=True)

    # 임시 파일에 먼저 저장
    temp_file = OUTPUT_FILE + ".tmp"

    with open(temp_file, "wb") as output:
        writer.write(output)

    # 최종 파일로 변경
    os.replace(temp_file, OUTPUT_FILE)

    print("PDF 저장 성공!")

except PermissionError as e:
    print("PDF 저장 권한 오류!")
    print(f"저장 경로: {OUTPUT_FILE}")
    print("다음 항목을 확인하세요.")
    print("1. 기존 PDF 파일이 열려 있는지")
    print("2. 저장 폴더에 쓰기 권한이 있는지")
    print("3. 다른 프로그램이 파일을 사용 중인지")
    print(f"상세 오류: {e}")

    if os.path.exists(OUTPUT_FILE + ".tmp"):
        os.remove(OUTPUT_FILE + ".tmp")

except Exception as e:
    print(f"PDF 저장 중 오류 발생: {e}")


# ==================================================
# 8. 결과 출력
# ==================================================

print("\n" + "=" * 50)
print("PDF 병합 완료!")
print("=" * 50)

print(f"입력 PDF 파일 수 : {len(pdf_files)}개")
print(f"원본 전체 페이지 : {total_pages}페이지")
print(f"페이지당 배치 수 : {PAGES_PER_SHEET}개")
print(f"최종 페이지 수   : {len(writer.pages)}페이지")
print(f"페이지 방향      : {PAGE_ORIENTATION}")
print(f"용지 크기        : {PAPER_SIZE}")
print(f"배치 형태        : {COLS}열 x {ROWS}행")
print(f"저장 경로        : {OUTPUT_FILE}")

print("=" * 50)
````

## 주요 예외 종류

| 예외 클래스 | 발생 상황 예시 |
|---|---|
| Exception | 모든 예외의 기본 클래스 |
| ZeroDivisionError | 0으로 나눴을 때 |
| ValueError | 잘못된 값 (예: `int("abc")`) |
| TypeError | 타입이 맞지 않을 때 (예: `"1" + 2`) |
| FileNotFoundError | 존재하지 않는 파일을 열 때 |
| IndexError | 리스트 인덱스 범위를 벗어날 때 |
| KeyError | 딕셔너리에서 존재하지 않는 키 접근 시 |
| NameError | 정의되지 않은 변수 참조 |
| AttributeError | 객체에 없는 속성 접근 |
| ImportError | 모듈을 찾지 못했을 때 |
| RuntimeError | 기타 런타임 오류 |
| AssertionError | assert 조건이 거짓일 때 |
