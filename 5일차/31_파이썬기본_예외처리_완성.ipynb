# 예외처리 기초

> 이 노트북 파일의 예제는 `input()`으로 값을 입력받습니다. 셀을 실행하면 입력 칸이 나타나므로, 값을 입력하고 Enter를 누릅니다. 정상 값과 잘못된 값을 번갈아 입력하면서 결과를 비교해 봅니다.

## 예외처리 기초

### 예외(Exception)란?

- 프로그램 실행 중 예상치 못한 오류가 발생하는 상황
- 예: 0으로 나누기, 잘못된 입력 값, 파일 없음 등..
- 이런 예외를 처리하지 않으면 프로그램이 중단됩니다.

### 예외처리의 목적

- 프로그램이 갑작스런 오류로 중단되는 것을 방지합니다.
- 사용자에게 명확한 에러 메시지를 보여줄 수 있습니다.

## 예외 처리 문법

### 기본 문법

```python
try:
    # 실행할 코드
except 예외종류:
    # 예외 발생 시 실행할 코드
```

- `try` 블록 : 예외가 발생할 수 있는 코드를 넣습니다.
- `except` 블록 : 예외가 발생했을 때 실행할 코드를 넣습니다.

## 예외 처리 문법

### 기본 문법

나이를 입력받아 정수로 바꿉니다. 숫자가 아닌 글자(예: `a`)를 입력하면 `int()`에서 `ValueError`가 발생하고, `except ValueError:` 블록이 실행됩니다.

````python
try:
  age = int(input("나이를 입력하세요 >>> "))
except ValueError:
  print("숫자만 입력 가능합니다.")
````

## 예외 처리 문법

### [연습] 다음 코드에 대한 예외처리를 해 봅시다.

```python
# 파일 읽기
with open("test.txt", "r") as f:
  content = f.read()
```

`test.txt` 파일이 없으면 `FileNotFoundError`가 발생합니다. 파일이 없을 때 `파일이 존재하지 않습니다.`를 출력하도록 고쳐 보세요.

````python
# 파일 읽기
try:
  with open("test.txt", "r") as f:
    content = f.read()
except FileNotFoundError:
  print("파일이 존재하지 않습니다.")
````

## 예외 처리 문법

### 두 개 이상의 예외 처리

예외처리 전 코드

```python
# 두 숫자를 입력 받아 나누기
num1 = float(input("첫번째 숫자 >>> "))
num2 = float(input("두번째 숫자 >>> "))
result = 10/1
print(f'{num1} / {num2} = {num1/num2}')
```

위 코드에서는 두 가지 예외가 발생할 수 있습니다.

- 숫자가 아닌 글자를 입력하면 `float()`에서 `ValueError`
- 두번째 숫자에 0을 입력하면 `num1/num2`에서 `ZeroDivisionError`

`except`를 여러 개 이어 쓰면 예외 종류마다 다른 처리를 할 수 있습니다.

````python
# 두 숫자를 입력 받아 나누기
try:
  num1 = float(input("첫번째 숫자 >>> "))
  num2 = float(input("두번째 숫자 >>> "))
  result = 10/1
  print(f'{num1} / {num2} = {num1/num2}')
except ValueError:
  print("숫자만 입력할 수 있습니다.")
except ZeroDivisionError:
  print("0으로 나눌 수 없습니다.")
````

## 예외 처리 문법

### 예외에 대한 상세한 내용 출력하기

```python
try:
    # 실행할 코드
except 예외종류 as e:
    # 예외 발생 시 실행할 코드
```

- `e`는 발생한 예외 정보를 담고 있는 객체입니다.
- 이 객체에는 예외와 관련된 메시지 같은 데이터가 들어있고, `print(e)` 를 하면 그 메시지가 출력됩니다.

````python
try:
  age = int(input("나이를 입력하세요 >>> "))
except ValueError as e:
  print('-'*50)
  print("숫자만 입력 가능합니다.")
  print(e)
````

실행 결과 (`a`를 입력한 경우)
```
나이를 입력하세요 >>> a
--------------------------------------------------
숫자만 입력 가능합니다.
invalid literal for int() with base 10: 'a'
```

## 예외 처리 문법

### 예외에 대한 상세한 내용 출력하기

두 개 이상의 예외 처리에도 `as e`를 붙여 예외 메시지를 함께 출력합니다.

````python
# 두 숫자를 입력받아 나누기
try:
  num1 = float(input("첫번째 숫자 >>> "))
  num2 = float(input("두번째 숫자 >>> "))
  result = 10/1
  print(f'{num1} / {num2} = {num1/num2}')
except ValueError as e:
  print('-'*50)
  print("숫자만 입력할 수 있습니다.")
  print(e)
except ZeroDivisionError as e:
  print('-'*50)
  print("0으로 나눌 수 없습니다.")
  print(e)
````

## 예외 처리 문법

### [연습] 다음 코드에 대한 예외처리를 해 봅시다.

```python
todo = ["운동", "독서", "공부"]
idx = int(input("몇 번째 할 일을 가져올까요?"))
print(f'{idx}번째 할 일은 {todo[idx-1]}입니다.')
```

먼저 아래 값을 입력하면 어떤 예외가 발생하는지 생각해 봅니다.

- `a` 입력
- `5` 입력

````python
try:
  todo = ["운동", "독서", "공부"]
  idx = int(input("몇 번째 할 일을 가져올까요?"))
  print(f'{idx}번째 할 일은 {todo[idx-1]}입니다.')
except ValueError as e:
  print("숫자만 입력할 수 있습니다.", e)
except IndexError as e:
  print("인덱스의 범위를 벗어납니다.", e)
````

**풀이**

- `a`를 입력하면 `int()`에서 `ValueError`가 발생합니다.
- `5`를 입력하면 `todo[4]`를 꺼내려다 `IndexError`가 발생합니다.
- 두 예외를 각각 `except`로 처리합니다.

## 예외 처리 문법

### 의도적으로 예외 발생시키기

```python
raise 예외명("예외메시지")
```

`raise`는 조건에 맞지 않는 값이 들어왔을 때 예외를 직접 발생시킵니다.

예를 들어 위 연습에서 `0`이나 `-1`을 입력하면 `todo[-1]`, `todo[-2]`처럼 뒤에서부터 값을 꺼내므로 오류가 나지 않습니다. 이런 경우를 막으려면 `raise`로 예외를 직접 발생시킵니다.

## 기본 예외 처리 문법

`idx`가 0 이하이면 `raise`로 `IndexError`를 발생시킵니다. `0`, `-1`, `2`, `5`를 차례로 입력해 결과를 비교해 봅니다.

````python
try:
  todo = ["운동", "독서", "공부"]
  idx = int(input("몇 번째 할 일을 가져올까요?"))
  if idx<=0:
    raise IndexError("인덱스는 0보다 커야합니다.")
  print(f'{idx}번째 할 일은 {todo[idx-1]}입니다.')
except IndexError as e:
  print("인덱스의 범위를 벗어납니다.", e)
````

## 주요 예외 종류

| 예외 클래스 | 발생 상황 예시 |
|---|---|
| Exception | 모든 예외의 기본 클래스 |
| ZeroDivisionError | 0으로 나눴을 때 |
| ValueError | 잘못된 값 (예: `int("abc")`) |
| TypeError | 타입이 맞지 않을 때 (예: `"1" + 2`) |
| FileNotFoundError | 존재하지 않는 파일을 열 때 |
| IndexError | 리스트 인덱스 범위를 벗어날 때 |
| KeyError | 딕셔너리에서 존재하지 않는 키 접근 시 |
| NameError | 정의되지 않은 변수 참조 |
| AttributeError | 객체에 없는 속성 접근 |
| ImportError | 모듈을 찾지 못했을 때 |
| RuntimeError | 기타 런타임 오류 |
| AssertionError | assert 조건이 거짓일 때 |
