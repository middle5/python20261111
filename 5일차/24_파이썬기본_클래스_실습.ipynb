## 객체(object)

### 자료는 모두 객체다.

파이썬에서 다루는 자료는 모두 객체(object)입니다. 자료마다 어떤 클래스의 객체인지가 정해져 있습니다.

| 자료 | 어떤 클래스의 객체인가 |
|---|---|
| `1` | int의 객체 |

## 클래스와 객체

### 클래스를 이용하여 객체를 생성합니다.

- **클래스** : 자동차를 만들기 위한 틀
- **객체** : 틀로 만든 실제 자동차

| 클래스 | 넘겨주는 값 | 객체 |
|---|---|---|
| 자동차를 만들기 위한 틀 | 빨간색, 소나타 | 철수의 자동차 |
| 자동차를 만들기 위한 틀 | 흰색, 벤츠 | 영희의 자동차 |

같은 틀(클래스)에 다른 값을 넘기면, 값이 서로 다른 객체가 만들어집니다.

## 클래스와 객체

### 클래스와 객체에는 속성과 메소드(기능)가 있습니다.

**클래스 : 자동차**

| 속성 (변수) | 메소드(기능) (함수) |
|---|---|
| 브랜드<br>모델명<br>색상 | 시동걸기()<br>시동끄기()<br>주행하기(speed) |

↓ 객체생성

**객체 : car1**

| 속성 (변수에 값 할당) | 메소드(기능) (함수호출) |
|---|---|
| 브랜드='벤츠'<br>모델명='E-CLASS'<br>색상='화이트' | 시동걸기()<br>시동끄기()<br>주행하기(60) |

- 속성은 클래스 안의 변수입니다. 객체를 생성하면 속성 변수에 값이 할당됩니다.
- 메소드는 클래스 안의 함수입니다. 객체에서 함수처럼 호출합니다.

## 클래스(class)

### 클래스 정의하기

```python
class 클래스이름:
    def __init__(self, param1, param2):
        self.속성1 = param1
        self.속성2 = param2

    def 메소드1(self, param):
        기능구현

    def 메소드2(self, param):
        기능구현
```

| 구분 | 설명 |
|---|---|
| 클래스 정의 | `class 클래스이름:`으로 클래스를 정의합니다. |
| 생성자 `__init__` | 객체가 생성될 때, 맨 처음 실행돼요.<br>객체가 생성될 때 실행할 기능을 구현해요.<br>객체의 초기 **속성을 지정**해요. |
| 메소드 | 함수의 형태로 기능을 구현해요. |

**※ self**
- 객체 자신을 가리키는 키워드.
- 모든 메소드의 첫번째 인수로 사용하여 메소드를 호출한 객체를 받습니다.

## 클래스(class)

### 클래스 정의하기

자동차 클래스 `Car`를 정의합니다.

- 속성 : `brand`(브랜드), `model`(모델명), `color`(색상)
- 메소드 : `turn_on()`(시동걸기), `turn_off()`(시동끄기), `drive(speed)`(주행하기)

클래스를 정의하는 셀은 실행해도 아무것도 출력되지 않습니다. 클래스는 틀만 만들어 두기 때문입니다.

**[직접 입력]** 아래 코드를 다음 셀에 입력하고 실행합니다.

```python
class Car:
  def __init__(self,b,m,c):
    self.brand = b
    self.model = m
    self.color = c
    print(b,m,c,'출고')
  def turn_on(self):
    print(self.model,'시동을 겁니다.')
  def turn_off(self):
    print(self.model, '시동을 끕니다.')
  def drive(self,speed):
    print(self.model, speed,'속도로 주행합니다.')
```

````python
# [직접 입력] 위 코드를 입력하세요.

class  Car:
    def __init__(self,b,m,c):
        self.brand = b
        self.model = m
        self.color = c
        print(b,m,c,'출고')
    def turn_on(self):
        print(self.model,'시동을 겁니다.')
    def turn_off(self):
        print(self.model,'시동을 끕니다.')
    def drive(self,speed=60):
        print(self.model, speed,'속도로 주행합니다.')
````

## 모듈/패키지의 클래스

| ③ 사용하기 | `객체.메소드()`  `객체.속성` |

| `d.isoformat()` | ③ 메소드 |
