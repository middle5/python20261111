# 파일 입출력

> 이 노트북 파일의 예제는 노트북 파일과 같은 폴더 아래의 `files` 폴더에서 파일을 읽고 씁니다. 아래 **예제 파일 준비** 셀을 가장 먼저 실행합니다.

## 예제 파일 준비

이 노트북 파일에서 쓰는 예제 파일을 `files` 폴더에 만듭니다.

| 파일 | 쓰는 곳 |
|---|---|
| 할일.txt | 연습문제 1 |
| 회의록.txt | 읽는 방법, 연습문제 3 |
| 시스템로그.txt | 연습문제 4 |
| 주문내역.csv | 연습문제 5 ~ 9, 미션 |
| 확인서양식.txt | 미션 |

`메모.txt`, `거래처.csv`는 수업 중에 코드로 직접 만듭니다.

````python
# 예제 파일 준비: files 폴더와 실습용 파일을 만듭니다.
# 다시 실행하면 예제 파일이 처음 상태로 돌아갑니다.
import os

os.makedirs("files", exist_ok=True)

samples = {
    "files/할일.txt": "결재 서류 제출\n",
    "files/회의록.txt": (
        "[논의] 하반기 워크숍 장소\n"
        "[결정] 워크숍은 11월 20일 진행\n"
        "[결정] 예산은 1인 5만원\n"
    ),
    "files/시스템로그.txt": (
        "09:00 INFO 프로그램 시작\n"
        "09:12 ERROR 거래처.csv 파일 없음\n"
        "09:30 INFO 주문 120건 처리\n"
        "09:45 WARN 디스크 사용량 85%\n"
        "10:05 ERROR 메일 서버 연결 실패\n"
    ),
    "files/주문내역.csv": (
        "주문일,거래처,담당자,지역,상품명,수량,금액\n"
        "11-03,미래물산,김민수,경기,노트북,2,2400000\n"
        "11-03,동해기획,이서연,강원,마우스,5,150000\n"
        "11-04,미래물산,김민수,경기,키보드,3,180000\n"
        "11-04,한빛상사,김지현,서울,노트북,1,1200000\n"
        "11-05,동해기획,이서연,강원,노트북,1,1200000\n"
    ),
    "files/확인서양식.txt": (
        "{거래처} {담당자}님께\n"
        "11월 주문 {건수}건, 합계 {금액}원을\n"
        "확인해 주셔서 감사합니다.\n"
    ),
}

for path, text in samples.items():
    # csv는 엑셀에서 열어도 한글이 깨지지 않도록 utf-8-sig로 저장합니다.
    enc = "utf-8-sig" if path.endswith(".csv") else "utf-8"
    with open(path, "w", newline="", encoding=enc) as f:
        f.write(text)

print("예제 파일을 준비했습니다:", os.listdir("files"))
````

# 1. 파이썬 파일 입출력 문법

## ① txt 파일 열고 닫기

**파일 작업 3단계** : 파일 열기 `open(파일, 모드, encoding)` → 읽기 · 쓰기 `read()` · `write()` → 파일 닫기 `close()` / with면 자동

| 항목 | 설명 |
|---|---|
| `open(파일, 모드, encoding)` | 파일 위치·여는 방식·글자 형식 |
| 모드 | r 읽기 · w 쓰기 · a 추가<br>w는 기존 내용을 덮어씁니다. |
| 경로 앞의 `r` | `\`를 글자 그대로 읽게 합니다. |
| FileNotFoundError | 작업 폴더 안에 files 폴더가 있는지 확인합니다. |

따라 쳐 보기: 작업 폴더 확인하고 메모.txt 열기

````python
# 1. 지금 작업 폴더 확인하기
import os
print(os.getcwd())  # 예) C:\Users\kim\업무

# 2. files 폴더의 파일 열기 (상대 경로)
f = open("files/메모.txt", "w", encoding="utf-8")
f.close()
````

- 1번 : `os.getcwd()`는 지금 작업 폴더(주피터 노트북 파일이 있는 폴더)를 알려 줍니다.
- 2번 : 실행하면 `files` 폴더에 빈 `메모.txt`가 생깁니다. 파일 탐색기에서 확인합니다.
- 아직 없는 파일을 r 모드로 열면 오류가 나서, 여기서는 w 모드로 엽니다.

**3. 다른 폴더의 파일 열기 (절대 경로, 앞에 r)**

`C:\업무` 폴더가 실제로 있어야 실행됩니다. 폴더가 없으면 `FileNotFoundError`가 납니다. 파일 탐색기 주소창에서 복사한 자기 폴더 경로로 바꿔서 실행해 봅니다.

````python
# 3. 다른 폴더의 파일 열기 (절대 경로, 앞에 r)
f = open(r"C:\업무\메모.txt", "w", encoding="utf-8")
f.close()
````

## ② txt 읽기·쓰기

| 항목 | 설명 |
|---|---|
| `write()`는 내용을 파일에 저장 | 줄을 바꿔 저장하려면 문장 끝에 `\n`을 붙입니다. |
| `read()`는 전체를 한 번에 읽기 | 한 줄씩 처리 하려면 `for line in f:` 형태를 사용합니다. |
| 실행 후 확인하기 | 코드를 실행한 후 메모.txt 를 메모장으로 열어 봅니다. |

따라 쳐 보기: 메모.txt 만들고 읽기

````python
# 1. 쓰기: 메모.txt 만들기 (w 모드)
f = open("files/메모.txt", "w", encoding="utf-8")
f.write("거래처 미팅 10시\n")
f.write("견적서 보내기\n")
f.close()

# 2. 읽기: 메모.txt 내용 출력 (r 모드)
f = open("files/메모.txt", "r", encoding="utf-8")
print(f.read())
f.close()
````

- `\n`을 지우고 다시 실행하면 두 문장이 한 줄로 붙어서 저장됩니다.
- w 모드는 실행할 때마다 내용을 새로 씁니다. 두 번 실행해도 두 줄만 남습니다.

## with 구문(권장): 파일을 열고 닫는 일을 한 번에

| 항목 | 설명 |
|---|---|
| `with open(...) as f:` | 연 파일을 f라는 이름으로 블록 안에서 씁니다. |
| 들여쓰기까지가 범위 | 4칸 들여 쓴 줄이 끝나면 파일이 닫힙니다. |
| close()가 필요 없다 | 오류가 나도 파일이 닫힙니다. 앞으로는 with로 씁니다. |

따라 쳐 보기: close()와 with 비교

````python
# 1. with 없이: 직접 열고 직접 닫기
f = open("files/메모.txt", "a", encoding="utf-8")
f.write("회의록 정리\n")
f.close()

# 2. with 사용: 블록이 끝나면 자동으로 닫힘
with open("files/메모.txt", "a", encoding="utf-8") as f:
    f.write("결재 서류 제출\n")

print(f.closed)  # True: 이미 닫혀 있음
````

- 메모.txt 끝에 두 줄이 추가돼 있으면 성공입니다.
- `print(f.closed)`가 `True`로 나오는 것으로, 블록이 끝나자 파일이 자동으로 닫혔음을 알 수 있습니다.
- 자주 하는 실수: 콜론(`:`)을 빠뜨리면 `SyntaxError`, 다음 줄을 들여 쓰지 않으면 `IndentationError`가 납니다.

## 파일 모드는 r·w·a 세 가지를 주로 쓴다

같은 파일이라도 어떤 모드로 여느냐에 따라 기존 내용이 남기도 하고 지워지기도 합니다.

| 모드 | 뜻 | 파일이 없으면 | 파일이 있으면 |
|---|---|---|---|
| "r" | 읽기 (기본값) | 오류 (FileNotFoundError) | 처음부터 읽음 |
| "w" | 쓰기 | 새로 만듦 | 기존 내용을 지우고 새로 씀 |
| "a" | 이어 쓰기 | 새로 만듦 | 기존 내용 끝에 덧붙임 |

- 기록을 쌓는 파일(업무일지, 방문기록)은 a, 매번 새로 만드는 보고서는 w로 엽니다.
- 실무 사고는 대부분 w로 기존 파일을 덮어써서 생깁니다.

## 읽는 방법은 파일을 어떻게 쓸지에 따라 고른다

| 방법 | 돌려주는 것 | 업무에서 이럴 때 |
|---|---|---|
| `f.read()` | 파일 전체를 문자열 하나로 | 안내문 양식처럼 통째로 다룰 때 |
| `for line in f` | 한 줄씩 차례로 | 로그·회의록에서 줄마다 조건을 볼 때 |
| `f.readlines()` | 모든 줄을 담은 리스트 | 줄 수를 세거나 n번째 줄을 꺼낼 때 |
| `f.readline()` | 한 줄 | 첫 줄(제목)만 따로 읽을 때 |

````python
with open("files/회의록.txt", encoding="utf-8") as f:
    for line in f:
        print(line.strip())  # 끝의 \n 지우기
````

**읽은 줄 끝에는 줄바꿈이 붙어 있다**

- 출력하거나 글자를 비교하기 전에 `strip()`으로 지웁니다.
- 큰 로그 파일은 `for line in f`로 읽어야 메모리를 적게 씁니다.

```
오늘 할 일: 견적서 발송
```

**실행 후 할일.txt**

### [연습문제 1] 해설

**처리 순서**

3. 한 줄 쓰기 (끝에 `\n` 붙이기)

**실행 후 할일.txt**

```
결재 서류 제출

주간 회의 자료 준비   ← 두 번째 실행
```

**핵심 포인트** : 여러 번 실행해 보세요. 실행할 때마다 한 줄씩 쌓입니다. w로 바꾸면 마지막 한 줄만 남습니다.

- 자주 하는 실수: `\n`을 빼면 다음 할 일이 같은 줄에 붙습니다. with 다음 줄 들여쓰기를 빼면 `IndentationError`가 납니다.

- `strftime()`은 날짜를 원하는 모양의 글자로 바꿉니다.

## [연습문제 2] 회의 참석자 명단 만들기

**상황** : 회의 전에 참석자 이름을 입력받아 참석자.txt로 저장합니다. 명단은 회의마다 새로 씁니다.

**요구 사항**
1. 참석자 이름을 여러 개 입력받는다 — 아무것도 없이 Enter를 누르면 입력 끝
2. 앞에 번호를 붙인다 — `1. 김민수` 형식
3. 참석자.txt에 새로 쓴다 — 첫 줄은 `===참석자 명단===`

**실행 화면**
```
참석자 >>> 김민수
참석자 >>> 이서연
참석자 >>>
```

**실행 후 참석자.txt**
```
===참석자 명단===
1. 김민수
2. 이서연
```

**스스로 확인**
- 두 번 실행해서 이전 명단이 남아 있지 않은지 확인합니다.
- w를 a로 바꾸면 어떻게 되는지 비교해 봅니다.

**힌트** `while True:`와 `break`, 모은 목록은 `writelines()`

````python
names = []
n = 1
# 1. 빈 칸을 입력할 때까지 반복
while True:
    name = input("참석자 >>> ")
    if name == "":
        break
    names.append(f"{n}. {name}\n")
    n += 1
# 2. w 모드로 새로 쓰기
with open("files/참석자.txt", "w", encoding="utf-8") as f:
    f.write("===참석자 명단===\n")
    f.writelines(names)
````

### [연습문제 2] 해설

- `"w"` 모드는 파일을 비우고 새로 씁니다. 회의마다 새 명단이 필요할 때 맞습니다.
- `writelines()`는 줄바꿈을 붙이지 않습니다. 그래서 목록에 넣을 때 `\n`을 붙였습니다.
- 번호 `n`은 참석자를 한 명 담을 때마다 1씩 늘립니다.
- with를 쓰면 `close()`를 빠뜨릴 일이 없습니다.

## [연습문제 3] 회의록에서 결정 사항만 뽑기

**상황** : 회의록.txt에서 `[결정]`으로 시작하는 줄만 골라 보고 싶습니다.

**회의록.txt 파일 내용**
```
[논의] 하반기 워크숍 장소
[결정] 워크숍은 11월 20일 진행
[결정] 예산은 1인 5만원
```

**요구 사항**
1. 회의록.txt를 읽기 모드로 연다 — with 구문 사용
2. 한 줄씩 읽는다 — `for line in f`
3. `[결정]`으로 시작하는 줄만 출력한다 — 줄 끝의 `\n`은 지우고 출력

**실행 결과**
```
[결정] 워크숍은 11월 20일 진행
[결정] 예산은 1인 5만원
```

**스스로 확인**
- 결과 사이에 빈 줄이 없는지 확인합니다.
- `[논의]` 줄만 뽑으려면 어디를 바꿀지 생각해 봅니다.

**힌트** `line.startswith("[결정]")` · `line.strip()`

**도전** 다 한 사람은 `[결정]` 줄을 새 파일 결정사항.txt에 저장해 보세요 — w 모드로 새 파일 열기 · 읽은 줄을 `f.write()`로 쓰기

````python
# 1. 읽기(r) 모드로 파일 열기
with open("files/회의록.txt", "r",
          encoding="utf-8") as f:
    # 2. 한 줄씩 읽기
    for line in f:
        # 3. [결정]으로 시작하는 줄만 출력
        if line.startswith("[결정]"):
            print(line.strip())
````

### [연습문제 3] 해설

**처리 순서**
1. r 모드로 파일 열기 — 파일이 없으면 오류가 납니다
2. for문으로 한 줄씩 읽기 — 줄마다 끝에 `\n`이 붙어 있습니다
3. `[결정]` 줄만 출력하기

**핵심 포인트** : `strip()`으로 줄 끝의 `\n`을 지워야 빈 줄 없이 출력됩니다.

- `startswith()`는 글자가 특정 단어로 시작하는지 True/False로 알려 줍니다.
- 모드를 생략하면 r로 열립니다. `open("files/회의록.txt", encoding="utf-8")`도 같은 뜻입니다.

**[도전] 결정 사항을 새 파일로 저장**

````python
with open("files/회의록.txt", "r", encoding="utf-8") as f:
    with open("files/결정사항.txt", "w", encoding="utf-8") as out:
        for line in f:
            if line.startswith("[결정]"):
                out.write(line)
````

- 읽을 파일(`f`)과 쓸 파일(`out`)을 각각 열고, 변수 이름을 다르게 둡니다.
- 읽은 줄 끝에 이미 `\n`이 있어서 `strip()` 없이 그대로 쓰면 한 줄씩 저장됩니다.
- `with open(...) as f, open(...) as out:` 처럼 한 줄에 두 파일을 열 수도 있습니다.

## [연습문제 4] 로그에서 오류만 골라 보고하기

**상황** : 매일 아침 시스템 로그에서 오류만 찾아 팀장에게 보고합니다.

**시스템로그.txt 파일 내용**
```
09:00 INFO 프로그램 시작
09:12 ERROR 거래처.csv 파일 없음
09:30 INFO 주문 120건 처리
09:45 WARN 디스크 사용량 85%
10:05 ERROR 메일 서버 연결 실패
```

**요구 사항**
1. 시스템로그.txt에서 ERROR 줄만 모은다 — `"ERROR" in line` · 리스트에 `append()`
2. 오류보고.txt를 새로 만든다 — 매번 새 보고서이므로 w 모드
3. 첫 줄에 건수, 이어서 오류 줄을 쓴다 — 오류 2건 → 오류 줄들

**실행 후 오류보고.txt**
```
오류 2건
09:12 ERROR 거래처.csv 파일 없음
10:05 ERROR 메일 서버 연결 실패
```

**스스로 확인**
- 오류보고.txt를 메모장으로 열어 세 줄인지 확인합니다.
- 두 번 실행해도 내용이 쌓이지 않는지 확인합니다.

**힌트** `len(errors)`로 건수 세기 · `f.writelines(errors)`

**도전** 다 한 사람은 WARN 줄도 따로 세어 "경고 N건"을 추가해 보세요 — 조건을 하나 더 만들고 리스트를 하나 더 씁니다

````python
errors = []
# 1. 한 줄씩 읽으며 ERROR 줄만 모으기
with open("files/시스템로그.txt", encoding="utf-8") as f:
    for line in f:
        if "ERROR" in line:
            errors.append(line)
# 2. 건수와 함께 오류보고.txt에 저장
with open("files/오류보고.txt", "w", encoding="utf-8") as f:
    f.write(f"오류 {len(errors)}건\n")
    f.writelines(errors)
````

### [연습문제 4] 해설

- 1단계에서 ERROR 줄을 `errors` 리스트에 모으고, 2단계에서 건수와 함께 한 번에 씁니다.
- 건수를 첫 줄에 쓰려면 다 읽은 뒤에 써야 합니다. 그래서 읽기와 쓰기를 두 개의 with 블록으로 나눕니다.
- **핵심 포인트** : 읽은 줄 끝에 이미 `\n`이 있어서 `writelines()`로 그대로 쓰면 한 줄씩 저장됩니다. 건수 줄에는 `\n`을 직접 붙입니다.
- `"ERROR" in line`은 줄 안 어디에든 ERROR가 있으면 True입니다. `startswith()`와 달리 시각이 앞에 있어도 찾습니다.

**[도전] 경고 건수 추가**

````python
errors, warns = [], []
with open("files/시스템로그.txt", encoding="utf-8") as f:
    for line in f:
        if "ERROR" in line:
            errors.append(line)
        elif "WARN" in line:
            warns.append(line)
with open("files/오류보고.txt", "w", encoding="utf-8") as f:
    f.write(f"오류 {len(errors)}건, 경고 {len(warns)}건\n")
    f.writelines(errors)
````

예제 로그에 WARN 줄이 하나 있어 도전 과제 결과는 `오류 2건, 경고 1건`입니다.

# csv 파일 다루기

## ① csv 파일 열고 닫기: 쓸 때는 newline=""을 추가한다.

**파일 작업 3단계** : 파일 열기 `open(파일, 모드, newline)` → 읽기 · 쓰기 `csv.reader` · `csv.writer` → 파일 닫기 with면 자동

txt와 비교해 보기 (여는 줄만)

```python
# txt 읽기 (비교용)
with open("files/메모.txt", "r", encoding="utf-8") as f:

# csv 읽기: 인코딩만 utf-8-sig
with open("files/거래처.csv", "r", encoding="utf-8-sig") as f:

# csv 쓰기: newline="" 추가
with open("files/거래처.csv", "w", newline="",
          encoding="utf-8-sig") as f:
```

※ 위 코드는 여는 줄만 보여 준 것이라 그대로 실행하면 오류가 납니다. 실제 실행은 다음 ② csv 읽기·쓰기에서 합니다.

| 항목 | 설명 |
|---|---|
| txt와 같은 open() | 경로 쓰는 법과 모드(r·w·a)는 txt와 똑같습니다. |
| newline=""을 붙이는 이유 | csv 파일을 쓸 때 넣지 않으면 엑셀에서 행 사이에 빈 줄이 생길 수 있습니다. |
| utf-8-sig를 쓰는 이유 | 한글이 들어있는 csv를 엑셀에서 열 때 글자가 깨지지 않도록 합니다. |

## 참고. 인코딩: 글자를 숫자로 바꿔 저장하는 규칙이다

글자 `'가'` → 인코딩 규칙 `utf-8` → 저장되는 바이트 `EA B0 80`

같은 글자도 규칙마다 저장되는 바이트가 다릅니다.

| 글자 | utf-8로 저장 | cp949로 저장 |
|---|---|---|
| A | 41 (1바이트) | 41 (1바이트) |
| 가 | EA B0 80 (3바이트) | B0 A1 (2바이트) |
| 한 | ED 95 9C (3바이트) | C7 D1 (2바이트) |

| 항목 | 설명 |
|---|---|
| 인코딩이란? | 컴퓨터는 숫자만 저장하므로, 글자를 숫자(바이트)로 바꾸는 규칙이 필요합니다. |
| 디코딩은 반대 과정 | 파일을 열 때 바이트를 다시 글자로 바꿉니다. 규칙은 encoding으로 정합니다. |
| 규칙이 다르면 글자가 깨진다 | utf-8로 저장한 파일을 cp949로 열면 글자가 깨지거나 오류가 납니다. |

## 참고. utf-8-sig: 파일 맨 앞에 BOM 3바이트를 붙인 utf-8이다

파일 맨 앞 `EF BB BF (BOM)` → 엑셀이 알아봄 "이 파일은 utf-8" → 한글 정상 표시

글자를 저장하는 방식은 같고, 맨 앞 3바이트만 다릅니다.

| 구분 | utf-8 | utf-8-sig |
|---|---|---|
| 맨 앞 3바이트 | 없음 | EF BB BF (BOM) |
| 엑셀에서 열면 | 한글이 깨질 수 있음 | 한글 정상 표시 |
| 주 용도 | 파이썬·웹·일반 파일 | 엑셀로 열 CSV |

| 항목 | 설명 |
|---|---|
| BOM이란? | 파일 맨 앞에 붙는 표시(EF BB BF)로, 어떤 인코딩인지 알려 줍니다. |
| 엑셀로 열 CSV는 utf-8-sig | 엑셀은 BOM이 없으면 utf-8 파일의 한글을 깨뜨려 보여 줍니다. |
| 읽을 때도 utf-8-sig | utf-8로 읽으면 첫 칸 이름 앞에 `\ufeff`가 붙어 나올 수 있습니다. |

## ② csv 읽기·쓰기: csv 모듈을 사용한다

| 항목 | 설명 |
|---|---|
| `writerow()`에는 한 행을 리스트로 전달 | `["상호", "담당자"]`처럼 넘기면 칸이 나뉘어 저장됩니다. |
| `reader()`는 csv를 한 행씩 읽기 | 파일의 내용을 한 행씩 리스트로 읽어옵니다. (예) `['미래물산', '김민수']`<br>`list(csv.reader(f))`로 감싸면 한 번에 받습니다. |
| 실행 후 확인하기 | 거래처.csv를 엑셀과 메모장으로 각각 열어 봅니다. |

따라 쳐 보기: 거래처 명단 쓰고 읽기

````python
import csv
# 1. 쓰기: 칸을 나눠 한 줄씩 저장
with open("files/거래처.csv", "w", newline="",
          encoding="utf-8-sig") as f:
    writer = csv.writer(f)
    writer.writerow(["상호", "담당자"])
    writer.writerow(["미래물산", "김민수"])
# 2. 읽기: 한 줄씩 리스트로 꺼냄
with open("files/거래처.csv", "r", encoding="utf-8-sig") as f:
    for row in csv.reader(f):
        print(row)
````

- 결과는 `['상호', '담당자']`, `['미래물산', '김민수']`처럼 리스트로 나옵니다. `row[0]`이 상호, `row[1]`이 담당자입니다.
- writer와 writerow 줄도 with 아래에 들여 써야 합니다. 들여쓰기를 빼면 파일이 이미 닫힌 뒤라 `ValueError`가 납니다.
- 자주 하는 실수: `writerow("미래물산")`처럼 글자를 그대로 넘기면 한 글자씩 칸이 나뉩니다. 꼭 리스트로 넘깁니다.
- 기존 csv 끝에 줄을 추가할 때는 a 모드로 열고, a 모드에도 `newline=""`을 꼭 붙입니다.

## csv의 한 줄은 리스트 또는 딕셔너리로 다룬다

| 구분 | 리스트 방식 | 딕셔너리 방식 |
|---|---|---|
| 읽기 | `csv.reader(f)` | `csv.DictReader(f)` |
| 쓰기 | `csv.writer(f)` | `csv.DictWriter(f, fieldnames=[...])` |
| 한 줄의 모양 | `['미래물산', '김민수']` | `{'상호': '미래물산', '담당자': '김민수'}` |
| 값 꺼내기 | `row[0]` (칸 번호) | `row["상호"]` (열 이름) |
| 첫 줄(제목 행) | 데이터 한 줄로 읽힘 | 열 이름으로 자동 사용 |
| 이럴 때 쓴다 | 열이 적고 순서가 고정된 파일 | 열이 많거나 순서가 바뀔 수 있는 실무 파일 |

딕셔너리 방식은 열 순서가 바뀌어도 코드를 고칠 필요가 없습니다. 열이 20개인 파일에서 `row[13]`보다 `row["연락처"]`가 훨씬 알아보기 쉽습니다.

## ③ csv 딕셔너리로 읽고 쓰기: 열 이름으로 다룬다

| 항목 | 설명 |
|---|---|
| fieldnames가 열 순서를 정한다 | 딕셔너리의 키 이름과 똑같이 적어야 합니다. |
| `writeheader()`가 제목 행을 쓴다 | 첫 줄에 상호, 담당자가 저장됩니다. |
| `row["열 이름"]`으로 꺼낸다 | 첫 줄이 열 이름이 되어 번호 대신 이름으로 꺼냅니다. |

따라 쳐 보기: 딕셔너리로 거래처 쓰고 읽기

````python
import csv
# 1. 쓰기: 딕셔너리 한 줄씩 저장
with open("files/거래처.csv", "w", newline="",
          encoding="utf-8-sig") as f:
    writer = csv.DictWriter(f, fieldnames=["상호", "담당자"])
    writer.writeheader()
    writer.writerow({"상호": "미래물산", "담당자": "김민수"})
# 2. 읽기: 열 이름으로 꺼내기
with open("files/거래처.csv", "r", encoding="utf-8-sig") as f:
    for row in csv.DictReader(f):
        print(row["상호"], row["담당자"])
````

- 실행 결과는 `미래물산 김민수`입니다.
- writerow에 넘기는 딕셔너리의 키가 fieldnames와 다르면 `ValueError`가 납니다.
- `writeheader()`를 빼먹으면 제목 행 없이 데이터만 저장되어, 나중에 DictReader로 읽을 때 첫 데이터 줄이 열 이름으로 쓰입니다.
- 여러 줄을 한 번에 쓸 때는 `writer.writerows([딕셔너리1, 딕셔너리2])`를 씁니다.

## [연습문제 8] 상품별 수량 집계하기

**상황** : 매주 상품별 판매 수량을 계산기로 더하고 있습니다. 파이썬으로 집계해 집계.csv로 저장합니다.

**주문내역.csv 파일 내용**
```
주문일,거래처,담당자,지역,상품명,수량,금액

11-04,한빛상사,김지현,서울,노트북,1,1200000
11-05,동해기획,이서연,강원,노트북,1,1200000
```

**요구 사항**
1. 주문내역.csv를 DictReader로 읽는다 — 수량은 `int()`로 숫자로 바꾸기
2. 상품별로 수량을 더한다 — 딕셔너리 `total`에 모으기
3. 집계.csv로 저장한다 — 첫 줄은 `상품명,수량`

```
상품명,수량
노트북,4
마우스,5
키보드,3
```

**스스로 확인**
- 집계.csv를 엑셀로 열어 숫자가 맞는지 확인합니다.
- `int()`를 빼고 실행해 비교해 봅니다.

````python
import csv
total = {}
# 1. 상품별로 수량 더하기
with open("files/주문내역.csv", "r", encoding="utf-8-sig") as f:
    reader = csv.DictReader(f)
    for row in reader:
        name = row["상품명"]
        num = int(row["수량"])
        if name in total:
            total[name] += num
        else:
            total[name] = num
# 2. 집계.csv로 저장
with open("files/집계.csv", "w", newline="", encoding="utf-8-sig") as f:
    writer = csv.writer(f)
    writer.writerow(["상품명", "수량"])
    for name, qty in total.items():
        writer.writerow([name, qty])
````

### [연습문제 8] 해설 ② get()으로 집계하기

- `total.get(name, 0)`은 키가 있으면 그 값을, 없으면 0을 돌려줍니다.
- 앞의 if/else 네 줄과 같은 일을 한 줄로 합니다.
- 결과는 ① 풀이와 똑같습니다. 이해가 어려우면 ① 풀이를 그대로 써도 됩니다.

````python
import csv
total = {}
# 1. 상품별로 수량 더하기 (if문 4줄을 1줄로)
with open("files/주문내역.csv", "r", encoding="utf-8-sig") as f:
    reader = csv.DictReader(f)
    for row in reader:
        name = row["상품명"]
        num = int(row["수량"])
        total[name] = total.get(name, 0) + num
# 2. 저장은 앞 장과 같음
with open("files/집계.csv", "w", newline="", encoding="utf-8-sig") as f:
    writer = csv.writer(f)
    writer.writerow(["상품명", "수량"])
    for name, qty in total.items():
        writer.writerow([name, qty])
````

### [연습문제 8] 해설 ③ 표준 라이브러리로 집계하기

- collections는 설치 없이 쓰는 파이썬 표준 라이브러리입니다.
- `defaultdict(int)`는 처음 나온 키를 0으로 만들어 줘서 `get()`이 필요 없습니다.
- `Counter`도 같은 일을 합니다: `total = Counter()` 후 `total[name] += 수량`
- 결과는 ①·② 풀이와 똑같습니다. 필수는 아니고 참고용입니다.
- `print(total)`을 하면 `defaultdict(<class 'int'>, {...})`처럼 보이지만 딕셔너리처럼 그대로 쓸 수 있습니다.

````python
import csv
from collections import defaultdict

# 1. 없는 키는 자동으로 0에서 시작
total = defaultdict(int)
with open("files/주문내역.csv", encoding="utf-8-sig") as f:
    for row in csv.DictReader(f):
        total[row["상품명"]] += int(row["수량"])

# 2. 저장은 앞 장과 같음
with open("files/집계.csv", "w", newline="", encoding="utf-8-sig") as f:
    writer = csv.writer(f)
    writer.writerow(["상품명", "수량"])
    for name, qty in total.items():
        writer.writerow([name, qty])
````

## [연습문제 9] 매출 집계하기

```
주문일,거래처,담당자,지역,상품명,수량,금액
```

**요구 사항**

**실행 결과**
```
합계 5,130,000원 · 하루 평균 1,710,000원
최고 매출일 11-03 (2,550,000원)
```

**스스로 확인**
- 11-03 매출이 2,550,000원인지 확인합니다.

````python
import csv
daily = {}
with open("files/주문내역.csv", encoding="utf-8-sig") as f:
    for row in csv.DictReader(f):
        day = row["주문일"]
        daily[day] = daily.get(day, 0) + int(row["금액"])
total = sum(daily.values())
avg = total // len(daily)
best = max(daily, key=daily.get)
print(f"합계 {total:,}원 · 하루 평균 {avg:,}원")
print(f"최고 매출일 {best} ({daily[best]:,}원)")
````

## [미션] 거래처별 주문 확인서 만들기

**상황** : 월말마다 거래처별 주문 건수와 합계를 정리해 확인서를 한 장씩 보내야 합니다.

```
주문일,거래처,담당자,지역,상품명,수량,금액
11-03,미래물산,김민수,경기,노트북,2,2400000
11-03,동해기획,이서연,강원,마우스,5,150000
… (연습문제와 같은 5건)
```

**입력 ② 확인서양식.txt**
```
{거래처} {담당자}님께
11월 주문 {건수}건, 합계 {금액}원을

```

**결과: 확인서 폴더**
```
확인서_미래물산.txt
확인서_동해기획.txt
```

**확인서_미래물산.txt 내용**
```
미래물산 김민수님께
11월 주문 2건, 합계 2,580,000원을

```

**도전** 발송기록.csv에 거래처, 파일 이름, 만든 시각을 한 줄씩 남겨 봅니다.

### 단계별 힌트

1. 양식을 통째로 읽는다 — `f.read()`로 문자열 하나에 담기
2. 거래처별로 건수와 금액을 모은다 — `info[거래처] = [담당자, 건수, 금액]`
3. 확인서 폴더를 만든다 — `os.makedirs("files/확인서", exist_ok=True)`
4. 거래처마다 바꿔 끼워 저장한다 — `replace("{거래처}", 이름)` · `f"files/확인서/확인서_{이름}.txt"`

### 제출 전 점검

- □ 확인서 폴더에 파일이 3개 생겼는가
- □ 미래물산 합계가 2,580,000원인가
- □ `{거래처}` 같은 글자가 남아 있지 않은가
- □ 다시 실행해도 오류 없이 덮어쓰는가
- □ 주문을 추가하면 건수와 합계가 바뀌는가
- □ (도전) 발송기록.csv에 3줄이 기록되었는가

````python
import csv, os
# 1. 양식 읽기
with open("files/확인서양식.txt", encoding="utf-8") as f:
    template = f.read()
# 2. 거래처별로 [담당자, 건수, 금액] 모으기
info = {}
with open("files/주문내역.csv", encoding="utf-8-sig") as f:
    for row in csv.DictReader(f):
        name = row["거래처"]
        if name not in info:
            info[name] = [row["담당자"], 0, 0]
        info[name][1] += 1
        info[name][2] += int(row["금액"])
# 3. 거래처마다 확인서 저장
os.makedirs("files/확인서", exist_ok=True)
for name, (person, cnt, amt) in info.items():
    text = template.replace("{거래처}", name).replace("{담당자}", person)
    text = text.replace("{건수}", str(cnt)).replace("{금액}", f"{amt:,}")
    with open(f"files/확인서/확인서_{name}.txt", "w", encoding="utf-8") as out:
        out.write(text)
print(f"{len(info)}개 저장")
````

### [미션] 예시 답안 해설

- 양식은 한 번만 읽고 거래처마다 복사본을 바꿔 씁니다.
- `info`는 거래처마다 `[담당자, 건수, 금액]`을 담습니다. `print(info)`로 `{'미래물산': ['김민수', 2, 2580000], …}`이 나오면 성공입니다.
- `replace()`는 원본을 바꾸지 않고 새 문자열을 돌려주므로, 결과를 변수에 다시 담아야 남습니다.
- `for name, (person, cnt, amt) in info.items()`는 리스트 `[담당자, 건수, 금액]`을 세 변수로 한 번에 나눠 받습니다.
- 금액은 `f"{amt:,}"`로 2,580,000처럼 쉼표를 넣어 글자로 바꿉니다. 건수는 `str(cnt)`로 바꿉니다.
- 정답 기준: 미래물산 2건 2,580,000원 · 동해기획 2건 1,350,000원 · 한빛상사 1건 1,200,000원

**[도전] 발송기록.csv 남기기**

````python
import csv
from datetime import datetime

# 앞 셀에서 만든 template, info를 그대로 씁니다.
with open("files/발송기록.csv", "a", newline="", encoding="utf-8-sig") as log:
    writer = csv.writer(log)
    for name, (person, cnt, amt) in info.items():
        text = template.replace("{거래처}", name).replace("{담당자}", person)
        text = text.replace("{건수}", str(cnt)).replace("{금액}", f"{amt:,}")
        path = f"files/확인서/확인서_{name}.txt"
        with open(path, "w", encoding="utf-8") as out:
            out.write(text)
        writer.writerow([name, path, datetime.now().strftime("%Y-%m-%d %H:%M")])
print("발송기록을 남겼습니다.")
````

- 발송기록.csv는 기록을 쌓는 파일이므로 a 모드로 엽니다. a 모드에도 `newline=""`과 `utf-8-sig`를 붙입니다.
- 확인서를 저장할 때마다 `[거래처, 파일 이름, 시각]`을 한 줄씩 씁니다.

엑셀·워드·PPT·PDF는 `open()`만으로 다룰 수 없고, 형식마다 전용 라이브러리를 설치해 씁니다.

## 업무 자동화의 시작, 파이썬으로 파일 내용 정리하기

파일 자동화는 모두 **읽기 → 가공 → 쓰기** 세 단계입니다.

| 단계 | 하는 일 | 예시 |
|---|---|---|
| ① 읽기(불러오기) | 파일의 데이터를 파이썬으로 불러옵니다. | 거래처.csv · 경비내역.csv |
| ② 가공(처리하기) | 필요한 기준으로 데이터를 정리하고 계산하거나 변환합니다. | if 조건 · 합계 · replace |
| ③ 쓰기(저장하기) | 가공한 결과를 새 파일로 저장합니다. | 보고서.txt · 정산표.xlsx |

## AI가 만든 코드는 실행 전후로 점검한다

**점검 순서**
1. 읽기: 무엇을 하는지 주석부터 본다 — 모르는 줄은 AI에게 '이 줄 설명해줘'
2. 실행 전: 파일 이름과 경로를 확인한다 — 열 이름·encoding이 내 파일과 같은가
3. 실행: 오류가 나면 메시지째 다시 묻는다 — 오류 전문 복사 + '이 오류 고쳐줘'
4. 실행 후: 결과 숫자를 직접 맞춘다 — 손으로 한두 개 계산해 비교하기

**제출 전 점검**
- □ 코드에 없는 열 이름을 지어내지 않았는가
- □ encoding, newline이 들어 있는가
- □ 원본 파일을 덮어쓰지 않는가 (w 모드 주의)
- □ 결과 파일을 열어 숫자를 확인했는가
- □ 개인정보·실제 금액을 AI에 붙여 넣지 않았는가
- □ 코드 한 줄 한 줄을 설명할 수 있는가

## 정리: 업무 자동화에 쓰는 파일 입출력 코드

| 하고 싶은 일 | 쓰는 코드 |
|---|---|
| 기존 내용 뒤에 덧붙이기 | `open(파일, "a", encoding="utf-8")` |
| 새로 쓰기 (기존 내용 지움) | `open(파일, "w", encoding="utf-8")` |
| 파일 자동으로 닫기 | `with open(...) as f:` |
| 줄마다 조건 보기 | `for line in f:` + `line.strip()` |
| csv를 열 이름으로 읽기 | `csv.DictReader(f)` → `row["열 이름"]` |
| csv 쓰기 (빈 줄 없이) | `open(..., "w", newline="", encoding="utf-8-sig")` |
| 키별로 합계 모으기 | `if 키 in 딕셔너리:` 더하기 / `else:` 새로 넣기 |
| 양식에 값 바꿔 끼우기 | `양식.replace("{거래처}", 이름)` |
