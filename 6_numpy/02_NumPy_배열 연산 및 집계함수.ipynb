## 배열 간 연산
- 동일한 인덱스(위치)끼리 연산한다.
- 배열 간 연산을 하려면 두 배열의 **모양(shape)이 같거나 브로드캐스팅이 가능**해야 한다.

```python
# --------------------------------------------------
# 배열 간 연산 → 같은 위치끼리 계산
# --------------------------------------------------

arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])
```

```python
# 배열 간 덧셈
arr1 + arr2
```

```python
# 배열 간 뺄셈

arr1 - arr2
```

```python
# 배열 간 곱셈 (원소별 곱)

arr1 * arr2
```

```python
# 배열 간 나눗셈 (원소별 나눗셈)

arr1 / arr2
```

> ⚠️ `*` 는 **원소별 곱**입니다. 행렬곱이 아니에요.
> 행렬곱이 필요하면 `arr1 @ arr2` 또는 `np.dot(arr1, arr2)` 를 사용합니다.

```python
# 배열 비교 연산 → 각 위치를 비교해 True/False 배열 반환

arr1 == arr2
```

## 브로드캐스팅 연산
* 모양이 **다른** 배열끼리 연산할 때, **작은 쪽을 늘려(복사해)** 맞춰주는 기능
  * (모양이 완전히 같으면 늘릴 게 없으니 그냥 원소별 연산)
* 늘어나는 규칙 : 크기가 **1인 쪽**이 상대 크기만큼 복사되어 늘어난다
  * 크기가 1도 아니고 서로 다르면 → 연산 불가(에러)

```python
l1 = [1,2,3]
[l+1 for l in l1]
```

```python
# --------------------------
# ① 배열과 스칼라 간 연산 (숫자 하나가 전체에 적용됨)
# --------------------------

arr = np.array([[1,2,3],[4,5,6]])
print(arr)
print('----------')

arr + 1
```

```python
# --------------------------
# ② 가로 행벡터와의 연산 (한 줄이 세로로 복사됨)
# --------------------------

arr1 = np.array([[1,2,3],[4,5,6]])
arr2 = np.array([1,2,3])
print(arr1)
print(arr2)
print('----------')

arr1 + arr2
```

```python
# --------------------------
# ③ 세로 열벡터와의 연산 (한 열이 가로로 복사됨)
# --------------------------

arr1 = np.array([[1,2,3],[4,5,6]])
arr2 = np.array([[10],[20]])
print(arr1)
print(arr2)
print('----------')
```

```python
# --------------------------
# ④ 열벡터 + 행벡터 → 격자로 확장
# --------------------------

arr1 = np.array([[1],[2],[3]])
arr2 = np.array([4,5,6])
print(arr1)
print(arr2)
print('----------')
```

## 0으로 나누기 (ZeroDivision)
- 파이썬에서 `1/0` 은 **오류(ZeroDivisionError)** 지만,
- NumPy 배열 연산에서는 **오류 대신 경고만** 내고 계속 진행한다.
  * `1/0 → inf` (무한대)
  * `0/0 → nan` (계산 불가, Not a Number)

```python
# 파이썬 기본 나눗셈은 오류
# 1/0   # → ZeroDivisionError
```

```python
# NumPy는 inf / nan 으로 처리하고 경고만 표시

arr = np.array([0, 1, 2])
```

## 집계함수
* 주요 집계함수
    * `sum()` : 합계  ·  `mean()` : 평균  ·  `median()` : 중앙값
    * `min()` : 최소값  ·  `max()` : 최대값  ·  `ptp()` : 최대-최소(범위)
    * `std()` : 표준편차  ·  `var()` : 분산  ·  `percentile()` : 백분위수
    * `argmin()` : 최소값의 인덱스  ·  `argmax()` : 최대값의 인덱스
    * `cumsum()` : 누적합계  ·  `cumprod()` : 누적곱
    * `unique()` : 유일한 값(중복 제거)
* 두 가지 형태로 쓸 수 있다.
    * 함수형 `np.sum(배열)`  ·  메서드형 `배열.sum()`
* 축(axis) 설정 — 2차원 이상에서 어느 방향으로 집계할지 지정
    * `axis=0` : 세로로 눌러 **열별** 결과 (shape의 0번 축이 사라짐)
    * `axis=1` : 가로로 눌러 **행별** 결과 (shape의 1번 축이 사라짐)

```python
arr = np.array([[1,2,3],[4,5,6]])
print(arr)
np.sum(arr, axis=0)
```

```python
# --------------------
# 합계(sum) — 전체 / 열별 / 행별
# --------------------
print(arr)
print('----------')
```

```python
# --------------------
# 평균(mean)
# --------------------
print(arr)
print('----------')
```

```python
# --------------------
# 함수형 np.sum(arr) vs 메서드형 arr.sum() → 결과는 같음
# --------------------
```

> **함수형 vs 메서드형**
> * 둘 다 되는 함수(`sum`, `mean`, `min`, `max`, `std`, `argmax`, `cumsum` 등)는 **결과가 완전히 같으니** 편한 걸 쓰면 된다. (보통 짧은 메서드형 `arr.sum()` 을 많이 씀)
> * 단, **`np.함수()` 형태만 있는** 것도 있다 → `np.median()`, `np.unique()`, `np.percentile()` 은 `arr.median()` 처럼 메서드로는 쓸 수 없다.

```python
# --------------------
# 메서드형이 없는 함수는 np.함수() 로만 사용
# --------------------

arr = np.array([1, 3, 2, 3, 1])
```

```python
# --------------------
# 최소 / 최대 / 표준편차
# --------------------
print(arr)
```

```python
# --------------------
# 유일한 값(unique) — 중복 제거 후 정렬
# --------------------

arr2 = np.array([3, 1, 2, 3, 1, 1, 2])
```

```python
# --------------------
# 누적합계(cumsum)
# --------------------
arr = np.arange(1,7)
print(arr)
print('----------')
```

```python
# --------------------
# 누적곱(cumprod)
# --------------------
arr = np.arange(1,7)
print(arr)
print('----------')
```

```python
# --------------------
# 최소값의 인덱스(argmin) / 최대값의 인덱스(argmax)
# --------------------

np.random.seed(0)
arr = np.random.randint(1, 100, 10)
print(arr)
print('----------')
arr.argmin()
# arr.argmax()
```

### 미션) 점수 표준화 (z-score)
점수를 '평균 0, 표준편차 1' 기준으로 바꾸는 표준화를 배열 연산으로 구현하세요.
z = (값 − 평균) ÷ 표준편차

```python
# -------------------------------
# [미션] z = (arr - 평균) / 표준편차
# -------------------------------

arr = np.array([50, 60, 70, 80, 90])
(arr - arr.mean()) / arr.std()
```
