## 배열의 shape 제어

#### `배열객체.reshape()`
* `배열객체.reshape(shape)`
* size(전체 개수)만 맞으면 자유롭게 shape를 바꿀 수 있다.
* 원본은 그대로 두고 **모양을 바꾼 새 배열을 돌려준다.** (원본 불변)
* **`-1`** 을 적으면 나머지 크기에 맞게 개수가 자동 계산된다.

```python
arr = np.arange(1,7)
print(arr)

print('--- 3행 2열 ---')
# arr을 3행 2열로 바꾸기

print(arr.reshape(3,2))
print('--- 2행 3열 ---')
print(arr.reshape(2,3))
print('--- 6행 1열 ---')
print(arr.reshape(6,1))
```

```python
# -1 을 쓰면 나머지 크기에 맞게 자동 계산
arr = np.arange(1,13)   # 12개
print(arr)

print(arr.reshape(-1, 3))   # 열=3 → 행은 자동(4)
print(arr.reshape(2, -1))   # 행=2 → 열은 자동(6)
```

#### `배열객체.flatten()`
* 다차원 배열을 1차원으로 펴준다.

```python
arr = np.array([[1,2],[3,4],[5,6]])
print(arr)
print('-----flatten-----')

print(arr.flatten())   # [1 2 3 4 5 6]
```

#### `배열객체.transpose()`
* 행과 열을 뒤바꾼다(전치). `배열.T` 로도 쓸 수 있다.

```python
arr = np.array([[1,2,3],[4,5,6]])
print(arr)
print('-----전치-----')

print(arr.transpose())
print('-----전치(.T)-----')
print(arr.T)
```

#### `배열객체.resize()`
* `배열객체.resize(shape)` : 배열의 크기를 바꾼다.
    * 크기를 키우면 → 모자란 자리는 **0으로 채움**
    * 크기를 줄이면 → 넘치는 값은 **버림**

> **📌 reshape vs resize (헷갈리기 쉬움)**
> | 함수 | 원본 | 반환값 |
> |---|---|---|
> | `reshape` | **안 바뀜**(그대로) | 모양 바꾼 **새 배열을 돌려줌** → `b = a.reshape(...)` |
> | `resize` | **바뀜**(직접 수정) | 없음 → `a.resize(...)` 하면 a 자체가 변함 |

```python
arr = np.array([[0,1],[2,3]])
print(arr)
print('----- 3x3으로 resize (0으로 채움) -----')

arr.resize((3,3))
print(arr)
print('----- 2x1로 resize (넘치는 값 버림) -----')
arr.resize((2,1))
print(arr)
```

## 배열 연결
#### `np.concatenate()`
* `np.concatenate([배열들], axis=축)`
    * `axis=0` : 행 방향(위아래로) 연결
    * `axis=1` : 열 방향(좌우로) 연결

```python
arr1 = np.arange(1,5).reshape(2,2)
arr2 = np.arange(5,9).reshape(2,2)
arr3 = np.arange(9,13).reshape(2,2)
print(arr1); print('---'); print(arr2); print('---'); print(arr3)

print('--- axis=0 (위아래) ---')
print(np.concatenate([arr1,arr2,arr3], axis=0))
print('--- axis=1 (좌우) ---')
print(np.concatenate([arr1,arr2,arr3], axis=1))
```

#### `np.vstack()` / `np.hstack()`
* `np.vstack([배열들])` : 수직(vertical) 연결 = `concatenate(axis=0)`
* `np.hstack([배열들])` : 수평(horizontal) 연결 = `concatenate(axis=1)`
* axis를 외우기 어려우면 vstack/hstack이 더 직관적이다.

```python
arr1 = np.arange(1,5).reshape(2,2)
arr2 = np.arange(5,9).reshape(2,2)
# 수직(vstack) / 수평(hstack) 연결

print('--- vstack (수직) ---')
print(np.vstack([arr1,arr2]))
print('--- hstack (수평) ---')
print(np.hstack([arr1,arr2]))
```

## 배열 분할
#### `np.split()`
* `np.split(배열, 개수, axis)` : 같은 크기로 N등분
* `np.split(배열, [기준인덱스], axis)` : 지정한 인덱스에서 자름

```python
arr = np.arange(1,10)
print(arr)

print('--- 3개로 분할 ---')
# 3등분

a,b,c = np.split(arr, 3)
print(a); print(b); print(c)

print('--- 인덱스 3,5 기준 분할 ---')
a,b,c = np.split(arr, [3,5])
print(a); print(b); print(c)
```

```python
# 2차원 분할 (axis 지정)
arr = np.arange(1,13).reshape(4,3)
print(arr)

print('--- axis=0 으로 2등분 ---')

a,b = np.split(arr, 2, axis=0)
print(a); print(b)
print('--- axis=1 으로 3등분 ---')
a,b,c = np.split(arr, 3, axis=1)
print(a); print(b); print(c)
```

#### `np.vsplit()` / `np.hsplit()`
* `np.vsplit(배열, [인덱스])` : 수직 분할 (= split axis=0)
* `np.hsplit(배열, [인덱스])` : 수평 분할 (= split axis=1)

```python
arr = np.arange(1,13).reshape(4,3)
print(arr)
# 인덱스 1 기준으로 수직/수평 분할

print('--- vsplit [1] ---')
a,b = np.vsplit(arr,[1])
print(a); print(b)
print('--- hsplit [1] ---')
a,b = np.hsplit(arr,[1])
print(a); print(b)
```

## 🧩 실습 예제 · 미션
reshape · flatten · transpose · concat · split 을 실전처럼 써봅니다.

### 예제 1) 이미지를 1차원 벡터로 펴기
머신러닝 입력으로 넣으려고 2차원 이미지를 `flatten` 으로 1줄로 펴는 상황.

```python
# 3x3 흑백 이미지를 1차원 벡터로
img = np.arange(1,10).reshape(3,3)
print(img)

vec = img.flatten()
print('벡터:', vec, '| 길이:', vec.size)   # 길이 9
```

### 예제 2) 두 반 점수를 하나로 합치기
1반·2반 점수표(각 2명 × 3과목)를 `vstack` 으로 위아래로 이어붙이기.

```python
class1 = np.array([[80,90,70],[60,85,95]])
class2 = np.array([[75,80,100],[90,70,60]])
# 두 반을 세로로 합치기

allscore = np.vstack([class1, class2])
print(allscore)
print('shape:', allscore.shape)   # (4, 3)
```

### 미션) reshape 와 transpose 로 데이터 재배열
1. `np.arange(1,13)` 을 4행 3열로 만들기
2. 그걸 전치해서 3행 4열로 바꾸기

```python
# -------------------------------
# [미션] arange(1,13) → (4,3) → 전치 (3,4)
# -------------------------------

arr = np.arange(1,13).reshape(4,3)
print(arr)
print('--- 전치 ---')
print(arr.T)
print('shape:', arr.T.shape)   # (3, 4)
```
