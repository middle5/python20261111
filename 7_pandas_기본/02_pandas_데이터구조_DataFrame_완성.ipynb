# 02. 데이터 구조 - DataFrame

* **DataFrame** : 2차원 데이터 구조 (여러 개의 **Series(컬럼)** 를 묶은 표)
* **index** (행 라벨) + **columns** (열 라벨) 로 이루어짐
* DataFrame의 컬럼 하나를 꺼내면 → 그게 바로 **Series** (01에서 배운 것)

```python
import pandas as pd
```

## DataFrame 생성
* **`pd.DataFrame(data, index, columns)`**
    * `data` : Series, 리스트, 딕셔너리, 넘파이 배열 등
    * `index` : 행 라벨 (생략 시 0, 1, 2 …)
    * `columns` : 열 라벨 (생략 시 0, 1, 2 …)

### ① Series로 생성
* Series 하나 → 1개 컬럼짜리 DataFrame

```python
s = pd.Series([100, 50, 30, 10],
              index=['A', 'B', 'O', 'AB'],
              name='혈액형별 인원수')

df = pd.DataFrame(s)

display(df)
print('type :', type(df))
print('shape:', df.shape)
```

### ② 중첩 리스트로 생성
* 한 행(row)씩 데이터를 나열

```python
data = [['김철수', 45, '남', 175, 'A'],
        ['이영희', 43, '여', 168, 'B'],
        ['김선우', 13, '남', 160, 'A'],
        ['김지우', 10, '여', 140, 'O'],
        ['김현우',  7, '남', 125, 'AB']]

idx = ['아빠', '엄마', '첫째', '둘째', '셋째']
cols = ['이름', '나이', '성별', '키', '혈액형']

df = pd.DataFrame(data, index=idx, columns=cols)

display(df)
print('shape:', df.shape)
```

### ③ 딕셔너리로 생성  (가장 많이 사용)
* **키 = 컬럼명**, 값 = 컬럼 데이터(리스트)

```python
data = {'이름': ['김철수', '이영희', '김선우', '김지우', '김현우'],
        '나이': [45, 43, 13, 10, 7],
        '성별': ['남', '여', '남', '여', '남'],
        '키':  [175, 168, 160, 140, 125],
        '혈액형': ['A', 'B', 'A', 'O', 'AB']}

idx = ['아빠', '엄마', '첫째', '둘째', '셋째']

df = pd.DataFrame(data, index=idx)
df.index.name = '역할'

df
```

## 인덱스 다루기
* `df.set_index('컬럼')` : 특정 컬럼을 **행 인덱스**로 지정
* `df.reset_index()` : 인덱스를 다시 **컬럼**으로 되돌림 (RangeIndex로 초기화)

```python
# -----------------------------
# '이름' 컬럼을 인덱스로 지정
# -----------------------------
df.set_index('이름')
```

```python
# -----------------------------
# 인덱스(역할)를 다시 컬럼으로 되돌리기
#   - 기존 인덱스를 버리려면 reset_index(drop=True)
# -----------------------------
df.reset_index()
```

## 인덱스 · 컬럼 이름 변경
* `df.rename({현재라벨: 바꿀라벨, ...}, axis=0)` : **인덱스** 라벨 변경
* `df.rename({현재라벨: 바꿀라벨, ...}, axis=1)` : **컬럼** 라벨 변경

```python
# -----------------------------
# 인덱스 라벨 변경 (아빠→아버지, 엄마→어머니)
# -----------------------------
df.rename({'아빠': '아버지', '엄마': '어머니'}, axis=0)
```

```python
# -----------------------------
# 컬럼 라벨 변경 (나이→연령)
# -----------------------------
df.rename({'나이': '연령'}, axis=1)
```

> 💡 **전체 컬럼명을 한 번에** 새로 지정할 때는 `df.columns = [리스트]` 를 씁니다. (실무에서 자주 사용)
> * 리스트 길이가 **열 개수와 같아야** 합니다.
> * 일부만 바꿀 땐 위의 `rename` 을 사용합니다.

```python
# 컬럼명을 한 번에 전부 새로 지정 (개수가 열 개수와 같아야 함)
df.columns = ['성명', '연령', '성별', '신장', '혈액형']
df
```

> 💡 여러 컬럼을 계층으로 묶는 **멀티인덱스(MultiIndex)** 는
> 마지막 **교통사고 데이터 분석**에서 실제 데이터로 다룹니다.
> 지금은 "한 컬럼을 인덱스로 지정한다(set_index)" 까지만 알면 충분합니다.
