# 01. 데이터 구조 - Series

* **Series** : 1차원 데이터 구조 (라벨(인덱스)이 붙은 값들의 나열)
* 앞으로 배울 **DataFrame(표)의 컬럼 하나하나가 바로 Series** 입니다.
* 관용적으로 pandas는 별칭 **`pd`** 로 import 합니다.

```python
import pandas as pd
```

## Series 생성
* **`pd.Series(data, index, name, dtype)`**
    * `data` : 저장할 데이터 (리스트, 딕셔너리 등)
    * `index` : 각 값에 붙일 인덱스 라벨 (생략 시 0, 1, 2 …)
    * `name` : 시리즈의 이름 (DataFrame으로 갈 때 컬럼명이 됨)
    * `dtype` : 데이터 타입 (생략 시 자동 추론)

> 아래 세 가지 방법은 모두 **같은 혈액형 데이터**(A·B·O·AB = 100·50·30·10)를 만듭니다.
> 바뀌는 것은 오직 **'생성 방법'** 뿐입니다.

```python
# -----------------------------
# ① 리스트로 생성 → 인덱스 자동 (0, 1, 2, 3)
#    혈액형별 인원수 (A, B, O, AB 순서)
# -----------------------------
data = [100, 50, 30, 10]

s = pd.Series(data)
s
```

```python
# -----------------------------
# ② 같은 데이터 + 인덱스 라벨(A/B/O/AB) + 이름 지정
# -----------------------------
idx = ['A', 'B', 'O', 'AB']

s = pd.Series([100, 50, 30, 10],
              index=idx,
              name='혈액형')
s
```

```python
# -----------------------------
# ③ 딕셔너리로 생성 → 키가 곧 인덱스 라벨 (②와 같은 결과)
# -----------------------------
data = {'A': 100, 'B': 50, 'O': 30, 'AB': 10}

s = pd.Series(data, name='혈액형')
s
```

## Series의 주요 속성
* `values` : 값 (넘파이 배열)
* `index` : 인덱스 라벨
* `dtype` : 데이터 타입
* `name` : 시리즈 이름
* `shape` : 형태

```python
print('values :', s.values)
print('index  :', s.index)
print('dtype  :', s.dtype)
print('name   :', s.name)
print('shape  :', s.shape)
```

## Series의 주요 메서드
* `info()` : 시리즈 정보
* 집계함수 : `sum()`, `mean()`, `min()`, `max()`, `std()` 등
* `sort_values()` : 값으로 정렬 / `sort_index()` : 인덱스로 정렬
* `isnull()` / `notnull()` : 결측치 여부
* `astype()` : 자료형 변경

> `value_counts()`, `dropna()`, `fillna()` 등은 **03 기본**에서 데이터프레임과 함께 자세히 다룹니다.

```python
# -----------------------------
# info()
# -----------------------------
s.info()
```

```python
# -----------------------------
# 집계함수
# -----------------------------
print('합계 :', s.sum())
print('평균 :', s.mean())
print('최대 :', s.max())
```

```python
# -----------------------------
# 값으로 정렬 (내림차순은 ascending=False)
# -----------------------------
s.sort_values()
```

```python
# -----------------------------
# 인덱스 라벨로 정렬
# -----------------------------
s.sort_index()
```

```python
# -----------------------------
# 결측치 여부 (null이면 True) / 반대는 s.notnull()
# -----------------------------
s.isnull()
```

```python
# -----------------------------
# 자료형 변경 (정수 → 문자열)
# -----------------------------
s.astype('str')
```

## Series 데이터 선택
* 라벨로 선택 : **`s.loc[라벨]`** (라벨 슬라이스는 **끝 라벨 포함**)
* 정수 위치로 선택 : **`s.iloc[정수]`** (슬라이스는 끝 미포함, 파이썬 리스트와 동일)

```python
# -----------------------------
# A형은 몇 명인가요?  (라벨 / 정수위치)
# -----------------------------
print(s.loc['A'])   # 라벨로
print(s.iloc[0])    # 0번째 위치로
```

```python
# -----------------------------
# A형, O형은 각각 몇 명인가요?  (라벨 리스트)
# -----------------------------
s.loc[['A', 'O']]
```

```python
# -----------------------------
# A형부터 O형까지  (라벨 슬라이스 → 끝(O) 포함)
# -----------------------------
s.loc['A':'O']
```

## Series 데이터 추가 / 수정 / 삭제
* 추가·수정 : `s[인덱스] = 값` (인덱스가 없으면 추가, 있으면 수정)
* 삭제 : `del s[인덱스]`

```python
# -----------------------------
# 추가 → 수정
# -----------------------------
s['기타'] = 3     # 없는 인덱스 → 추가
s['기타'] = 5     # 있는 인덱스 → 수정
s
```

```python
# -----------------------------
# 삭제
# -----------------------------
del s['기타']
s
```

> 💡 **정리** : Series는 *라벨이 붙은 1차원 값들*입니다.
> 다음 노트북에서 이런 Series를 여러 개 묶어 **표(DataFrame)** 를 만듭니다.
> (= DataFrame의 각 컬럼이 곧 Series)
