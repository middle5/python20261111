# 자료형과 변수

# 자료형(data type)

## 파이썬의 자료형 종류

| 구분 | 설명 | 예 |
|---|---|---|
| 내장 자료형 | 파이썬에서 기본적으로 제공하는 핵심 자료형 | int, float, bool, str, list, tuple, range, dict, set, NoneType 등… |
| 라이브러리 자료형 | 내장 자료형을 확장하여 라이브러리에서 제공하는 자료형 | datetime.datetime, numpy.ndarray 등… |
| 사용자 정의 자료형 | 개발자가 직접 클래스를 만들어 정의한 자료형 | |

## 내장 자료형의 유형별 구분

| 유형 | 우리말 | 자료형 |
|---|---|---|
| Numeric Types | 숫자형 | int, float, complex |
| Boolean Type | 논리형 | bool |
| Sequence Types | 순서가 있는 시퀀스형 | str, list, tuple, range, bytes, bytearray, memoryview |
| Mapping Type | 키:값 매핑형 | dict |
| Set Types | 집합형 | set, frozenset |
| None Type | 값이 없음 | NoneType |

- 숫자형에는 정수를 다루는 int형, 실수를 다루는 float형, 복소수를 다루는 complex형이 있습니다.
- complex형은 과학계산이나 신호처리 분야에서 중요하게 사용되지만, 일반적인 프로그램에서는 잘 사용되지 않습니다.

## int

- integer(정수)의 약자
- 소수점 이하에 정보가 없는 숫자
- 양의 정수, 음의 정수, 0을 모두 포함 (ex. 100, -100, 0)
- 다양한 진법 지원 (10진수, 2진수, 8진수, 16진수)
- 메모리가 허용하는 한 크기 제한이 없음 → 매우 큰 수를 표현할 수 있음

**활용 예시** : 조회수, 상품가격, 손실금액

````python
views = 12563
price = 10000
loss = -300000
print(type(views), type(price), type(loss))
````

실행 결과
```
<class 'int'> <class 'int'> <class 'int'>
```

`type()`은 값의 자료형을 알려 주는 함수입니다. 자세한 내용은 뒤의 「변수 특징② - 동적 타이핑」에서 다시 봅니다.

## float

- 소수점을 포함한 실수 형태의 자료형
- 소수점 표기법 (ex. -1.1, 0.5, 3.14)
- 지수 표기법 (ex. 1.5e10, 1.67e-27)
- 크기 제한이 있음 (IEEE 754 표준의 64비트 배정밀도 형식)
- 정밀도에 한계가 있음 (약 15~17자리)
  - 부동소수점 연산 시 미세한 오차 발생 가능
  - 정밀한 연산이 필요한 경우, decimal 모듈 등 사용 고려
- 성능 측면에서 장점이 있음 (계산 속도가 빠름)

**활용 예시** : 환율, 온도, 대기중 이산화탄소 농도 등

````python
exchange_rate = 1450.5
temperature = 25.7
co2_ppm = 4.16e-4
print(type(exchange_rate), co2_ppm)
````

실행 결과
```
<class 'float'> 0.000416
```

`4.16e-4`는 지수 표기법으로, 4.16 × 10⁻⁴ = 0.000416입니다.

## bool

- 참과 거짓을 표현하는 논리형 데이터 타입
  - 조건문, 반복문 등 프로그램의 흐름 제어를 위한 핵심
- 값의 종류 : `True`(참), `False`(거짓) — 첫 글자는 대문자입니다.
- int의 하위 클래스
  - `True` → 1
  - `False` → 0

````python
is_member = True
print(type(is_member))
print(True + True)
````

실행 결과
```
<class 'bool'>
2
```

`True`는 1처럼 동작하므로 `True + True`는 2입니다.

## str

- 연결된 문자들을 따옴표(`"` 또는 `'`)로 묶어 표현하는 자료형
- 삼중 따옴표(`"""` 또는 `'''`)를 사용하여 여러 줄의 문자열 표현 가능

**활용 예시** : 컨텐츠 제목, 댓글 내용, 전화번호, 우편번호

전화번호나 우편번호처럼 숫자로만 이루어져 있어도 문자열로 다루는 것이 적합한 경우가 많습니다. 우편번호 `04515`를 숫자로 저장하면 앞의 0이 사라지기 때문입니다.

````python
title = "파이썬 기초문법"
review = "좋은 내용 감사합니다"
phone = "0101234567"
postal_code = "04515"
print(type(phone), postal_code)
````

실행 결과
```
<class 'str'> 04515
```

## index

시퀀스 자료형에는 index가 있습니다. index를 이용하여 데이터의 개별 요소에 접근합니다.

| index | 0 | 1 | 2 | 3 | 4 |
|---|---|---|---|---|---|
| 값 | h | e | l | l | o |
| 음수 index | -5 | -4 | -3 | -2 | -1 |

- index는 0부터 시작합니다.
- 뒤에서부터 역순으로 접근할 때는 마이너스 index를 사용합니다. 마이너스 index는 -1부터 시작합니다.

````python
word = "hello"
print(word[0], word[4], word[-1])
````

실행 결과
```
h o o
```

## list

- 여러 아이템을 대괄호`[]`로 묶어서 다루는 자료형
- 아이템을 변경할 수 있음

**활용 예시** : 처리해야 할 작업 목록

````python
todo_list = ["이메일 확인", "회의 준비", "보고서 작성"]
print(todo_list[0], todo_list[-1])
````

실행 결과
```
이메일 확인 보고서 작성
```

| index | 0 | 1 | 2 |
|---|---|---|---|
| 값 | "이메일 확인" | "회의 준비" | "보고서 작성" |
| 음수 index | -3 | -2 | -1 |

list에는 index가 있습니다. index를 이용하여 데이터의 개별 요소에 접근합니다. 리스트는 개별 아이템의 내용을 수정할 수 있습니다.

````python
todo_list[2] = "회의록 작성"   # 2번 아이템 바꾸기
print(todo_list)
````

실행 결과
```
['이메일 확인', '회의 준비', '회의록 작성']
```

## tuple

- 여러 데이터를 소괄호`()`로 묶어서 다루는 자료형
- 내용을 변경할 수 없음

**활용 예시** : 설정값, 고정된 데이터

````python
background_color = (255, 30, 24)
seoul_location = (37.5665, 126.9780)
print(seoul_location[0], seoul_location[-1])
````

실행 결과
```
37.5665 126.978
```

| index | 0 | 1 |
|---|---|---|
| 값 | 37.5665 | 126.9780 |
| 음수 index | -2 | -1 |

tuple에는 index가 있습니다. index를 이용하여 데이터의 개별 요소에 접근합니다.

## range

- 일정한 간격의 정수 시퀀스를 표현하는 자료형
- 일정한 패턴의 정수 목록이 필요할 때 사용
- 시작값, 종료값, 스텝 크기만 저장하여 메모리 효율적

**예시** : 1부터 99까지 홀수 시퀀스 생성 (1, 3, 5, …, 99) — 시작값 1, 종료값 100, 스텝 2

````python
odd_numbers = range(1,100,2)
print(odd_numbers[0], odd_numbers[1], odd_numbers[-1])
````

실행 결과
```
1 3 99
```

| index | 0 | 1 | 2 | 3 | 4 | … |
|---|---|---|---|---|---|---|
| 값 | 1 | 3 | 5 | 7 | 9 | … |

range에는 index가 있습니다. index를 이용하여 데이터의 개별 요소에 접근합니다. 종료값 100은 포함되지 않습니다.

## dict

- dictionary(사전)의 줄임말 — 사전의 "단어:정의"처럼 "키:값"을 짝지어 저장
- 키:값 매핑하여 데이터를 저장, 키를 이용하여 값에 접근
- 여러 속성을 가진 데이터를 구조화 할 때 사용

**활용 예시** : 사용자 프로필

````python
user_profile = {"name" : "James",
                "email" : "abc@google.com",
                "age" : 20}
print(user_profile["name"], user_profile["age"])
````

실행 결과
```
James 20
```

## set

- 중복된 요소를 허용하지 않음
- 순서가 없음
- 집합 연산 기능을 제공 (합집합, 교집합, 차집합)

**활용 예시** : 두 포스트의 모든 태그를 중복 없이 추출

````python
post1_tags = {"파이썬", "프로그래밍", "데이터분석"}
post2_tags = {"파이썬", "웹개발", "프레임워크"}
post1_tags | post2_tags
````

실행 결과 (순서는 실행할 때마다 다를 수 있습니다)
```
{'파이썬', '프로그래밍', '데이터분석', '웹개발', '프레임워크'}
```

`|`는 합집합 연산자입니다. 두 포스트에 모두 있는 `"파이썬"`은 한 번만 나옵니다.

## None Type

- `None`이라는 단 하나의 값만 존재
- `None`은 값이 없음을 나타냄

**활용 예시**
- 함수의 매개변수 기본값으로 설정
- 함수에서 명시적인 반환 값이 없을 때 암묵적 반환
- 함수의 처리과정 중 특정 조건에서 명시적으로 반환
- 변수에 아직 유효한 값이 할당되지 않음을 표시

````python
result = print('hello')   # print()는 반환 값이 없는 함수
print(result)
````

실행 결과
```
hello
None
```

`print()`는 화면에 출력만 하고 값을 반환하지 않습니다. 그래서 변수 `result`에는 `None`이 담깁니다.

## 내장 자료형 : 기본 자료형과 컨테이너 자료형

| 구분 | 설명 | 자료형 |

| 컨테이너 자료형(자료구조) | 여러 개의 데이터를 묶어서 하나의 변수에 담아 사용한다. | list, tuple, range, dict, set |

## 자료형 정리

| 기본 자료형 | 설명 | 컨테이너 자료형(자료구조) | 설명 |
|---|---|---|---|
| int 정수형 | 소수점이 없는 숫자<br>(예) -100, 0, 245 | list 리스트 | 여러 개의 자료들의 묶음<br>(예) ['python', 'c언어', 'java'] |
| float 실수형 | 소수점이 있는 숫자<br>(예) -55.0, 0.0, 3.14 | tuple 튜플 | 여러 개의 자료들의 묶음(읽기전용)<br>(예) ('python', 'c언어', 'java') |
| str 문자열형 | 문자들이 연결된 값. 따옴표로 묶어줍니다.<br>※ 쌍따옴표(""), 홑따옴표('') 모두 가능 | dict 딕셔너리 | key:value의 쌍으로 이루어진 자료들의 묶음<br>(예) {'국어':100, '영어':95, '수학':90} |
| bool 부울형 | 참, 거짓을 구분하는 자료형<br>True, False 두가지 값만 가져요. | set 집합 | 중복을 제거한 자료들의 집합<br>(예) {'사과', '딸기', '포도'} |

## 자료형 변환

`input()`함수는 실행 결과를 무조건 **'문자열'로 반환**합니다.

아래 셀은 **오류가 나는 것을 확인하는 셀**입니다. 나이를 입력하면 셋째 줄에서 `TypeError`가 납니다.

````python
# [오류 확인] 오류가 나는 것이 정상입니다.
age = input("나이를 입력하세요 >>> ")
print("나이는 ", age, "살이시군요.", sep="")
print("내년이면 ", age+1, "살이 되시겠네요.", sep="")
````

오류 메시지
```
TypeError: can only concatenate str (not "int") to str
```

`age`는 str(문자열), `1`은 int(정수)입니다. str(문자열)과 int(정수)를 연결할 수 없어 에러가 발생합니다.

## 자료형 변환 : int()

`int()` 함수를 이용하여 문자열을 정수로 변환합니다.

1. 사용자가 입력한 값을 str(문자열)로 반환
2. `int()`가 int(정수)형으로 자료형 변환하여 반환
3. int(정수)형으로 변환된 값을 변수 `age`에 할당

````python
age = int(input("나이를 입력하세요 >>> "))
print("나이는 ", age, "살이시군요.", sep="")
print("내년이면 ", age+1, "살이 되시겠네요.", sep="")
````

실행 결과 (20 입력 시)
```
나이는 20살이시군요.
내년이면 21살이 되시겠네요.
```

| 함수 | 변환 | 함수 | 변환 |

| `float()` | 실수형으로 변환 | `tuple()` | 튜플형으로 변환 |
| `str()` | 문자열형으로 변환 | `dict()` | 딕셔너리형으로 변환 |
| `bool()` | 논리형으로 변환 | `set()` | 집합형으로 변환 |

````python
print(int("100") + 1)
print(float("3.5"))
print(str(2026) + "년")
print(list(range(1, 6)))
````

실행 결과
```
101
3.5
2026년
[1, 2, 3, 4, 5]
```

# 변수

## 변수란?

- 변수란, 프로그램에서 사용하는 자료(데이터)를 저장하는 공간입니다.
- 먼저 변수명을 정하고, 할당연산자(`=`)를 사용하여 변수에 데이터를 할당합니다.

```
변수 = 데이터
```

**변수와 데이터베이스**

| 구분 | 저장하는 데이터 |
|---|---|
| 변수(Variable) | 프로그램이 실행되는 동안 필요한 데이터 (메모리에 저장) |
| DataBase | 영구 저장하는 데이터 (하드디스크·SSD에 저장) |

영구 저장이 필요한 데이터는 데이터베이스에 보관합니다. 프로그램이 실행되는 동안에는 필요한 데이터를 데이터베이스에서 불러와 메모리에 두고 다루는데, 이때 변수가 사용됩니다.

## 변수명 작성 규칙 : 필수 규칙 (위반시 즉시 에러 발생)

1. 변수 이름은 알파벳, 숫자, 언더스코어(`_`)만으로 구성 가능
   - 📌 한글도 사용할 수 있지만 실무적으로 권장하지 않습니다.
2. 첫 글자는 알파벳이나 언더스코어(`_`)로 시작 (숫자는 첫 글자로 쓸 수 없다)
3. 파이썬 문법에서 사용하는 예약어는 변수명으로 사용 불가
   - 예약어 : `if`, `for`, `while` 등

아래 셀은 **오류가 나는 것을 확인하는 셀**입니다. 숫자로 시작하는 변수명을 쓰면 `SyntaxError`가 납니다.

````python
# [오류 확인] 오류가 나는 것이 정상입니다.
1st_name = "김철수"
````

## 변수명 작성 규칙 : 주의사항 (잠재적 문제 발생 가능)

- 변수명은 대문자와 소문자를 구분합니다. (`Age`와 `age`는 다른 변수)
- 함수명을 변수명으로 사용하지 않도록 주의합니다. (예: `print = 10`처럼 쓰면 그 뒤로 `print()` 함수를 쓸 수 없습니다.)

````python
age = 20
Age = 30
print(age, Age)
````

실행 결과
```
20 30
```

## 변수명 작성 규칙 : 관행 (좋은 코드를 만들기 위한 약속)

1. 영문 소문자로 시작
2. 두 개 이상의 단어 조합인 경우 스네이크 케이스 사용
3. 의미를 잘 나타내는 이름 사용
   - `age = 20` (good)
   - `a = 20` (bad)

| 표기법 | 설명 | 예 |
|---|---|---|

## 변수 특징② - 동적 타이핑

- 변수에는 어떠한 type의 자료도 담을 수 있으며, 변수에 값이 할당될 때 할당되는 값의 type에 따라 변수의 type이 정해집니다.
- `type()` : 자료형을 리턴 해주는 내장함수

````python
a = 10
print(type(a))
a = 3.14
print(type(a))
a = '김철수'
print(type(a))
a = True
print(type(a))
````

실행 결과
```
<class 'int'>
<class 'float'>
<class 'str'>
<class 'bool'>
```

````python
a = [1,2,3]
print(type(a))
a = (1,2,3)
print(type(a))
a = {'name':'김철수'}
print(type(a))
a = {'김철수','홍길동'}
print(type(a))
````

실행 결과
```
<class 'list'>
<class 'tuple'>
<class 'dict'>
<class 'set'>
```

### [응용문제 1] 신규 고객 정보 정리 (영업)

신규 고객의 이름, 나이, 연락처를 입력받아 딕셔너리 `customer`에 담고, 딕셔너리 전체와 자료형을 출력하세요.

- 나이는 숫자로 저장합니다. (`int()` 사용)
- 연락처는 문자열로 저장합니다. (`010`의 0이 사라지면 안 되기 때문입니다.)

실행 결과 (김하늘, 29, 01012345678 입력 시)
```
이름: 김하늘
나이: 29
연락처: 01012345678
{'이름': '김하늘', '나이': 29, '연락처': '01012345678'}
<class 'int'> <class 'str'>
```
