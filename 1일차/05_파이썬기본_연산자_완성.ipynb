# 연산자

## 연산자

| 연산자 종류 | 연산자 |
|---|---|
| 산술연산 | `+` `-` `*` `/` `//` `%` `**` |
| 비교연산 | `>` `<` `>=` `<=` `==` `!=` |
| 논리연산 | `and` `or` `not` |
| 할당연산 | `=` `+=` `-=` `*=` `/=` `%=` `//=` `**=` |

## 산술연산

**Python 산술연산자의 종류**

| 연산 | 연산자 |
|---|---|
| 더하기 | `+` |
| 빼기 | `-` |
| 곱하기 | `*` |
| 나누기 | `/` |
| 몫 | `//` |
| 나머지 | `%` |
| 거듭제곱 | `**` |

**산술연산의 우선순위**

- 수학에서와 동일한 연산의 우선순위 규칙을 가집니다.
- 괄호 안이 가장 먼저 계산되고, 제곱 → 곱셈과 나눗셈 → 덧셈과 뺄셈의 순서로 계산됩니다.
- ※ 단, 파이썬에서는 중괄호나 대괄호를 사용하지 않고, 소괄호만 사용합니다.

````python
print('10+3=',10+3)
print('10-3=',10-3)
print('10*3=',10*3)
print('10/3=',10/3)
print('10//3=',10//3)
print('10%3=',10%3)
print('10**3=',10**3)
````

실행 결과
```
10+3= 13
10-3= 7
10*3= 30
10/3= 3.3333333333333335
10//3= 3
10%3= 1
10**3= 1000
```

- `/`는 결과가 항상 float(실수)입니다.
- `//`는 나눈 몫, `%`는 나눈 나머지입니다.

## 비교연산

값을 비교합니다. 결과는 부울값(True/False)으로 계산됩니다.

| 비교연산자 | 사용 | 비고 |
|---|---|---|
| `==` | `x == y` | 등호 2개로 표현한다는 점에 유의 (`=`는 할당연산자이다.) |
| `!=` | `x != y` | 느낌표를 먼저 쓴다는 점에 유의 (`=!`로 쓰지 않는다.) |
| `>` | `x > y` | |
| `<` | `x < y` | |
| `>=` | `x >= y` | 부등호를 먼저 쓴다는 점에 유의 (`=>`로 쓰지 않는다.) |
| `<=` | `x <= y` | 부등호를 먼저 쓴다는 점에 유의 (`=<`로 쓰지 않는다.) |

````python
print('1==1', 1==1)
print('1!=1', 1!=1)
print('1>1', 1>1)
print('1<1', 1<1)
print('1>=1', 1>=1)
print('1<=1', 1<=1)
````

실행 결과
```
1==1 True
1!=1 False
1>1 False
1<1 False
1>=1 True
1<=1 True
```

## 논리연산

두 개 이상의 비교연산을 논리적으로 결합합니다. 결과는 bool입니다.

| 논리연산자 | 기능 |
|---|---|
| `and` | 그리고 (모두 True일 때만 True) |
| `or` | 또는 (하나 이상이 True이면 True) |
| `not` | ~가 아니다 (결과를 반대로 출력) |

**진리표**

| a | b | a and b | a or b |
|---|---|---|---|
| True | True | True | True |
| True | False | False | True |
| False | True | False | True |
| False | False | False | False |

| a | not a |
|---|---|
| True | False |
| False | True |

````python
print(True and False)
print(True or False)
print(not True)
````

실행 결과
```
False
True
False
```

## 할당연산

`+=`, `-=` 같은 할당연산자는 변수의 값을 계산한 뒤 같은 변수에 다시 담습니다.

| 할당연산 | 같은 뜻 |
|---|---|
| `x += 3` | `x = x + 3` |
| `x -= 3` | `x = x - 3` |
| `x *= 3` | `x = x * 3` |
| `x /= 3` | `x = x / 3` |
| `x //= 3` | `x = x // 3` |
| `x %= 3` | `x = x % 3` |
| `x **= 3` | `x = x ** 3` |

````python
stock = 100     # 재고 100개
stock -= 30     # 30개 출고
stock += 50     # 50개 입고
print(stock)
````

실행 결과
```
120
```

### [응용문제 1] 견적서 금액 계산 (영업)

단가, 수량, 할인율(%)을 입력받아 공급가액, 할인 금액, 할인 후 금액을 출력하세요. 금액은 모두 정수로 만들고 천 단위 쉼표를 넣습니다.

- 공급가액 = 단가 × 수량
- 할인 금액 = 공급가액 × 할인율 / 100
- 할인 후 금액 = 공급가액 - 할인 금액

실행 결과 (35000, 40, 15 입력 시)
```
단가: 35000
수량: 40
할인율(%): 15
공급가액: 1,400,000원
할인 금액: 210,000원
할인 후 금액: 1,190,000원
```

````python
price = int(input('단가: '))
qty = int(input('수량: '))
rate = int(input('할인율(%): '))
supply = price * qty
discount = int(supply * rate / 100)
print(f'공급가액: {supply:,}원')
print(f'할인 금액: {discount:,}원')
print(f'할인 후 금액: {supply - discount:,}원')
````

### [응용문제 2] 캠페인 성과 판단 (마케팅)

광고비, 매출, 신규 회원 수를 입력받아 아래 두 가지를 출력하세요.

- ROAS(광고비 대비 매출, %) = 매출 / 광고비 × 100 — 소수점 첫째 자리까지 반올림
- 캠페인 성공 여부 : ROAS가 300% 이상**이고** 신규 회원이 100명 이상이면 True

실행 결과 (2000000, 7300000, 85 입력 시)
```
광고비: 2000000
매출: 7300000
신규 회원 수: 85
ROAS: 365.0 %
캠페인 성공: False
```

````python
cost = int(input('광고비: '))
revenue = int(input('매출: '))
members = int(input('신규 회원 수: '))
roas = round(revenue / cost * 100, 1)
print('ROAS:', roas, '%')
print('캠페인 성공:', roas >= 300 and members >= 100)
````

### [응용문제 3] 연차 사용 가능 여부 (일반 업무)

남은 연차 일수와 신청할 일수를 입력받아 아래를 출력하세요.

- 신청 후 남는 연차 일수 (`-=` 사용)
- 신청 가능 여부 : 신청 일수가 1일 이상**이고**, 신청 후 남는 일수가 0일 이상이면 True

실행 결과 (5, 3 입력 시)
```
남은 연차: 5
신청 일수: 3
신청 후 남는 연차: 2
신청 가능: True
```

````python
left = int(input('남은 연차: '))
days = int(input('신청 일수: '))
left -= days
print('신청 후 남는 연차:', left)
print('신청 가능:', days >= 1 and left >= 0)
````
