## f스트링

- f-string(formatted string literal)은, 문자열을 보다 간결하게 포매팅(Formatting)하기 위해 사용되는 문법입니다.
- 파이썬 3.6부터 도입되었으며, 문자열 앞에 `f`를 붙여 사용합니다.
- f-string을 사용하면, 문자열 안에 변수나 표현식을 중괄호`{}`로 감싸 바로 삽입할 수 있습니다.

## 슬라이싱 : 문자열의 일부 꺼내기

다음 미션에서 쓰는 문법입니다. 인덱스 하나 대신 `[시작:끝]`을 쓰면 여러 글자를 한 번에 꺼냅니다.

| 쓰는 법 | 뜻 | `'python'`에 쓰면 |
|---|---|---|
| `문자열[시작:끝]` | 시작 인덱스부터 끝 인덱스 **바로 앞**까지 | `[1:4]` → `'yth'` |
| `문자열[:끝]` | 처음부터 끝 인덱스 바로 앞까지 | `[:2]` → `'py'` |
| `문자열[시작:]` | 시작 인덱스부터 마지막까지 | `[-2:]` → `'on'` |

**[직접 입력]** 아래 코드를 다음 셀에 입력하고 실행합니다.

```python
word = 'python'
print(word[1:4], word[:2], word[-2:])
```

````python
# [직접 입력] 위 코드를 입력하세요.

word = 'python'

# py
word[0:2] # === word[:2]

# on
word[4:6] # === word[4:] === word[-2:]
````

실행 결과
```
yth py on
```

## [미션] 비밀번호 만들기

1. 영문 이름을 입력 받는다.

3. 마지막 2글자를 추출한다.
4. ②와 ③을 연결하여 비밀번호를 생성한다.
5. 비밀번호를 출력한다.

실행 결과 (python 입력 시)
```
영문이름을 입력하세요(4글자 이상):python

```

## 문자열 메소드

문자열 메소드는 `문자열.메소드()` 형태로 호출합니다. 모든 메소드를 외울 필요는 없고, 필요할 때 이 표에서 찾아 씁니다.

| 메소드 | 기능 | 메소드 | 기능 |
|---|---|---|---|
| capitalize | 문장의 첫 글자를 대문자로 바꾸기 | startswith | 문자열이 특정 문자로 시작하는지 여부 |
| title | 문장 속 각 단어의 첫 문자를 대문자로 바꾸기 | endswith | 문자열이 특정 문자로 끝나는지 여부 |
| upper | 모든 문자를 대문자로 바꾸기 | isalpha | 문자열의 모든 문자가 알파벳/한글이면 True |
| lower | 모든 문자를 소문자로 바꾸기 | isalnum | 문자열의 모든 문자가 알파벳/한글/숫자이면 True |
| casefold | 영문을 소문자로 바꾸기 | isascii | 문자열의 모든 문자가 ASCII문자이면 True |
| swapcase | 소문자는 대문자로, 대문자는 소문자로 바꾸기 | isdecimal | 문자열이 숫자로 이루어져 있으면 True |
| expandtabs | 문장의 tab 간격 지정 | isdigit | 문자열이 숫자로 이루어져 있으면 True |
| find | 찾을 문자열의 첫번째 인덱스 (문자가 없으면 -1) | isnumeric | 문자열이 숫자로 이루어져 있으면 True |
| rfind | 찾을 문자열의 마지막 인덱스 (문자가 없으면 -1) | isidentifier | 문자열이 식별자로 사용될 수 있으면 True |
| index | 찾을 문자의 첫번째 인덱스 (문자가 없으면 ValueError) | islower | 문자열이 소문자로만 이루어져 있으면 True |
| rindex | 찾을 문자의 마지막 인덱스 (문자가 없으면 ValueError) | isupper | 문자열이 대문자로만 이루어져 있으면 True |
| count | 문자열에서 특정 문자의 개수 세기 | isprintable | 문자열이 출력 가능한 문자이면 True |

| 메소드 | 기능 | 메소드 | 기능 |
|---|---|---|---|
| isspace | 문자열이 공백으로만 이루어져 있으면 True | rjust | 전체 자릿수를 지정하여 오른쪽 정렬하기 |
| istitle | 문자열의 각 단어가 대문자로 시작하면 True | ljust | 전체 자릿수를 지정하여 왼쪽 정렬하기 |

| rsplit | 특정 문자 기준으로 문자열을 나누어 오른쪽부터 리스트로 만들기 | translate | 문자열 치환(maketrans와 함께 사용) |
