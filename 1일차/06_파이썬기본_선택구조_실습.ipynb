# 선택구조

> **실습 파일 사용법**
>
> - **[직접 입력]** 표시가 있는 셀: 바로 위 마크다운 셀의 코드를 보고 직접 입력한 뒤 실행합니다.
> - **[직접 작성]** 표시가 있는 셀: 문제를 읽고 코드를 스스로 작성합니다. 정답은 완성 파일에 있습니다.
> - **[실행만]** 표시가 있는 셀: 코드를 입력하지 않고 그대로 실행합니다.

## 프로그램의 3가지 기본 제어 구조

| 구조 | 설명 |
|---|---|
| 순차구조 | 명령들이 순차적으로 실행되는 구조 |
| 선택구조 | 조건에 따라 명령을 선택하여 실행되는 구조 (조건이 True이면 한쪽, False이면 다른 쪽 실행) |
| 반복구조 | 동일한 명령이 반복되면서 실행되는 구조 |

## 흐름 제어

| 흐름제어 | 키워드 |
|---|---|
| 조건 | `if` `elif` `else` |
| 반복 | `for` `while` |
| 반복제어 | `break` `continue` |

이 노트북 파일에서는 조건(`if`, `elif`, `else`)을 다룹니다.

## 선택구조 if

**[ 파이썬 문법 ]**

```python
if 조건식:
    코드블록
```

- 조건식 뒤에는 콜론(`:`)을 붙입니다.
- 조건식의 결과가 True일 때 실행할 코드들은 들여쓰기 합니다.
- 조건식의 결과가 True이면 실행할 코드블록을 실행합니다.
- 조건식의 결과가 False이면 조건문을 빠져나옵니다.

**[직접 입력]** 아래 코드를 다음 셀에 입력하고 실행합니다.

```python
# 60점 이상이면 합격
score = 60
if score>=60:
    print('합격입니다.')
    print('축하합니다.')
print('수고하셨습니다.')
```

````python
# [직접 입력] 위 코드를 입력하세요.

score = 60

if score >= 60:
    print('합격입니다.', '축하합니다.', sep='\n')
print('수고하셨습니다.')
````

실행 결과
```
합격입니다.
축하합니다.
수고하셨습니다.
```

`score`를 50으로 바꿔 실행해 보세요. 들여쓰기 한 두 줄은 실행되지 않고, 마지막 줄 `수고하셨습니다.`만 출력됩니다. 마지막 줄은 들여쓰기 하지 않아서 if문 밖에 있기 때문입니다.

## 선택구조 if~else~

**[ 파이썬 문법 ]**

```python
if 조건식:
    코드블록1
else:
    코드블록2
```

- 조건식의 결과가 True이면 코드블록1을 실행하고 조건문을 빠져나옵니다.
- 조건식의 결과가 False이면 코드블록2를 실행하고 조건문을 빠져나옵니다.

**[직접 입력]** 아래 코드를 다음 셀에 입력하고 실행합니다.

```python
# 60점 이상이면 합격, 아니면 불합격
score = 30
if score>=60:
    print('합격입니다.')
    print('축하합니다.')
else:
    print('불합격입니다')
    print('안타깝군요.')
print('수고하셨습니다.')
```

````python
# [직접 입력] 위 코드를 입력하세요.

# [직접 입력] 위 코드를 입력하세요.

score = 55

if score >= 60:
    print('합격입니다.', '축하합니다.', sep='\n')
else:
    print('불합격입니다.', '안타깝군요.', sep='\n')
print('수고하셨습니다.')
````

실행 결과
```
불합격입니다
안타깝군요.
수고하셨습니다.
```

## 선택구조 if~elif~else~

조건을 위에서부터 차례로 검사하다가, 처음으로 True가 나온 곳의 코드블록만 실행합니다.

| 순서 | 조건 | True이면 |
|---|---|---|
| 1 | 점수 >= 90 | A |
| 2 | 점수 >= 80 | B |
| 3 | 점수 >= 70 | C |
| 4 | 점수 >= 60 | D |
| 5 | 모두 False | F |

**[직접 입력]** 아래 코드를 다음 셀에 입력하고 실행합니다.

```python
# 점수를 입력받아 A,B,C,D,F 등급 출력하기

score = int(input("점수:"))
if score >= 90:
  print("A")
elif score >=80:
  print("B")
elif score >=70:
  print("C")
elif score >= 60:
  print("D")
else:
  print("F")
```

````python
# [직접 입력] 위 코드를 입력하세요.

score = int(input('점수 :'))
if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
elif score >=  60:
    print("D")
else:
    print("F")
````

실행 결과 (85 입력 시)
```
점수:85
B
```

**순서를 바꾸면?** `score >= 60`을 맨 위에 두면, 85점도 첫 조건에서 True가 되어 `D`가 출력됩니다. 범위로 나누는 조건은 좁은 범위(높은 점수)부터 검사해야 합니다. 아래 「무엇이 잘못됐을까요?」 연습문제에서 같은 문제를 다시 봅니다.

| 구매 금액 | 할인율 |
