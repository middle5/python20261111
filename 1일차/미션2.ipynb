### 어려운 미션: 견적서 자동 작성 (영업·일반 업무)

거래처명과 주문 정보를 입력받아 견적서를 출력하세요. 주문 정보는 `제품코드-수량` 형식으로 한 줄에 입력받습니다.

**조건**

| 제품코드 | 품목 | 단가 |
| --- | --- | --- |
| A | 노트 | 1,500원 |
| B | 볼펜 | 800원 |
| C | 클리어파일 | 2,000원 |
- 제품코드는 소문자로 입력해도 인식합니다. 예를 들어 `b-120`도 받아들입니다.
- 위 표에 없는 제품코드를 입력하면 "등록되지 않은 제품 코드입니다."만 출력합니다.
- 수량 할인: 100개 이상이면 10%, 50개 이상이면 5%, 그 밖에는 0%입니다.
- 공급가액은 단가 × 수량에서 할인을 적용한 금액입니다. 부가세는 공급가액의 10%입니다. 두 금액 모두 정수로 계산합니다.
- 합계가 1,000,000원 이상이거나 거래처명이 `(주)`로 시작하면 결제 조건은 "월말 정산"입니다. 그 밖에는 "선결제"입니다.

**실행 예시 1**

```
거래처명: (주)한빛상사
주문 정보(제품코드-수량): b-120
================================
견적서 - (주)한빛상사 귀하
================================
품목: 볼펜 (B)
수량: 120개 (할인 10%)
공급가액: 86,400원
부가세: 8,640원
합계: 95,040원
결제 조건: 월말 정산
================================
```

**실행 예시 2**

```
거래처명: 새솔카페
주문 정보(제품코드-수량): D-30
등록되지 않은 제품 코드입니다.
```

**힌트**

- `split('-')`으로 제품코드와 수량을 나눕니다.
- 대문자 변환은 `upper()`, 시작 문자 확인은 `startswith()`를 씁니다.
- 제품코드를 찾지 못했을 때 단가를 0으로 정해 두면, 오류 안내와 견적서 출력을 `if ~ else`로 나눌 수 있습니다. 이때 `if` 안에 다시 `if`를 넣어도 됩니다.

````python
# 1. 입력 : 거래처명, 주문정보(형식 : 제품코드-수량)
# client = input('거래처명 :')
client = "새솔카페"
# order = input('주문 정보(제품코드-수량) :')
order = "c-1200"
product_code = order[:order.index('-')].upper()
qty = int(order[order.index('-')+1:])


# 2. 처리
if product_code == "A":
    item = "노트 (A)"
    price = 1500
elif product_code == "B":
    item = "볼펜 (B)"
    price = 800
elif product_code == "C":
    item = "클리어파일 (C)"
    price = 2000
else:
    print("등록되지 않은 제품 코드입니다.")

if qty >= 100:
    sale = "10%"
    supply = price * qty * 0.9
elif qty >= 50:
    sale = "5%"
    supply = price * qty * 0.95
else:
    sale = "0%"
    supply = price * qty

# client.startswith('(주)')
if int(supply * 1.1) >= 1000000 or client.startswith('(주)'):
    terms = "월말 정산"
else:
    terms = "선결제"


# 3. 출력
print(f"""
        {'='*25}
        견적서 - {client} 귀하
        {'='*25}
        품목 :{item}
        수량 :{qty}개 (할인 {sale})
        공급가액 :{int(supply):>8,}원
        부가세 :{int(supply * 0.1):>9,}원
        합계 :{int(supply * 1.1):>11,}원
        결제 조건 :{terms}
        {'='*25}
        """)
````
