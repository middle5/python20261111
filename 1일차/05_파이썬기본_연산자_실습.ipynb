## 연산자

| 연산자 종류 | 연산자 |
|---|---|
| 산술연산 | `+` `-` `*` `/` `//` `%` `**` |
| 비교연산 | `>` `<` `>=` `<=` `==` `!=` |
| 논리연산 | `and` `or` `not` |
| 할당연산 | `=` `+=` `-=` `*=` `/=` `%=` `//=` `**=` |

## 산술연산

**Python 산술연산자의 종류**

| 연산 | 연산자 |
|---|---|
| 더하기 | `+` |
| 빼기 | `-` |
| 곱하기 | `*` |
| 나누기 | `/` |
| 몫 | `//` |
| 나머지 | `%` |
| 거듭제곱 | `**` |

**산술연산의 우선순위**

- 수학에서와 동일한 연산의 우선순위 규칙을 가집니다.
- 괄호 안이 가장 먼저 계산되고, 제곱 → 곱셈과 나눗셈 → 덧셈과 뺄셈의 순서로 계산됩니다.
- ※ 단, 파이썬에서는 중괄호나 대괄호를 사용하지 않고, 소괄호만 사용합니다.

**[직접 입력]** 아래 코드를 다음 셀에 입력하고 실행합니다.

```python
print('10+3=',10+3)
print('10-3=',10-3)
print('10*3=',10*3)
print('10/3=',10/3)
print('10//3=',10//3)
print('10%3=',10%3)
print('10**3=',10**3)
```

````python
# [직접 입력] 위 코드를 입력하세요.

10^3
````

실행 결과
```
10+3= 13
10-3= 7
10*3= 30
10/3= 3.3333333333333335
10//3= 3
10%3= 1
10**3= 1000
```

- `/`는 결과가 항상 float(실수)입니다.
- `//`는 나눈 몫, `%`는 나눈 나머지입니다.

## 비교연산

값을 비교합니다. 결과는 부울값(True/False)으로 계산됩니다.

| 비교연산자 | 사용 | 비고 |
|---|---|---|
| `==` | `x == y` | 등호 2개로 표현한다는 점에 유의 (`=`는 할당연산자이다.) |
| `!=` | `x != y` | 느낌표를 먼저 쓴다는 점에 유의 (`=!`로 쓰지 않는다.) |
| `>` | `x > y` | |
| `<` | `x < y` | |
| `>=` | `x >= y` | 부등호를 먼저 쓴다는 점에 유의 (`=>`로 쓰지 않는다.) |
| `<=` | `x <= y` | 부등호를 먼저 쓴다는 점에 유의 (`=<`로 쓰지 않는다.) |

## 논리연산

| 논리연산자 | 기능 |
|---|---|
| `and` | 그리고 (모두 True일 때만 True) |

| False | True |

## 할당연산

`+=`, `-=` 같은 할당연산자는 변수의 값을 계산한 뒤 같은 변수에 다시 담습니다.

| 할당연산 | 같은 뜻 |
|---|---|
| `x += 3` | `x = x + 3` |
| `x -= 3` | `x = x - 3` |
| `x *= 3` | `x = x * 3` |
| `x /= 3` | `x = x / 3` |
| `x //= 3` | `x = x // 3` |
| `x %= 3` | `x = x % 3` |
| `x **= 3` | `x = x ** 3` |

```python
print(stock)
```
