### [쉬운 미션] 회의 준비 안내문 만들기

회의 이름, 참석 인원, 회의 시간(분)을 입력받아 회의 준비 안내문을 출력하는 프로그램을 작성하세요.

1. 회의 이름은 문자열, 참석 인원과 회의 시간은 정수로 입력받습니다.
2. 참석 인원이 10명 이상이면 `'대회의실'`, 아니면 `'소회의실'`을 변수 `room`에 담습니다.
3. 다과비는 1인당 4,000원입니다. 다과비 합계를 구합니다.
4. 회의 시간을 `"1시간 30분"` 모양으로 바꿉니다. (`//`, `%` 사용)
5. 아래 실행 결과 예시처럼 f-string으로 안내문을 출력합니다. 위아래 줄은 `'-'`를 30번 반복합니다.

힌트: 다과비는 `{값:,}`으로 천 단위 쉼표를 넣습니다.

````python
title = input('회의 이름: ')
people = int(input('참석 인원: '))
minutes = int(input('회의 시간(분): '))

# 2. 회의실 정하기
if people >= 10:
    room = '대회의실'
else:
    room = '소회의실'

# 3. 다과비
snack = people * 4000

# 4. 회의 시간 바꾸기
hours = minutes // 60
mins = minutes % 60

# 5. 안내문 출력
print('-' * 30)
print(f'[{title}] 준비 안내')
print(f'회의실: {room}')
print(f'회의 시간: {hours}시간 {mins}분')
print(f'다과비: {snack:,}원 ({people}명)')
print('-' * 30)
````

### [어려운 미션] 주문 접수 처리 프로그램

주문 내용이 `"거래처,상품,수량"` 모양의 한 줄 문자열로 들어옵니다. 아래 순서대로 주문 확인서를 출력하는 프로그램을 만드세요.

1. 주문 한 줄과 회원 등급을 입력받습니다. 주문 줄은 앞뒤 공백을 `strip()`으로 지웁니다.
2. `split(',')`으로 나눈 리스트에서 거래처(0번), 상품(1번), 수량(2번)을 꺼냅니다. 수량은 `int()`로 바꿉니다.
3. 단가표 딕셔너리 `price`에서 상품 단가를 꺼내 공급가액(단가 × 수량)을 구합니다.
4. 수량에 따라 할인율을 정합니다.
   - 50개 이상 : 10%
   - 20개 이상 : 5%
   - 그 외 : 0%
5. 할인 금액 = 공급가액 × 할인율 / 100 (정수), 할인 후 금액 = 공급가액 - 할인 금액
6. 부가세 = 할인 후 금액 × 0.1 (정수), 합계 = 할인 후 금액 + 부가세
7. 배송비 : 합계가 500,000원 이상**이거나** 회원 등급이 `'VIP'`이면 0원, 아니면 5,000원입니다. 회원 등급은 `upper()`로 대문자로 바꾼 뒤 비교합니다.
8. 한 상자에 12개씩 담습니다. 꽉 찬 상자 수(`//`)와 남는 개수(`%`)를 구합니다.
9. 아래 실행 결과 예시처럼 주문 확인서를 출력합니다. 금액은 모두 천 단위 쉼표를 넣습니다.

힌트: 거래처 이름은 `client`, 상품은 `item`처럼 의미가 드러나는 변수명을 씁니다.

````python
price = {'노트북': 1200000, '모니터': 280000, '키보드': 45000, '마우스': 18000}

# 1. 입력받기
line = input('주문(거래처,상품,수량): ').strip()
grade = input('회원 등급: ').upper()

# 2. 나누기
parts = line.split(',')
client = parts[0]
item = parts[1]
qty = int(parts[2])

# 3. 공급가액
supply = price[item] * qty

# 4. 할인율
if qty >= 50:
    rate = 10
elif qty >= 20:
    rate = 5
else:
    rate = 0

# 5. 할인 금액, 할인 후 금액
discount = int(supply * rate / 100)
after = supply - discount

# 6. 부가세, 합계
vat = int(after * 0.1)
total = after + vat

# 7. 배송비
if total >= 500000 or grade == 'VIP':
    fee = 0
else:
    fee = 5000

# 8. 상자 수
boxes = qty // 12
left = qty % 12

# 9. 주문 확인서 출력
print('=' * 34)
print(f'주문 확인서 - {client} ({grade})')
print('=' * 34)
print(f'상품: {item} {qty}개 (단가 {price[item]:,}원)')
print(f'공급가액: {supply:,}원')
print(f'할인({rate}%): -{discount:,}원')
print(f'부가세: {vat:,}원')
print(f'배송비: {fee:,}원')
print(f'결제 금액: {total + fee:,}원')
print(f'포장: {boxes}상자 + 낱개 {left}개')
print('=' * 34)
````
