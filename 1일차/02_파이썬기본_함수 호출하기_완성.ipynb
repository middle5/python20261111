# 함수 호출하기

> `input()`이 들어 있는 셀을 실행하면 입력 칸이 나타납니다. 값을 입력하고 Enter를 누르면 다음 줄이 실행됩니다.

## 함수 호출하기

**함수란, 특정 기능을 미리 만들어 놓은 것.**

**호출하는 법**

```
함수이름( 함수에 전달할 값,… )
```

- 함수에 전달할 값이 없으면 빈 괄호를 사용합니다.
- 함수에 값을 전달하는 방법은 함수에 따라 다릅니다.
- 함수에 전달하는 값을 "인자", "인수", "아규먼트"라고 합니다.
- 함수는 기능을 수행한 후 그 결과를 반환합니다.

## 출력함수

**출력기능** : `print( 출력할 내용 )`

- 문자열은 따옴표로 묶어줍니다.
- 쌍따옴표(""), 홑따옴표('') 모두 사용 가능합니다.

````python
print('Hello')
print(1+1)
print("1+1")
````

실행 결과
```
Hello
2
1+1
```

`1+1`은 계산한 결과 `2`가 출력되고, 따옴표로 묶은 `"1+1"`은 글자 그대로 출력됩니다.

## 입력함수

**입력기능** : `input( 안내문구 )`

셀을 실행하면 안내문구와 입력 칸이 나타납니다. 이름을 입력하고 Enter를 누릅니다.

````python
input("이름을 입력하세요 >>> ")
````

## 반올림함수

- 일반 파이썬에서는 **help(함수명)**을 사용하여 함수의 도움말을 볼 수 있습니다.

````python
print?
````

도움말 화면 (일부)
```
Docstring:
print(value, ..., sep=' ', end='\n', file=sys.stdout, flush=False)

Prints the values to a stream, or to sys.stdout by default.
Optional keyword arguments:
file:  a file-like object (stream); defaults to the current sys.stdout.
```

- 괄호 안에서 함수에 전달할 값들(parameters)을 확인할 수 있습니다.
- 파이썬 버전에 따라 첫 줄이 `print(*args, sep=' ', end='\n', file=None, flush=False)`처럼 조금 다르게 보일 수 있습니다. `sep`, `end`의 디폴트 값은 같습니다.
- 디폴트 값이 할당되어 있는 경우(`sep=' '`, `end='\n'`처럼 `=` 뒤에 값이 있는 경우) 값을 전달하지 않으면 디폴트 값이 사용됩니다.

일반 파이썬과 같은 방법인 `help(print)`로도 같은 내용을 볼 수 있습니다.

````python
help(print)
````

## print 함수 도움말 : value, ...

`value, ...` : 가변 개수의 값을 받을 수 있습니다. 두 개 이상의 값을 전달할 때는 콤마(,)로 구분하여 전달합니다.

````python
# 두개 이상의 값 출력하기
print('python', 'java', 'c')
print('1+1=', 1+1)
````

실행 결과
```
python java c
1+1= 2
```

`sep` : 출력할 각 값 사이에 string(문자열)이 삽입됩니다. 디폴트 값은 space(`' '`)입니다. 즉, sep를 별도로 지정하지 않으면 각 값을 space(`' '`)로 구분하여 출력됩니다.

## print 함수 – sep 옵션

````python
# sep를 다른 문자로 지정하기
print('python', 'java', 'c', sep='-')
print('1+1=', 1+1, sep='')
````

실행 결과
```
python-java-c
1+1=2
```

## print 함수 – end 옵션

`end` : 마지막 value 다음에 문자열이 추가됩니다. 디폴트 값은 `'\n'`(개행문자, 줄바꿈)입니다. 즉, end를 별도로 지정하지 않으면 마지막에 줄바꿈을 붙여 출력합니다.

````python
# 끝문자를 별도로 지정하지 않으면 '\n'
print('Life is short')
print('You need python')
````

실행 결과
```
Life is short
You need python
```

````python
# 끝문자 지정하기
print('Life is short', end=', ')
print('You need python')
````

실행 결과
```
Life is short, You need python
```

## 함수의 옵션

````python
print('Hello')
print(1+1)
print('1+1')
print()
print('Hello', 'World')
print('Hello', 'World', sep=",", end='!')
````

실행 결과
```
Hello
2
1+1

Hello World
Hello,World!
```

- `print()`처럼 값을 전달하지 않으면 빈 줄이 출력됩니다.

**옵션의 디폴트 값 : 세탁기 비유**

| 옵션 | 디폴트 | 변경 |
|---|---|---|
| 코스 | 표준 | 표준 |
| 헹굼 | 2회 | 5회 |
| 탈수 | 중 | 중 |
| 물온도 | 40도 | 60도 |

세탁기는 아무 설정도 바꾸지 않으면 디폴트 값으로 동작하고, 바꾸고 싶은 옵션만 골라서 바꿉니다. 함수의 옵션도 같습니다. `print()`의 `sep`, `end`를 지정하지 않으면 디폴트 값이 쓰이고, 바꾸고 싶은 옵션만 `sep='-'`처럼 지정합니다.

사용자에게 키보드로부터 값을 입력 받습니다. 보통 어떤 값을 입력할 지 prompt를 주어 사용합니다.

- 순서에 맞게 값을 전달하는 경우는 매개변수명(`prompt=`)을 생략할 수 있습니다.
- `input(prompt=...)`처럼 매개변수명을 쓰는 방식은 주피터 노트북(코랩 포함)에서만 동작합니다. 일반 파이썬(.py 파일)에서는 오류가 나므로 `input("...")`처럼 매개변수명을 생략해서 씁니다.

## input 함수 호출하기 : 반환(return)

실행이 완료되면 사용자가 입력한 값을 **반환(return)**합니다.

**※ 함수의 반환(return) 값**
- 함수는 실행 결과를 반환하는 함수와 반환하지 않는 함수가 있습니다.
- 함수가 반환한 값은 변수에 할당하여 사용할 수 있습니다.

**함수의 반환 값을 변수에 담아 사용하기**

1. `input` 함수는 사용자가 입력한 값을 반환합니다.
2. 함수가 반환한 값을 변수 `name`에 할당합니다.
3. 함수의 반환 값을 이용하여 문장을 만들어 출력합니다.

````python
name = input(prompt="이름을 입력하세요:")
print(name, '님, 안녕하세요?', sep='')
````

실행 결과 (김철수 입력 시)
```
이름을 입력하세요:김철수
김철수님, 안녕하세요?
```

## input 함수 호출하기 : 입력값은 문자열

사용자로부터 입력 받은 값은 무조건 **'문자열'로 반환**합니다.

아래 셀은 **오류가 나는 것을 확인하는 셀**입니다. 나이를 입력하면 `TypeError`가 납니다.

````python
# [오류 확인] 오류가 나는 것이 정상입니다.
age = input(prompt='나이를 입력하세요 : ')
print('내년이면 ', 1+age, '살이 되시겠네요', sep='')
````

오류 메시지
```
TypeError: unsupported operand type(s) for +: 'int' and 'str'
```

int(정수)와 str(문자열)을 연산자 `+`를 이용하여 덧셈할 수 없어 에러가 발생합니다. `1`은 int, `age`는 str이기 때문입니다.

## input 함수 호출하기 : int()로 변환하기

`int()` 함수를 이용하여 자료형을 정수형으로 변환합니다.

1. 사용자가 입력한 값을 str(문자열)로 반환
2. `int()`가 int(정수)형으로 자료형 변환하여 반환
3. int(정수)형으로 변환된 값을 변수 `age`에 할당

````python
age = int(input(prompt='나이를 입력하세요 : '))
print('내년이면 ', age+1, '살이 되시겠네요', sep='')
````

실행 결과 (20 입력 시)
```
내년이면 21살이 되시겠네요
```

## 함수 호출하기 정리

- 함수를 호출할 때, 괄호 안에 필요한 값들(parameters)을 전달합니다.
- parameter에 디폴트 값이 할당되어 있는 경우, 값을 전달하지 않으면 디폴트 값이 사용됩니다.
- 함수 실행 결과를 반환하는 함수와 반환하지 않는 함수가 있습니다.
- 실행 결과를 반환하는 함수는 반환 값을 변수에 담아 사용할 수 있습니다.

### [응용문제 2] 분기 매출 합계와 평균 (영업)

1월, 2월, 3월 매출(만 원)을 하나씩 입력받아, 1분기 매출 합계와 월평균을 출력하세요. 월평균은 소수점 둘째 자리까지 반올림합니다.

- 입력받은 값은 `int()`로 바꿔서 변수에 담습니다.
- 합계는 `sum([값1, 값2, 값3])`로 구합니다.
- 평균은 `합계 / 3`으로 구합니다.

실행 결과 (1250, 980, 1410 입력 시)
```
1월 매출(만 원): 1250
2월 매출(만 원): 980
3월 매출(만 원): 1410
1분기 합계: 3640 만 원
월평균: 1213.33 만 원
```

````python
jan = int(input('1월 매출(만 원): '))
feb = int(input('2월 매출(만 원): '))
mar = int(input('3월 매출(만 원): '))
total = sum([jan, feb, mar])
print('1분기 합계:', total, '만 원')
print('월평균:', round(total / 3, 2), '만 원')
````

### [응용문제 3] 광고 클릭률 계산 (마케팅)

광고 노출 수와 클릭 수를 입력받아 클릭률(%)을 출력하세요.

- 클릭률 = 클릭 수 / 노출 수 × 100
- 클릭률은 소수점 첫째 자리까지 반올림합니다.
- `%` 기호는 숫자 뒤에 빈칸 없이 붙입니다.

실행 결과 (12000, 456 입력 시)
```
노출 수: 12000
클릭 수: 456
클릭률: 3.8%
```

````python
views = int(input('노출 수: '))
clicks = int(input('클릭 수: '))
print('클릭률: ', round(clicks / views * 100, 1), '%', sep='')
````
