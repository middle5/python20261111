# 선택구조

## 프로그램의 3가지 기본 제어 구조

| 구조 | 설명 |
|---|---|
| 순차구조 | 명령들이 순차적으로 실행되는 구조 |
| 선택구조 | 조건에 따라 명령을 선택하여 실행되는 구조 (조건이 True이면 한쪽, False이면 다른 쪽 실행) |
| 반복구조 | 동일한 명령이 반복되면서 실행되는 구조 |

## 흐름 제어

| 흐름제어 | 키워드 |
|---|---|
| 조건 | `if` `elif` `else` |
| 반복 | `for` `while` |
| 반복제어 | `break` `continue` |

이 노트북 파일에서는 조건(`if`, `elif`, `else`)을 다룹니다.

## 선택구조 if

**[ 파이썬 문법 ]**

```python
if 조건식:
    코드블록
```

- 조건식 뒤에는 콜론(`:`)을 붙입니다.
- 조건식의 결과가 True일 때 실행할 코드들은 들여쓰기 합니다.
- 조건식의 결과가 True이면 실행할 코드블록을 실행합니다.
- 조건식의 결과가 False이면 조건문을 빠져나옵니다.

````python
# 60점 이상이면 합격
score = 60
if score>=60:
    print('합격입니다.')
    print('축하합니다.')
print('수고하셨습니다.')
````

실행 결과
```
합격입니다.
축하합니다.
수고하셨습니다.
```

`score`를 50으로 바꿔 실행해 보세요. 들여쓰기 한 두 줄은 실행되지 않고, 마지막 줄 `수고하셨습니다.`만 출력됩니다. 마지막 줄은 들여쓰기 하지 않아서 if문 밖에 있기 때문입니다.

## 선택구조 if~else~

**[ 파이썬 문법 ]**

```python
if 조건식:
    코드블록1
else:
    코드블록2
```

- 조건식의 결과가 True이면 코드블록1을 실행하고 조건문을 빠져나옵니다.
- 조건식의 결과가 False이면 코드블록2를 실행하고 조건문을 빠져나옵니다.

````python
# 60점 이상이면 합격, 아니면 불합격
score = 30
if score>=60:
    print('합격입니다.')
    print('축하합니다.')
else:
    print('불합격입니다')
    print('안타깝군요.')
print('수고하셨습니다.')
````

실행 결과
```
불합격입니다
안타깝군요.
수고하셨습니다.
```

## 선택구조 if~elif~else~

조건을 위에서부터 차례로 검사하다가, 처음으로 True가 나온 곳의 코드블록만 실행합니다.

| 순서 | 조건 | True이면 |
|---|---|---|
| 1 | 점수 >= 90 | A |
| 2 | 점수 >= 80 | B |
| 3 | 점수 >= 70 | C |
| 4 | 점수 >= 60 | D |
| 5 | 모두 False | F |

````python
# 점수를 입력받아 A,B,C,D,F 등급 출력하기

score = int(input("점수:"))
if score >= 90:
  print("A")
elif score >=80:
  print("B")
elif score >=70:
  print("C")
elif score >= 60:
  print("D")
else:
  print("F")
````

실행 결과 (85 입력 시)
```
점수:85
B
```

**순서를 바꾸면?** `score >= 60`을 맨 위에 두면, 85점도 첫 조건에서 True가 되어 `D`가 출력됩니다. 범위로 나누는 조건은 좁은 범위(높은 점수)부터 검사해야 합니다. 아래 「무엇이 잘못됐을까요?」 연습문제에서 같은 문제를 다시 봅니다.

## [연습문제] 구매 금액별 할인

구매 금액을 입력 받아 아래 기준으로 할인율을 정하고, 할인율과 최종 결제 금액을 출력하세요.

| 구매 금액 | 할인율 |
|---|---|
| 1,000,000원 이상 | 10% |
| 500,000원 이상 | 5% |
| 그 외 | 0% |

실행 결과

| 입력 | 할인율 | 결제 금액 |
|---|---|---|
| 1200000 | 10 | 1080000 |
| 650000 | 5 | 617500 |
| 300000 | 0 | 300000 |

````python
amount = int(input('구매 금액(원):'))

if amount >= 1000000:
    rate = 10
elif amount >= 500000:
    rate = 5
else:
    rate = 0

pay = int(amount * (100 - rate) / 100)

print('할인율:', rate, '%')
print('결제 금액:', pay)
````

## [연습문제] 무엇이 잘못됐을까요?

아래 코드에 1,200,000을 입력하면 할인율이 몇 %로 나올까요? 왜 그럴까요? 먼저 예상한 뒤 실행해 봅니다.

````python
amount = int(input('구매 금액(원):'))

if amount >= 500000:
    rate = 5
elif amount >= 1000000:
    rate = 10
else:
    rate = 0

print('할인율:', rate, '%')
````

실행 결과 (1200000 입력 시)
```
할인율: 5 %
```

**이유** : 1,200,000은 첫 조건 `amount >= 500000`에서 이미 True가 됩니다. if~elif는 처음 True가 나온 곳만 실행하고 빠져나오므로, 아래의 `amount >= 1000000`은 검사하지 않습니다. 금액이 큰 조건(1,000,000원 이상)을 먼저 검사해야 합니다.

````python
# 조건 순서를 바로잡은 코드
amount = 1200000

if amount >= 1000000:
    rate = 10
elif amount >= 500000:
    rate = 5
else:
    rate = 0

print('할인율:', rate, '%')
````

## [연습문제] 배송비 계산

주문 금액과 회원 등급을 입력 받아, 배송비와 최종 결제 금액을 출력하세요.

- 주문 금액이 50,000원 이상이거나 회원 등급이 'VIP'이면 배송비는 0원입니다.
- 위 조건에 해당하지 않으면 배송비는 3,000원입니다.
- 최종 결제 금액은 주문 금액에 배송비를 더한 금액입니다.

실행 결과

| 입력 (금액, 등급) | 배송비 | 결제 금액 |
|---|---|---|
| 42000, 일반 | 3000 | 45000 |
| 42000, VIP | 0 | 42000 |

````python
amount = int(input('주문 금액(원):'))
grade = input('회원 등급:')

if amount >= 50000 or grade == 'VIP':
    fee = 0
else:
    fee = 3000

print('배송비:', fee)
print('결제 금액:', amount + fee)
````

### [연습문제 3] 파일 형식 확인

파일 이름을 입력받아 `.xlsx`로 끝나면 "엑셀 파일", `.pdf`로 끝나면 "PDF 파일", 그 외에는 "지원하지 않는 형식"을 출력하세요. 대문자 확장자(`.PDF`)도 같은 규칙을 적용하도록 `lower()`를 먼저 씁니다.

실행 결과 (`견적서.PDF` 입력 시)

### [응용문제 1] 경비 결재 라인 정하기 (일반 업무)

경비 금액을 입력받아 결재자를 출력하세요.

| 경비 금액 | 결재자 |

| 300,000원 이상 | 팀장 |

### [응용문제 2] 고객 등급과 적립금 (마케팅)

연간 구매 금액을 입력받아 고객 등급과 적립금을 출력하세요. 적립금은 구매 금액 × 적립률이고, 정수로 만듭니다.

| 연간 구매 금액 | 등급 | 적립률 |
|---|---|---|
| 5,000,000원 이상 | VIP | 5% |
| 1,000,000원 이상 | 골드 | 3% |
| 그 외 | 일반 | 1% |

실행 결과 (2400000 입력 시)
```
연간 구매 금액: 2400000

```

### [응용문제 3] 영업 목표 달성 판정 (영업)

월 목표 매출과 실제 매출(만 원)을 입력받아 달성률(%)을 구하고, 아래 기준으로 판정을 출력하세요. 달성률은 소수점 첫째 자리까지 반올림합니다.

| 달성률 | 판정 |
|---|---|
| 120% 이상 | 초과 달성 (인센티브 지급) |
| 100% 이상 | 목표 달성 |
| 80% 이상 | 분발 필요 |
| 그 외 | 미달 (면담 대상) |

실행 결과 (5000, 4350 입력 시)
```
목표 매출(만 원): 5000
실제 매출(만 원): 4350
달성률: 87.0%
판정: 분발 필요
```

````python
target = int(input('목표 매출(만 원): '))
sales = int(input('실제 매출(만 원): '))
rate = round(sales / target * 100, 1)
if rate >= 120:
    result = '초과 달성 (인센티브 지급)'
elif rate >= 100:
    result = '목표 달성'
elif rate >= 80:
    result = '분발 필요'
else:
    result = '미달 (면담 대상)'
print(f'달성률: {rate}%')
print('판정:', result)
````
