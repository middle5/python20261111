# 파이썬 업무자동화 개요

## 업무자동화란?

**업무자동화** : 수작업으로 반복하던 컴퓨터 업무를 프로그램으로 처리하는 것

→ 업무 시간 단축, 업무 정확성 향상, 생산성 향상, 업무 표준화

**예: 지점별 엑셀 보고서 50개를 하나로 합치기**

| 구분 | 수작업 | 파이썬 자동화 |
|---|---|---|
| 방법 | ① 엑셀 파일 열기<br>② 데이터 범위 복사<br>③ 취합 파일에 붙여넣기<br>④ 저장하고 닫기<br>→ 50개 파일 모두 반복 | 코드 5줄 (아래 코드) |
| 시간 | 파일당 1분, 약 50분 | 코드 5줄, 실행은 몇 초 |
| 정확성 | 복사 범위 실수가 생기기 쉬움 | 파일이 늘어나도 매번 같은 결과 |

```python
import glob
import pandas as pd
files = glob.glob("보고서/*.xlsx")
df = pd.concat([pd.read_excel(f) for f in files])
df.to_excel("취합.xlsx", index=False)
```

※ 위 코드는 자동화의 모습을 보여 주는 예시입니다. 보고서 폴더가 없어서 이 노트북 파일에서는 실행하지 않습니다.

## 자동화하기 좋은 업무 고르기

| 기준 | 자동화하기 좋은 업무 | 사람이 하는 게 나은 업무 |
|---|---|---|
| 반복성 / 일회성 | 자주 반복된다<br>매일·매주·매월 하는 일 | 한 번만 하고 끝난다<br>예: 연 1회 행사 기획 |
| 규칙성 / 비규칙성 | 일하는 순서와 기준이 정해져 있다<br>예: 100만 원 이상이면 팀장 결재 | 매번 판단 기준이 달라진다<br>예: 상황 따라 결정하는 협상·민원 대응 |
| 정형성 / 비정형성 | 다루는 자료의 형식이 일정하다<br>예: 매달 같은 양식의 엑셀 보고서 | 자료의 형식이 매번 제각각이다<br>예: 거래처마다 다른 양식의 견적서 |
| 대량성 / 소량성 | 양이 많아 실수가 잦다<br>예: 파일 수십~수백 개 | 양이 적어 손으로도 금방 끝난다<br>예: 한 달에 파일 몇 개 |

**판단 기준** : 이 일을 다른 사람에게 순서대로 설명할 수 있다면, 코드로도 설명할 수 있습니다.

## 파일·폴더 자동화란?

폴더와 파일을 다루는 반복 작업을 코드 몇 줄로 처리합니다.

| 기능 | 하는 일 | 예시 → 처리 결과 |
|---|---|---|
| 01 파일 탐색 | 조건에 맞는 파일 찾기 | 보고서.pdf, 매출.xlsx, 회의록.txt 중 조건 `확장자 .pdf` → 보고서.pdf |

| 03 파일 이동·복사 | 위치 변경·복제, 필요한 폴더 만들기 | 다운로드/보고서.pdf → 이동: 문서/보고서.pdf, 복사: 백업/보고서.pdf |

## 도구: os · pathlib · shutil

| 모듈 | 역할 | 자주 쓰는 기능 |
|---|---|---|

| shutil | 파일·폴더 이동, 복사, 압축 | `shutil.move()`, `shutil.copy()`, `shutil.copytree()` |
| os | 운영체제 기능 사용 | `os.listdir()`, `os.rename()`, `os.makedirs()` |

- 세 모듈 모두 파이썬에 기본 포함 → 설치 없이 import만 하면 됩니다.

## 사용할 도구 ① 파일 탐색

**목표** : 폴더 안에서 조건에 맞는 파일만 찾기

|---|---|

```python
    print(f.name)

    print(f.name, f.stem, f.suffix)
```

## 기본 사용법 ② 이름 변경

**목표** : 규칙에 맞춰 파일 이름 바꾸기

| 기능 | 설명 |
|---|---|
| `rename(새 경로)` | 파일 이름(또는 위치) 바꾸기 |
| `sorted()` | 순서대로 정렬해 번호가 꼬이지 않게 |
| `enumerate(목록, 1)` | 1부터 번호 붙이며 하나씩 꺼내기 |
| `f"{i:03d}"` | 숫자를 001처럼 3자리로 맞추기 |

**주의** : 같은 이름의 파일이 이미 있으면 오류가 납니다. 복사본 폴더에서 먼저 실행하세요.

**[직접 입력]** 아래 코드를 다음 셀에 입력하고 실행합니다.

```python
from pathlib import Path
folder = Path("사진")
files = sorted(folder.glob("IMG_*.jpg"))
for i, f in enumerate(files, 1):
    new = f"현장A_{i:03d}{f.suffix}"   # 현장A_001.jpg
    f.rename(folder / new)
    print(f.name, "→", new)
```

## 사용할 도구 ③ 파일 이동·복사

**목표** : 파일을 다른 폴더로 옮기거나 복제하기

| 기능 | 설명 |
|---|---|
| `shutil.copy(원본, 대상)` | 파일 복제. 원본은 그대로 남음 |
| `shutil.move(원본, 대상)` | 파일 이동. 원본 위치에서 사라짐 |
| 대상 = 폴더 | 같은 이름으로 그 폴더 안에 저장 |
| 대상 = 파일 경로 | 새 이름까지 지정해 저장 |

**주의** : 대상 폴더가 없으면 오류가 납니다. `mkdir()`로 먼저 만들어 두세요.

**[직접 입력]** 아래 코드를 다음 셀에 입력하고 실행합니다.

```python
import shutil
from pathlib import Path
src = Path("다운로드/보고서.pdf")
Path("문서").mkdir(exist_ok=True)
shutil.copy(src, "문서")    # 복사: 원본 그대로
Path("보관").mkdir(exist_ok=True)
shutil.move(src, "보관")    # 이동: 원본 사라짐
```

````python
# [직접 입력] 위 코드를 입력하세요.
````

실행 후 폴더

```
보관/보고서.pdf   ← 이동 (다운로드에서 사라짐)
```

한 번 실행하면 `다운로드/보고서.pdf`가 보관 폴더로 옮겨져서, 다시 실행하면 오류가 납니다. 처음부터 다시 하려면 `%run 실습파일만들기0.py` 셀을 다시 실행합니다.

## 예제 ① 거래명세서 날짜별 분류

**정리 전: 한 폴더에 모두 섞여 있음**

**정리 후: 날짜별 폴더로 분류**
```

    거래명세서_20261105_B물산.xlsx
```

**자동화 규칙** : 파일명에서 날짜(20261103)를 읽고 → 2026-11-03 폴더를 만든 뒤 → 그 폴더로 복사

### 예제 ① 풀이: 단계별로 생각하기

- **입력** : 거래명세서 폴더의 파일들
- **처리** : 파일명에서 날짜를 읽어 날짜 폴더로 복사
- **출력** : 날짜별로 정리된 폴더

| 단계 | 하는 일 | 코드 |
|---|---|---|
| 1. 거래명세서 파일만 찾기 | 메모.txt 같은 다른 파일은 제외 | `folder.glob("거래명세서_*.xlsx")` |
| 2. 파일명에서 날짜 꺼내기 | `_` 로 나누면 [이름, 날짜, 거래처] | `ymd = f.stem.split("_")[1]  # "20261103"` |
| 3. 날짜 폴더 만들기 | 20261103을 2026-11-03 모양으로 | `day = f"{ymd[:4]}-{ymd[4:6]}-{ymd[6:]}"`<br>`(folder / day).mkdir(exist_ok=True)` |
| 4. 파일을 해당 날짜 폴더로 복사 | 원본은 그대로, 날짜 폴더에 사본 생성 | `shutil.copy(f, folder / day / f.name)` |

### 예제 ① 풀이: 완성 코드

**[직접 입력]** 아래 코드를 다음 셀에 입력하고 실행합니다.

```python
from pathlib import Path   # 파일 경로를 다루는 클래스
import shutil              # 파일·폴더 복사, 이동

# 교안: Path(r"C:\workspace\python20261103\6일차\거래명세서")
# 노트북 파일과 같은 폴더에 있으면 폴더 이름만 써도 됩니다.
folder = Path("거래명세서")

# 1. 거래명세서 파일만 찾기
files = list(folder.glob("거래명세서_*.xlsx"))
for f in files:
    # 2. 파일명에서 날짜 꺼내기
    ymd = f.stem.split("_")[1]               # 20261103
    day = f"{ymd[:4]}-{ymd[4:6]}-{ymd[6:]}"  # 2026-11-03
    # 3. 날짜 폴더 만들기
    target = folder / day
    target.mkdir(exist_ok=True)
    # 4. 날짜 폴더로 복사하기 (원본은 그대로)
    shutil.copy(f, target / f.name)
    print(f.name, "→", day)
```

### 예제 ② 풀이: 완성 코드

**[직접 입력]** 아래 코드를 다음 셀에 입력하고 실행합니다.

```python
from pathlib import Path
from datetime import datetime

# 교안: Path(r"C:\workspace\python20261103\6일차\현장사진")
folder = Path("현장사진")

# 1. 사진 파일만 찾기 (이름순 정렬)
files = sorted(folder.glob("IMG_*.jpg"))
for i, f in enumerate(files, 1):
    # 2. 수정한 날짜 읽기
    t = f.stat().st_mtime
    day = datetime.fromtimestamp(t)
    # 3. 새 이름 만들기
    name = f"{day:%Y-%m-%d}_현장A_{i:03d}"
    new = name + f.suffix
    # 4. 이름 바꾸기
    f.rename(folder / new)
    print(f.name, "→", new)
```
