## 준비

```python
# ------------------------------------------------------
# 라이브러리 · 한글 폰트 · 데이터 준비
# ------------------------------------------------------
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import platform
_os = platform.system()
if _os == 'Windows':
    plt.rc('font', family='Malgun Gothic')
elif _os == 'Darwin':
    plt.rc('font', family='AppleGothic')
else:
    plt.rc('font', family='NanumGothic')
plt.rc('axes', unicode_minus=False)

# 하나의 데이터로 진행합니다 : tips (식당 결제·팁 데이터)
tips = sns.load_dataset('tips')
tips.head()
```

## A. matplotlib 기본

### 문제 1. 꺾은선 — 요일별 평균 결제금액
요일별 평균 `total_bill`을 구해 **꺾은선**으로 그리세요. (marker='o', 제목·축라벨)

> 힌트: `tips.groupby('day')['total_bill'].mean()`

```python
day_bill = tips.groupby('day')['total_bill'].mean()
plt.plot(day_bill.index.astype(str), day_bill.values, marker='o', color='green')
plt.title('요일별 평균 결제금액')
plt.xlabel('요일'); plt.ylabel('평균 결제금액')
plt.show()
```

### 문제 2. 막대 — 요일별 평균 팁
요일별 평균 `tip`을 막대그래프로 그리세요.

```python
day_tip = tips.groupby('day')['tip'].mean()
plt.bar(day_tip.index.astype(str), day_tip.values, color='#4C72B0')
plt.title('요일별 평균 팁')
plt.ylabel('평균 팁')
plt.show()
```

### 문제 3. 파이 — 요일별 방문 비율
요일별 방문 횟수의 비율을 파이그래프로 그리세요. (autopct 사용)

> 힌트: `tips['day'].value_counts()`

```python
day_count = tips['day'].value_counts()
plt.pie(day_count.values, labels=day_count.index, autopct='%.1f%%')
plt.title('요일별 방문 비율')
plt.show()
```

### 문제 4. 히스토그램 — 결제금액 분포
`total_bill`의 분포를 히스토그램으로 그리세요. (bins=20)

```python
plt.hist(tips['total_bill'], bins=20, color='skyblue', edgecolor='white')
plt.title('결제금액 분포')
plt.xlabel('total_bill'); plt.ylabel('빈도')
plt.show()
```

### 문제 5. 산점도 — 결제금액 vs 팁
`total_bill`과 `tip`의 관계를 산점도로 그리세요. (alpha=0.5)

```python
plt.scatter(tips['total_bill'], tips['tip'], alpha=0.5)
plt.title('결제금액 vs 팁')
plt.xlabel('total_bill'); plt.ylabel('tip')
plt.show()
```

## B. 서브플롯

### 문제 6. 2×2 서브플롯
`tips`를 2×2 한 그림에 요약하세요.
1. total_bill 히스토그램  2. tip 히스토그램
3. 요일별 방문 수 막대  4. total_bill vs tip 산점도
- 각 제목 + `suptitle` + `tight_layout`

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 7))

axes[0][0].hist(tips['total_bill'], bins=20); axes[0][0].set_title('결제금액 분포')
axes[0][1].hist(tips['tip'], bins=20); axes[0][1].set_title('팁 분포')

day_count = tips['day'].value_counts()
axes[1][0].bar(day_count.index.astype(str), day_count.values); axes[1][0].set_title('요일별 방문 수')
axes[1][1].scatter(tips['total_bill'], tips['tip'], alpha=0.5); axes[1][1].set_title('결제금액 vs 팁')

fig.suptitle('tips 데이터 요약')
fig.tight_layout()
plt.show()
```

## C. seaborn

### 문제 7. barplot — 요일·성별 평균 결제금액
요일별 평균 `total_bill`을 **성별(hue)**로 나눠 그리세요. (errorbar=None)

```python
sns.barplot(data=tips, x='day', y='total_bill', hue='sex', errorbar=None)
plt.title('요일·성별 평균 결제금액')
plt.show()
```

### 문제 8. boxplot — 요일별 결제금액 분포 (흡연 여부)
요일별 `total_bill` 분포를 **흡연 여부(smoker)**로 나눠 상자그림으로 그리세요.

```python
sns.boxplot(data=tips, x='day', y='total_bill', hue='smoker')
plt.title('요일별 결제금액 분포 (흡연 여부)')
plt.show()
```

### 문제 9. countplot — 요일별 방문 수 (성별)
요일별 방문 수를 **성별(hue)**로 나눠 그리세요.

```python
sns.countplot(data=tips, x='day', hue='sex')
plt.title('요일별 방문 수 (성별)')
plt.show()
```

### 문제 10. heatmap — 요일×성별 평균 팁
요일×성별 평균 `tip`을 피벗테이블로 만든 뒤 히트맵으로 그리세요. (annot=True, fmt='.2f')

> 힌트: `tips.pivot_table(index='day', columns='sex', values='tip', aggfunc='mean')`

```python
pivot = tips.pivot_table(index='day', columns='sex', values='tip', aggfunc='mean')
plt.figure(figsize=(4, 3))
sns.heatmap(pivot, annot=True, fmt='.2f', cmap='YlOrRd')
plt.title('요일·성별 평균 팁')
plt.show()
```

### 문제 11. regplot — 결제금액과 팁의 관계
`total_bill`과 `tip`을 산점도+회귀선으로 그리세요. (회귀선 빨강)

```python
sns.regplot(data=tips, x='total_bill', y='tip', line_kws={'color': 'red'})
plt.title('결제 금액과 팁의 관계')
plt.show()
```

## D. 미션 (실무형)

### 미션 1. 팁이 가장 높은 요일 강조
요일별 평균 팁 막대에서 **가장 높은 요일만 다른 색**으로 강조하세요.

```python
day_tip = tips.groupby('day')['tip'].mean()
colors = ['tomato' if v == day_tip.max() else 'lightgray' for v in day_tip.values]
plt.bar(day_tip.index.astype(str), day_tip.values, color=colors)
plt.title('요일별 평균 팁 (최고 요일 강조)')
plt.ylabel('평균 팁')
plt.show()
```

### 미션 2. 흡연 여부별 평균 팁
흡연 여부(smoker)에 따라 평균 팁이 다른지 막대로 비교하세요.

```python
sns.barplot(data=tips, x='smoker', y='tip', errorbar=None)
plt.title('흡연 여부별 평균 팁')
plt.show()
```
