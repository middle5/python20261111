## 전역 설정

```python
# ----------------------------------
# font 전역설정
# ----------------------------------
import platform
_os = platform.system()
if _os == 'Windows':
    plt.rc('font', family='Malgun Gothic')     # 윈도우
elif _os == 'Darwin':
    plt.rc('font', family='AppleGothic')        # 맥
else:
    plt.rc('font', family='NanumGothic')        # 리눅스 / 구글 코랩
    # ※ 코랩에서 한글이 깨지면 아래 줄을 먼저 실행하고 런타임을 재시작하세요:
    # !apt-get -qq install fonts-nanum && rm -rf ~/.cache/matplotlib
plt.rc('axes', unicode_minus=False)
pd.Series([-1,2,3]).plot(title='테스트', figsize=(3,2))

# ----------------------------------
# figsize 전역설정
# ----------------------------------
plt.rc('figure', figsize=(4,3))
```
