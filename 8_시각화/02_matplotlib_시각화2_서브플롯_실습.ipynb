## 앤스콤 데이터셋 — 왜 시각화가 필요한가
* 1973년 통계학자 프랜시스 앤스콤이 만든 **4개의 데이터셋**
* 네 데이터셋은 **평균·표준편차·상관계수가 거의 같다**
* 하지만 **그래프로 그리면 완전히 다른 모양** → 숫자만 봐선 안 되고, 시각화가 필요하다!

```python
# --------------------------
# 앤스콤 데이터셋 불러오기
# --------------------------
import seaborn as sns
anscombe = sns.load_dataset('anscombe')

df1 = anscombe[anscombe['dataset'] == 'I']
df2 = anscombe[anscombe['dataset'] == 'II']
df3 = anscombe[anscombe['dataset'] == 'III']
df4 = anscombe[anscombe['dataset'] == 'IV']
display(df1.head())
display(df2.head())
display(df3.head())
display(df4.head())

# 👉 여기에 코드를 작성하세요
```

```python
# --------------------------
# 통계 수치 확인 : 평균 · 표준편차 · 상관계수 (거의 같다!)
# --------------------------
print(f"df1 >>>>> mean:{df1['y'].mean().round(2)}, std:{df1['y'].std().round(2)}, corr:{df1['x'].corr(df1['y']).round(2)}")
print(f"df2 >>>>> mean:{df2['y'].mean().round(2)}, std:{df2['y'].std().round(2)}, corr:{df2['x'].corr(df2['y']).round(2)}")
print(f"df3 >>>>> mean:{df3['y'].mean().round(2)}, std:{df3['y'].std().round(2)}, corr:{df3['x'].corr(df3['y']).round(2)}")
print(f"df4 >>>>> mean:{df4['y'].mean().round(2)}, std:{df4['y'].std().round(2)}, corr:{df4['x'].corr(df4['y']).round(2)}")

# 👉 여기에 코드를 작성하세요
```

## 서브플롯
* 여러 그래프를 **한 그림에 나눠서** 그리는 방법
* **`fig, axes = plt.subplots(nrows, ncols)`**
    * `nrows` : 행의 수, `ncols` : 열의 수
    * `figsize` : 전체 그림 크기
* 각 칸에 그리기
    * `axes[행][열].plot(...)`
    * `axes[행][열].set_title(...)`
* 전체 설정
    * `fig.suptitle(...)` : 전체 제목
    * `fig.tight_layout()` : 그래프가 겹치지 않게 간격 조정

#### 기본 2 × 2 서브플롯
앤스콤 네 데이터셋을 각 칸에 그려 **모양이 다름**을 확인한다.

```python
# 👉 여기에 코드를 작성하세요
fig, axes = plt.subplots(2,2, sharex=True, sharey=True)
axes[0][0].plot(df1['x'], df1['y'], 'o')
axes[0][1].plot(df2['x'], df2['y'], 'o')
axes[1][0].plot(df3['x'], df3['y'], 'o')
axes[1][1].plot(df4['x'], df4['y'], 'o')
fig.suptitle("앤스콤 4분할 데이터")
plt.tight_layout()
pass
```
