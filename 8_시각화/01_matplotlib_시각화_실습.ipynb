## 꺾은선 그래프
* 시간의 흐름에 따른 데이터의 변화 시각화
* **`plt.plot(x축데이터, y축데이터)`**
>* 주요 옵션
>   * `color` : 색상이름, 16진수색상코드
>   * `marker` : `^`,`D`,`o`,`.`,`*` 등
>   * `linestyle` : `--`,`:`, `-.` 

```python
# --------------------------------------------
# 샘플데이터 : 어떤 아이의 나이별 키의 변화
# --------------------------------------------
age = [7,8,9,10,11,12,13,14,15]
height = [122.1, 127.9, 133.4, 138.8, 142.7, 151.4, 155.6, 165.0, 169.2]

# 👉 여기에 코드를 작성하세요

plt.figure(figsize=(3,2))
plt.plot(age,height,marker='o', ls='--', color='lightgreen')
plt.grid(ls = ':')
plt.title("oo의 키 성장 그래프")
plt.xlabel('나이')
plt.ylabel('키(cm)')
plt.xticks(age)
plt.yticks(height)
pass
```

## 파이그래프
* 비율의 시각화
* **`plt.pie(데이터)`**
>* 주요 옵션
>   * `labels` : 각 조각에 표시할 라벨 리스트
>   * `autopct` : 퍼센트 표시 형식 (예:'%.1f%%')
>   * `colors` : 조각 색상 리스트
>   * `explode` : 각 조각을 얼마나 띄울지 결정
>   * `wedgeprops` : 각 부채꼴 조각의 스타일 지정

```python
# --------------------------------------------
# 샘플데이터 : 100명 중 혈액형 인원 비교
# --------------------------------------------
blood_type=['A','B','O','AB']
count=[45,35,15,5]

# 👉 여기에 코드를 작성하세요

plt.pie(count, labels=[i+'형' for i in blood_type], autopct="%.1f%%")
pass
```

## 중첩플롯
* 하나의 axes 위에 여러 개의 그래프를 겹쳐서 시각화하는 방식
* 여러 변수 간의 비교나 추이를 한 눈에 파악할 수 있다.
* 하나의 코드셀에 그래프를 여러개 그리면 된다.
* 각 그래프를 구별하기 위한 범례를 표시한다.
    * 범례를 표시하는 방법
        * 그래프 label 옵션 지정
        * `plt.legend()` 설정

```python
# --------------------------------------------
# 샘플 데이터 
# 두 사람의 월별 몸무게 변화
# --------------------------------------------
df1 = pd.DataFrame({'월':[1,2,3,4,5],'몸무게':[80,78,75,73,70]}) #철수
df2 = pd.DataFrame({'월':[1,2,3,4,5],'몸무게':[60,62,59,55,54]}) #영희

# 👉 여기에 코드를 작성하세요

plt.figure(figsize=(3,2))
display(df1)
display(df2)
```

```python
# 👉 여기에 코드를 작성하세요

plt.plot(df1['월'], df1['몸무게'], marker='o', label='철수')
plt.plot(df2['월'], df2['몸무게'], marker='o', label='영희')
plt.xticks(df1['월'])
plt.legend()
plt.ylim(50,85)
pass
```

```python
# 👉 여기에 코드를 작성하세요

plt.plot(df1['월'], df1['몸무게'], marker='o', label='철수', color='red')
plt.bar(df2['월'], df2['몸무게'], label='영희')
plt.legend()
```
