### [연습문제 1-10] for 반복구조를 리스트 컴프리헨션으로 바꾸기

아래 셀의 for 반복구조는 교육 평가 점수가 80점 이상이면 '수료', 아니면 '재교육'을 담은 리스트를 만듭니다. 같은 결과가 나오도록 리스트 컴프리헨션 한 줄로 바꿔 쓰세요.

````python
scores = [85, 72, 90, 64]

result = []
for score in scores:
    if score >= 80:
        result.append('수료')
    else:
        result.append('재교육')
print(result)

# 여기에 리스트 컴프리헨션으로 작성하세요
````

```

```
