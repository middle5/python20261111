## 함수 정의

```
def 함수명(매개변수):
    함수의 기능 구현
    return 반환값
```

- 매개변수(입력) : 없을 경우 생략.
- 함수의 기능(처리)
- 결과값(반환) : 없을 경우 생략.

## 함수 정의와 호출

**<함수 정의>**

```
def 함수명(매개변수):
    함수의 기능 구현
    return 반환 값
```

- 함수명은 변수명과 동일한 규칙으로 만들어진다.
- 매개변수는 필요한 만큼 기술할 수 있으며, 없을 수도 있다.
- 반환 값은 없거나 1개이다.

**<함수 호출>**

```
함수명(인자)
```

- 함수 이름 뒤에 매개변수에 전달할 인자를 기술한다. (없는 경우 생략)
- 함수를 호출하기 전에 함수가 정의되어 있어야 한다.
- 반환 값이 있을 때는 변수로 받거나 수식 내에 기술한다.

※ 용어정리
- 매개변수(parameter) : 값을 전달받는 변수
- 인자(argument) : 함수를 호출할 때 전달하는 값

## 함수의 정의와 호출 - 매개변수 있음, 반환 값 있음

## 함수의 정의와 호출 - 매개변수 있음, 반환 값 있음

반환 값이 여러 개인 경우, 튜플 패킹과 언패킹

함수에서 여러 값 반환 – 튜플 패킹

````python
def get_plus_minus(n1,n2):
    plus = n1+n2
    minus = n1-n2
    return plus, minus
result = get_plus_minus(1, 2)
print(result)
````

````python
def get_plus_minus(n1,n2):
    plus = n1+n2
    minus = n1-n2
    return plus, minus
result1, result2 = get_plus_minus(1, 2)
print(result1, result2, sep=',')
````

공급가액, 부가세, 배송비, 최종 결제 금액을 한 번에 리턴 하고 양식에 맞게 출력하도록 코드를 수정하세요.


## [미션] 상품 재고 확인

사용자가 입력한 상품의 재고가 있는지 확인하는 check_stock 함수를 작성하세요.

1. 상품별 재고 수량

| 상품명 | 재고 수량 |
|---|---|
| 볼트 | 15 |
| 너트 | 0 |
| 와셔 | 32 |
| 베어링 | 0 |
| 모터 | 8 |

2. 요구사항

| 구분 | 내용 |
|---|---|
| 데이터 | 상품별 재고 수량을 변수에 저장합니다. |
| 함수 정의 | check_stock 함수를 정의합니다.<br>- 매개변수 product로 상품명을 받습니다.<br>- 재고가 1 이상이면 상품명 재고 개수를 출력합니다. (예) 볼트 : 재고 있음(15개)<br>- 재고가 0이면 재고가 없다고 출력합니다. (예) 너트 : 재고 없음<br>- 등록되지 않은 상품은 등록되지 않았다고 출력합니다. (예) 나사 : 등록되지 않은 상품<br>- 반환값(return)은 사용하지 않습니다. |

정답 예시

````python
stock = {'볼트': 15, '너트': 0, '와셔': 32, '베어링': 0, '모터': 8}

def check_stock(product):
    if product not in stock:
        print(f'{product} : 등록되지 않은 상품')
    elif stock[product] >= 1:
        print(f'{product} : 재고 있음({stock[product]}개)')
    else:
        print(f'{product} : 재고 없음')

check_stock('볼트')
check_stock('너트')
check_stock('나사')
````

사용자가 상품명을 입력하는 경우

````python
product = input('상품명을 입력하세요 : ')
check_stock(product)
````

## [미션] 합격 여부 안내

학생 이름을 입력 받아 합격 여부를 리턴하는 프로그램을 작성하세요.

1. 합격자 명단 : 김민준, 이하늘, 박선우, 최유진, 정다희

2. 요구사항

| 구분 | 내용 |
|---|---|
| 데이터 | 합격자 명단을 passed_students 변수에 할당합니다. |
| 함수 정의 | 합격여부를 판별하여 리턴하는 isPassed 함수를 정의합니다.<br>- 학생 이름을 매개변수 student_name으로 받습니다.<br>- 합격자 명단에 이름이 있는지 여부를 True/False로 반환합니다. |
| 함수 호출 | 학생 이름을 input()으로 입력 받아 변수 name에 저장합니다.<br>isPassed 함수에 학생 이름을 전달하여 호출하고, 결과(합격여부)를 리턴 받아 변수 result에 저장합니다. |
| 결과 출력 | 합격 여부에 따라 다르게 메시지를 출력합니다.<br>- 김민준님, 합격을 축하합니다!<br>- 김철수님은 합격자 명단에 없습니다. |

정답 예시

````python
passed_students = ['김민준', '이하늘', '박선우', '최유진', '정다희']

def isPassed(student_name):
    return student_name in passed_students

name = input('이름을 입력하세요 : ')
result = isPassed(name)

if result:
    print(f'{name}님, 합격을 축하합니다!')
else:
    print(f'{name}님은 합격자 명단에 없습니다.')
````

## 매개변수와 인자

가변인자 : 함수를 호출할 때 전달하는 인자의 개수가 일정하지 않은 경우, 여러 개의 인자를 전달받을 수 있도록 하는 기능입니다.

**위치 가변인자 (\*args)**
- 매개변수 앞에 `*`를 붙여 정의합니다.
- 인자의 개수에 관계 없이 함수를 호출할 수 있습니다.
- 전달 받은 여러 인자는 튜플(tuple) 형태로 묶입니다.

````python
def param1(*args):
   print(args)
param1(1,2,3,4,5)
param1(1,2,3)
````

**키워드 가변인자 (\*\*kwargs)**
- 매개변수 앞에 `**`를 붙여 정의합니다.
- 이름=값 형태로 여러 인자를 전달합니다.
- 전달 받은 여러 인자는 딕셔너리(dict) 형태로 묶입니다.

````python
def param2(**kwargs):
  print(kwargs)
param2(a=1, b=2, c=3)
param2(a=1, b=2)
````

※ args, kwargs는 관례적으로 사용하는 이름입니다.

## [예제] 여러 상품의 판매금액 합계 계산

여러 상품의 판매 금액에 대한 합계를 계산하는 함수를 작성하세요.

| 구분 | 요구사항 |
|---|---|
| 함수 정의 | cal_totalprice 함수를 정의합니다.<br>\*\*args로 여러 금액을 받습니다.<br>입력받은 금액을 출력합니다.<br>금액의 합계를 리턴합니다. |
| 함수 호출 | 가변 개수의 금액을 인자로 넘깁니다.<br>(예) cal_totalprice(1000, 2000)<br>cal_totalprice(100, 300, 500, 700) |

## [연습문제] 오늘의 식단 출력

오늘의 식단을 입력 받아 출력하는 함수를 완성하세요.

````python
def print_menu(*foods):
    print("=== 오늘의 식단 ===")
    for i, food in enumerate(foods, 1):
        print(f"{i}. {food}")
    print("================")
# 함수 사용 예시
print_menu("김치찌개", "제육볶음", "미역국")
print()  # 빈 줄 출력
# 더 많은 메뉴 출력
print_menu("된장찌개", "불고기", "김치", "잡곡밥", "시금치나물")
````

## [연습문제] 학생 정보 출력

학생 정보를 입력 받아 출력하는 함수를 완성하세요.

````python
def print_student_info(**students):
    print("=== 학생 정보 ===")
    for key, value in students.items():
        print(f"{key}: {value}")
    print("================")
# 함수 사용
print_student_info(이름="홍길동", 나이=18, 학년=2, 반=3)
print()
print_student_info(이름="김철수", 학번="2024001", 전공="컴퓨터공학")
````

## 지역변수와 전역변수

함수에서 사용하는 변수는 선언된 위치와 유효 범위에 따라 지역변수와 전역변수로 구분할 수 있습니다.

**지역변수(Local Variable)**
- 함수 내부에서 정의한 변수
- 해당 함수 내부에서만 사용할 수 있습니다.

````python
def calc_total(price, count):
    subtotal = price * count
    return subtotal

total = calc_total(3000, 4)
print(total)
print(subtotal)  # NameError
````

**전역변수(Local Variable)**
- 함수 외부에서 정의한 변수
- 함수 내부와 외부에서 참조할 수 있습니다.

````python
shipping = 3000  # 전역변수

def calc_total(price, count):
    subtotal = price * count
    return subtotal + shipping

total = calc_total(3000, 4)

print(total)
print(shipping)
````

## 지역변수와 전역변수

변수의 유효 범위 : 함수 내부에서는 전역변수를 참조할 수 있지만, 함수 외부에서는 지역변수를 직접 참조할 수 없습니다.

```
┌ 전역 영역 (Global Scope) ─────────────────┐
│  shipping = 3000                          │
│  ┌ 지역 영역 (Local Scope) ─────────────┐ │
│  │ def calc_total(price, count):        │ │
│  │     subtotal = price * count         │ │
│  │     return subtotal + shipping       │ │
│  └──────────────────────────────────────┘ │
└───────────────────────────────────────────┘
```

## 지역변수와 전역변수

지역변수와 전역변수의 이름이 같다면? 지역변수가 우선 사용됩니다.

````python
shipping = 3000  # 전역변수

def calc_total(price, count):
    shipping = 0  # 지역변수
    return price * count + shipping

total = calc_total(3000, 4)

print(total)
print(shipping)
````

## 지역변수와 전역변수

global 키워드 : 함수 내부에서 전역변수의 값을 변경해야 할 때는 global 키워드를 사용합니다.

````python
order_count = 0  # 전역변수

def add_order():
    order_count += 1 # UnboundLocalError

add_order()
add_order()

print(order_count)
````

````python
order_count = 0  # 전역변수

def add_order():
    global order_count
    order_count += 1

add_order()
add_order()

print(order_count)
````
