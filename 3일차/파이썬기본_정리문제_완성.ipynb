### [연습문제 2-1] 카운트다운

`count`를 5에서 시작해서 1까지 하나씩 줄이며 출력하세요. 반복이 끝나면 "메일 발송!"을 출력하세요.

힌트: 초기식, 조건식, 변화식 세 가지를 모두 써야 무한루프에 빠지지 않습니다.

````python
count = 5

while count > 0:
    print(count)
    count -= 1

print('메일 발송!')
````

### [연습문제 2-5] 할부 상환 개월 수 구하기

남은 할부 금액은 1,000,000원이고, 매월 150,000원씩 갚습니다. 남은 금액이 0원 이하가 될 때까지 몇 개월이 걸리는지 출력하세요.

### [연습문제 2-6] break로 목표 달성한 날 찾기

일별 판매량 리스트입니다. 첫째 날부터 판매량을 누적하다가, 누적 판매량이 100개 이상이 되면 "○일째에 목표 달성 (누적 ○개)"를 출력하고 반복을 멈추세요.

힌트: `i`는 리스트의 인덱스입니다. `while i < len(sales):`로 리스트 끝까지 반복하고, 목표를 달성하면 `break`로 빠져나옵니다.

````python
sales = [18, 25, 12, 30, 22, 15]
i = 0
total = 0

while i < len(sales):
    total += sales[i]
    i += 1
    if total >= 100:
        print(f'{i}일째에 목표 달성 (누적 {total}개)')
        break
````

### [연습문제 2-7] 남은 업무가 없을 때까지 처리하기

업무 목록에 업무가 남아 있는 동안, `pop()`으로 업무를 하나씩 꺼내 "처리 완료: ○○"를 출력하세요. 업무를 모두 처리하면 "오늘 업무 끝!"을 출력하세요.

힌트: `pop()`은 마지막 요소부터 꺼냅니다. 반복 조건은 `len(tasks) > 0`입니다.

````python
tasks = ['회의록 정리', '견적서 발송', '주간보고서 작성']

while len(tasks) > 0:
    task = tasks.pop()
    print('처리 완료:', task)

print('오늘 업무 끝!')
````

### [연습문제 2-8] 횟수를 정해 두고 반복하기

로그인은 최대 3번까지 시도할 수 있습니다. 사용자가 입력한 비밀번호 대신, 시도한 비밀번호를 리스트 `tries`에 미리 적어 두었습니다. 아래 규칙대로 출력하세요.

- 시도할 때마다 "○번째 시도: 비밀번호"를 출력합니다.
- 비밀번호가 맞으면 "로그인 성공"을 출력하고 반복을 멈춥니다.
- 3번 모두 틀리면 반복이 끝난 뒤 "로그인 3회 실패. 계정이 잠겼습니다."를 출력합니다.

힌트: 반복 조건은 `attempt < max_attempt`입니다. 반복이 끝난 뒤 성공했는지 알 수 있게 `success = False`로 시작합니다.

````python
password = 'abc123'
tries = ['abc111', 'abc123', 'abc999']   # 사용자가 차례로 입력한 비밀번호
max_attempt = 3
attempt = 0
success = False

while attempt < max_attempt:
    pw = tries[attempt]
    attempt += 1
    print(f'{attempt}번째 시도: {pw}')
    if pw == password:
        success = True
        print('로그인 성공')
        break

if success == False:
    print('로그인 3회 실패. 계정이 잠겼습니다.')
````

`tries = ['abc111', 'abc000', 'abc999']`로 바꿔서 3번 모두 틀리는 경우도 확인해 보세요.

### [연습문제 2-9] while True와 break

이번 주 매출은 100만 원입니다. 매주 매출이 10%씩 늘어난다고 할 때, 매출이 200만 원 이상이 되는 데 몇 주가 걸리는지 구하세요. `while True:`로 반복하고, 조건을 만족하면 `break`로 빠져나옵니다. 매주 매출은 정수로 계산합니다.

````python
sales = 1000000
weeks = 0

while True:
    sales = int(sales * 1.1)
    weeks += 1
    if sales >= 2000000:
        break

print(f'{weeks}주 후 매출 : {sales:,}원')
````

`while True:`는 조건이 늘 참이라서 break를 만나야만 멈춥니다. break를 빠뜨리면 무한루프가 됩니다.

### [연습문제 3-7] 여러 값을 돌려주는 함수

단가와 수량을 받아서 공급가액, 부가세, 합계 금액 세 가지를 한 번에 돌려주는 함수 `calc_order`를 정의하세요. 부가세는 공급가액의 10%이고 정수로 계산합니다. 반환 값을 변수 세 개로 나눠 받아 출력하세요. 단가는 7,000원, 수량은 3개입니다.

````python
def calc_order(unit_price, quantity):
    supply_amount = unit_price * quantity
    vat = int(supply_amount * 0.1)
    total = supply_amount + vat
    return supply_amount, vat, total

supply_amount, vat, total = calc_order(7000, 3)
print('공급가액:', supply_amount)
print('부가세:', vat)
print('합계 금액:', total)
````

### [연습문제 3-8] 디폴트 매개변수

- `apply_discount(30000)` → 30000
- `apply_discount(30000, 0.2)` → 24000

````python
def apply_discount(price, rate=0):
    return int(price * (1 - rate))

print(apply_discount(30000))
print(apply_discount(30000, 0.2))
````

### [연습문제 3-10] 위치 가변인자 (*args)

금액을 몇 개든 받아서 합계를 돌려주는 함수 `sum_amounts`를 정의하세요. 금액 3개와 금액 5개로 한 번씩 호출해서 합계를 출력하세요.

````python
def sum_amounts(*amounts):
    total = 0
    for amount in amounts:
        total += amount
    return total

print(sum_amounts(12000, 5000, 3000))
print(sum_amounts(1000, 2000, 3000, 4000, 5000))
````

`*amounts`로 받은 값은 튜플입니다. for 반복구조 대신 `sum(amounts)`를 써도 됩니다.

### [연습문제 4-2] 함수 안에서 while 반복구조 쓰기

목표 금액과 월 저축액을 받아서, 목표 금액 이상이 될 때까지 걸리는 개월 수를 돌려주는 함수 `count_months`를 정의하세요. 아래 두 경우의 개월 수를 출력하세요.

- 목표 2,000,000원, 월 300,000원
- 목표 5,000,000원, 월 450,000원

````python
def count_months(target, monthly):
    savings = 0
    months = 0
    while savings < target:
        savings += monthly
        months += 1
    return months

print(count_months(2000000, 300000), '개월')
print(count_months(5000000, 450000), '개월')
````

### [쉬운 미션] 교육 수강료 계산기

교육생별 수강 시간으로 수강료를 계산하는 프로그램을 만드세요.

1. 수강 시간과 시간당 요금을 받아서 수강료를 돌려주는 함수 `calc_fee`를 정의하세요. 시간당 요금은 디폴트 매개변수로 하고, 기본값은 20,000원입니다.
2. 리스트 컴프리헨션과 `calc_fee`로 수강료 리스트 `fees`를 만드세요.
3. 수강료가 300,000원 이상이면 10% 할인한 금액(정수), 아니면 수강료 그대로 담은 리스트 `final_fees`를 리스트 컴프리헨션(if-else)으로 만드세요.
4. for 반복구조와 `range(len(names))`로 "김하늘 : 12시간 → 240,000원" 모양으로 출력하세요. 금액은 `final_fees`를 씁니다.
5. 마지막에 최종 수강료 합계를 출력하세요.

````python
names = ['김하늘', '이도윤', '박서준', '최지우']
hours = [12, 20, 8, 16]

# 1. 수강료 함수
def calc_fee(hour, rate=20000):
    return hour * rate

# 2. 수강료 리스트
fees = [calc_fee(h) for h in hours]

# 3. 할인 적용
final_fees = [int(fee * 0.9) if fee >= 300000 else fee for fee in fees]

# 4. 교육생별 출력
print('=== 수강료 안내 ===')
for i in range(len(names)):
    print(f'{names[i]} : {hours[i]}시간 → {final_fees[i]:,}원')

# 5. 합계
print(f'최종 수강료 합계 : {sum(final_fees):,}원')
````

이도윤(20시간, 400,000원)과 최지우(16시간, 320,000원)는 300,000원 이상이라 10% 할인이 적용됩니다.

### [어려운 미션] 재고 자동 발주 시뮬레이션

A4 용지 재고를 하루씩 따라가며 자동으로 발주하는 프로그램을 만드세요. 처음 재고는 40박스입니다.

1. 엑셀에서 복사한 일별 판매량 `raw_sales`는 문자열입니다. 리스트 컴프리헨션으로 정수 리스트 `sales`를 만드세요.
2. 함수 세 개를 정의하세요.
   - `sell(stock, qty)` : 재고에서 판매량을 뺀 재고를 돌려줍니다.
   - `need_order(stock, reorder_point=15)` : 재고가 기준(기본값 15박스)보다 적으면 True를 돌려줍니다.
   - `place_order(stock, amount=30)` : 재고에 발주량(기본값 30박스)을 더한 재고를 돌려줍니다. 이 함수를 부를 때마다 전역변수 `order_count`를 1씩 늘립니다.
3. while 반복구조로 첫째 날부터 마지막 날까지 하루씩 처리하세요. 하루마다 아래 순서로 처리합니다.
   - 판매량만큼 재고를 줄이고 "○일째 : 판매 ○박스 → 재고 ○박스"를 출력합니다.
   - 발주가 필요하면 발주하고 "   발주! 재고 ○박스"를 출력합니다.
   - 그날의 최종 재고를 리스트 `stock_log`에 추가합니다.
4. 반복이 끝나면 아래 내용을 출력하세요.
   - 발주 횟수
   - 총 판매량과 하루 평균 판매량(`//`로 정수 몫만)
   - 최종 재고가 20박스 미만이었던 날짜 리스트 (리스트 컴프리헨션)

힌트

- 3번은 17 반복구조 while의 "1부터 10까지 출력"처럼 `day = 0`, `while day < len(sales):`, `day += 1`로 반복합니다.
- 4번의 날짜 리스트는 `[i + 1 for i in range(len(stock_log)) if stock_log[i] < 20]`처럼 인덱스로 고릅니다.

````python
raw_sales = ['7', '9', '6', '12', '8', '10', '5']
stock = 40
order_count = 0

# 1. 판매량을 정수로 바꾸기
sales = [int(s) for s in raw_sales]

# 2. 함수 정의
def sell(stock, qty):
    return stock - qty

def need_order(stock, reorder_point=15):
    return stock < reorder_point

def place_order(stock, amount=30):
    global order_count
    order_count += 1
    return stock + amount

# 3. 하루씩 처리
stock_log = []
day = 0
while day < len(sales):
    stock = sell(stock, sales[day])
    print(f'{day + 1}일째 : 판매 {sales[day]}박스 → 재고 {stock}박스')
    if need_order(stock):
        stock = place_order(stock)
        print(f'   발주! 재고 {stock}박스')
    stock_log.append(stock)
    day += 1

# 4. 결과 요약
total_sold = sum(sales)
low_days = [i + 1 for i in range(len(stock_log)) if stock_log[i] < 20]

print()
print('=== 결과 ===')
print('발주 횟수 :', order_count, '번')
print('총 판매량 :', total_sold, '박스')
print('하루 평균 판매량 :', total_sold // len(sales), '박스')
print('재고 20박스 미만인 날 :', low_days)
````

`need_order(stock, 20)`이나 `place_order(stock, 50)`처럼 기준과 발주량을 바꿔 보세요. 디폴트 매개변수 덕분에 함수를 고치지 않고 조건만 바꿀 수 있습니다.
