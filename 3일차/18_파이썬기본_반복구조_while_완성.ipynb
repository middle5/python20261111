# 반복구조

## while 반복구조

조건의 결과가 True인 동안 반복합니다.

```
초기식
while 조건식:
        실행할 코드블록
        변화식        ← 무한루프 방지
```

동작 순서: 시작 → 조건식 검사 → True이면 반복 코드 실행 → 조건 재검사 → … → False이면 종료

## while 반복구조

카운트다운

- `count = 3` : 초기식 (반복에 사용할 변수 초기화)
- `count > 0` : 반복을 계속할 조건
- `count = count-1` : 변화식

````python
count = 3
while count > 0:
  print(count)
  count = count-1
````

## [예제1] 목표 금액에 도달할 때까지 저축하기

매일 20만원씩 저축할 때, 저축액이 100만원 이상이 될 때까지 반복한다.

````python
savings = 0
month = 0

while savings<1000000:
  savings += 200000
  month += 1

print(f"저축 기간 : {month}개월")
print(f"저축 금액 : {savings:,}원")
````

## [연습문제] 재고관리

초기 재고 100개에서 주문 수량 15개씩 반복 출고하고, 남은 재고가 15개 미만이 되면 종료합니다. 총 출고 횟수와 남은 재고를 출력하세요.

````python
stock = 100
order_quantity = 15
count = 0

while stock>=15:
  count += 1
  stock -= 15

print(f"총 출고 횟수 : {count}")
print(f"남은 재고량 : {stock}")
````

## break를 이용한 반복문 종료

조건식이 True인 상태에서도 break를 만나면 반복문을 종료합니다.

```
while 조건식:
    실행할 코드
    if 종료조건:
        break
```

## [예제1] 특정 숫자가 나오면 반복 종료하기

1부터 10까지 출력

````python
num = 1

while num <= 10:
  print(num)
  num += 1
````

1부터 10까지 출력하다가 5를 만나면 반복 종료

````python
num = 1

while num <= 10:
  if num == 5:
    break
  print(num)
  num += 1
````

## 무한루프와 반복 종료

◾ 무한루프와 break

무한반복 하다가 특정 조건을 만나면 반복을 종료합니다.

````python
while True:
  user_input = input("아무거나 입력하세요. 중단하려면 'q'를 입력하세요.")
  if user_input=='q':
    break
````


## [미션] 숫자 맞추기 게임

- 1~100 사이의 랜덤 정수로 정답을 만든다.
- 사용자로부터 반복하여 숫자를 입력 받는다.
- 정답이 사용자가 입력한 숫자보다 크면 'up!'을 출력한다.
- 정답이 사용자가 입력한 숫자보다 작으면 'down!'을 출력한다.
- 사용자가 정답을 입력하면 '정답!' 출력하고 반복을 종료한다.

````python
import random
# 1부터 100 사이의 랜덤 숫자 생성
secret_number = random.randint(1, 50)
print("=== 숫자 맞추기 게임 ===")
print("1부터 100 사이의 숫자를 맞춰보세요!")
print()
while True:
    guess = int(input(f"숫자를 입력하세요: "))

    # 입력한 숫자 검증
    if guess < secret_number:
        print("그보다 큰 숫자입니다!")
    elif guess > secret_number:
        print("그보다 작은 숫자입니다!")
    else:
        print("정답입니다!")
        break
````

## [미션] 숫자 맞추기 게임

````python
import random
# 1부터 100 사이의 랜덤 숫자 생성
secret_number = random.randint(1, 50)
attempt = 0
max_attempt = 10
print("=== 숫자 맞추기 게임 ===")
print("1부터 100 사이의 숫자를 맞춰보세요!")
print()
while attempt <= max_attempt:
    # 남은 기회 표시
    guess = int(input(f"[남은기회:{max_attempt-attempt}번]숫자를 입력하세요: "))
    attempt += 1
    # 입력한 숫자 검증
    if guess < secret_number:
        print("그보다 큰 숫자입니다!")
    elif guess > secret_number:
        print("그보다 작은 숫자입니다!")
    else:
        print(f"축하합니다! {attempt}번 만에 정답을 맞추셨습니다!")
        break
if attempt > max_attempt:
  print(f"게임오버! 정답은 {secret_number}입니다.")
````
